```python
def mesh_deform_loss(arg_deformed_template, arg_trg_mesh, n_samples_trg=5000, use_temp_verts=False, n_samples_temp=5000,
                     chamfer_on=True, edge_on=True, normal_on=True, laplacian_on=True, ptedge_on=False, ptface_on=False,
                     given_trg_samples=False, double_chamfer=True, em_weights=False, knn_k=1, sigma_sq_inv=None, rec_runtime=False):
    # arg_deformed_template: structures.Meshes
    # use_temp_verts: use the vertices of template instead of sampling from template meshes (assume templates in a batch all same number of vertices)
    if rec_runtime:
        runtime_split = np.zeros(8)
        runtime_split_names = ['trg sampling', 'model sampling', 'knn finding and distance', \
                               'em for chamfer', 'edge loss', 'normal loss', 'lap loss', 'other losses']
        start_time = timeit.default_timer()
    if given_trg_samples:
        assert(not ptedge_on and not ptface_on) # require arg_trg_mesh to be Meshes
        # overload the arg_trg_mesh to be [bs,n,3] for points input or Pointclouds object for clouds of different len
        sample_trg = arg_trg_mesh
    else:
        sample_trg = pytorch3d.ops.sample_points_from_meshes(arg_trg_mesh, n_samples_trg) # (bs,n_samples=5000,3)
    if rec_runtime:
        runtime_split[0] = timeit.default_timer() - start_time
        start_time = timeit.default_timer()
    if use_temp_verts:
        nmb_verts_temp = arg_deformed_template.num_verts_per_mesh()
        assert(nmb_verts_temp.min() == nmb_verts_temp.max()) # all same number of vertices
        sample_template = arg_deformed_template.verts_padded()
    else:
        sample_template = pytorch3d.ops.sample_points_from_meshes(arg_deformed_template, n_samples_temp) # (bs,n_samples,3)
    if rec_runtime:
        runtime_split[1] = timeit.default_timer() - start_time
        start_time = timeit.default_timer()
    if chamfer_on:
        if double_chamfer: # chamfer is double sided between two point clouds (see src code). Assume trg or template point cloud all same len. (trg and template can have different lens)
            loss_chamfer_out, _ = pytorch3d.loss.chamfer_distance(sample_trg, sample_template) # (bs,n,3) to (1) avg.
            if rec_runtime:
                runtime_split[3] = timeit.default_timer() - start_time
                start_time = timeit.default_timer()
        else: # one sided assumes trg point clouds have different lengths
            if not em_weights:
                knn_k = 1
            # the chamfer loss uses default mean point reduction and mean batch reduction
            knn_output = pytorch3d.ops.knn_points(sample_trg.points_padded(), sample_template,
                                                  lengths1=sample_trg.num_points_per_cloud(), lengths2=None,
                                                  K=knn_k, return_nn=False, return_sorted=False) # default L2 norm output squared dist
            if rec_runtime:
                runtime_split[2] = timeit.default_timer() - start_time
                start_time = timeit.default_timer()
            if em_weights:
                # sigma_sq_inv: [bs,n] with one/zero padding
                post_prob = association_prob(knn_k) * gauss_prob(knn_output.dists, sigma_sq_inv) # [bs,n,k]
                post_prob = post_prob / torch.sum(post_prob, dim=-1, keepdim=True) # [bs,n,k]
                weights = (post_prob * sigma_sq_inv.unsqueeze(-1)).detach() # [bs,n,k]
                chamfer_loss = (knn_output.dists * weights).sum((-1,-2)) # [bs] the sum over k is weighted by the weights so not mean over k
                chamfer_loss /= sample_trg.num_points_per_cloud().clamp(min=1).to(torch.float32)
                loss_chamfer_out = chamfer_loss.mean(0)
            else:
                chamfer_loss = knn_output.dists[...,0].sum(-1) # [bs]
                chamfer_loss /= sample_trg.num_points_per_cloud().clamp(min=1).to(torch.float32)
                loss_chamfer_out = chamfer_loss.mean(0)
            if rec_runtime:
                runtime_split[3] = timeit.default_timer() - start_time
                start_time = timeit.default_timer()
                
    else:
        loss_chamfer_out = 0
    if edge_on:
        loss_edge_out = pytorch3d.loss.mesh_edge_loss(arg_deformed_template) # each mesh in a batch contributes equally (weighted by 1/n_edges) avg. across b.
    else:
        loss_edge_out = 0
    if rec_runtime:
        runtime_split[4] = timeit.default_timer() - start_time
        start_time = timeit.default_timer()
    if normal_on:
        loss_normal_out = pytorch3d.loss.mesh_normal_consistency(arg_deformed_template) # avg. across the batch
    else:
        loss_normal_out = 0
    if rec_runtime:
        runtime_split[5] = timeit.default_timer() - start_time
        start_time = timeit.default_timer()
    if laplacian_on:
        loss_laplacian_out = pytorch3d.loss.mesh_laplacian_smoothing(arg_deformed_template, method='uniform') # avg. across the batch (make nearby verts close)
    else:
        loss_laplacian_out = 0
    if rec_runtime:
        runtime_split[6] = timeit.default_timer() - start_time
        start_time = timeit.default_timer()
    if ptedge_on or ptface_on:
        verts_template = arg_deformed_template.verts_list() # list[(n_verts,3)] of len bs
        vert_pts_template = pytorch3d.structures.Pointclouds(verts_template) # try with the sampled pts (bs,ptc)
    if ptedge_on:
        loss_ptedge_out = pytorch3d.loss.point_mesh_edge_distance(arg_trg_mesh, vert_pts_template) # avg. across pairs of ptc and mesh in a batch
    else:
        loss_ptedge_out = 0
    if ptface_on:
        loss_ptface_out = pytorch3d.loss.point_mesh_face_distance(arg_trg_mesh, vert_pts_template) # avg. across pairs of ptc and mesh in a batch
    else:
        loss_ptface_out = 0
    if rec_runtime:
        runtime_split[7] = timeit.default_timer() - start_time
    if rec_runtime:
        return loss_chamfer_out, loss_edge_out, loss_normal_out, loss_laplacian_out, loss_ptedge_out, loss_ptface_out, (runtime_split, runtime_split_names)
    else:
        return loss_chamfer_out, loss_edge_out, loss_normal_out, loss_laplacian_out, loss_ptedge_out, loss_ptface_out
```

```python
# cubic template
# template_mesh_vox_np = np.ones((1,32,32,32)) # 32 with 1/2 scale
# template_mesh_vox = torch.from_numpy(template_mesh_vox_np).to(device)
# template_mesh = pytorch3d.ops.cubify(template_mesh_vox, 0.0, device=torch.device(device), align='center')
# template_mesh.scale_verts_(1.0/2)

# spherical template
template_mesh = pytorch3d.utils.ico_sphere(4, device) # n of iterations to subdivide the mesh faces
pytorch3d.io.IO().save_mesh(template_mesh, 'mesh_template.ply')
print(template_mesh.num_verts_per_mesh())
```

```python
# code following https://pytorch3d.org/tutorials/deform_source_mesh_to_target_mesh
deform_verts = torch.full(template_mesh.verts_packed().shape, 0.0, device=device, requires_grad=True) # (sum(n_verts),3)
# optimizer = torch.optim.Adam([deform_verts], lr=lrate)
optimizer = torch.optim.SGD([deform_verts], lr=1.0, momentum=0.9)
n_iters = 2000
loop = tqdm(range(n_iters))
chamfer_losses, laplacian_losses, edge_losses, normal_losses = [], [], [], []
ptedge_losses, ptface_losses = [], []
```

```python
for i in loop:
    optimizer.zero_grad()  
    deformed_template = template_mesh.offset_verts(deform_verts) # new meshes object (bs,mesh)
    if save_intermediate_meshes and (i%save_frequency == 0): # not i+1 since deformed_template here uses last estimate
        pytorch3d.io.IO().save_mesh(deformed_template, 'mesh_template_deformed_{:08d}.ply'.format(i))
    [loss_chamfer, loss_edge, loss_normal, loss_laplacian, loss_ptedge, loss_ptface] \
        = mesh_deform_loss(deformed_template, one_mesh,
                           chamfer_on=wt_chamfer, edge_on=wt_edge, normal_on=wt_normal,
                           laplacian_on=wt_laplacian, ptedge_on=wt_ptedge, ptface_on=wt_ptface)
    total_loss = wt_chamfer * loss_chamfer + wt_edge * loss_edge + \
                 wt_normal * loss_normal + wt_laplacian * loss_laplacian + \
                 wt_ptedge * loss_ptedge + wt_ptface * loss_ptface
    loop.set_description('total_loss = %.6f' % total_loss)
    if wt_chamfer:
        chamfer_losses.append(float(loss_chamfer.detach().cpu()))
    else:
        chamfer_losses.append(0)
    if wt_edge:
        edge_losses.append(float(loss_edge.detach().cpu()))
    else:
        edge_losses.append(0)
    if wt_normal:
        normal_losses.append(float(loss_normal.detach().cpu()))
    else:
        normal_losses.append(0)
    if wt_laplacian:
        laplacian_losses.append(float(loss_laplacian.detach().cpu()))
    else:
        laplacian_losses.append(0)
    if wt_ptedge:
        ptedge_losses.append(float(loss_ptedge.detach().cpu()))
    else:
        ptedge_losses.append(0)
    if wt_ptface:
        ptface_losses.append(float(loss_ptface.detach().cpu()))
    else:
        ptface_losses.append(0)
    total_loss.backward()
    optimizer.step()
fig = plt.figure(figsize=(13, 5))
ax = fig.gca()
ax.plot(chamfer_losses, label="chamfer loss")
ax.plot(edge_losses, label="edge loss")
ax.plot(normal_losses, label="normal loss")
ax.plot(laplacian_losses, label="laplacian loss")
ax.plot(ptedge_losses, label="point-mesh edge loss")
ax.plot(ptface_losses, label="point-mesh face loss")
ax.legend(fontsize="16")
ax.set_xlabel("Iteration", fontsize="16")
ax.set_ylabel("Loss", fontsize="16")
ax.set_title("Loss vs iterations", fontsize="16")

deformed_template = template_mesh.offset_verts(deform_verts)
pytorch3d.io.IO().save_mesh(deformed_template, 'mesh_template_deformed.ply')
```

**Summary of Good Settings**<br>
=> *Chair, Bottle, Bowl*<br>
wt_chamfer = 1.0<br>
wt_edge = 1.0<br>
wt_normal = 0.01<br>
wt_laplacian = 0.1<br>
wt_ptedge = 0.0<br>
wt_ptface = 0.0<br>
n_iters = 2000<br>
ico_sphere = 4<br>
lr = 1.0, momentum = 0.9<br>
=> *Mug*<br>
Normal settings will have artifacts (pits and extrusions) due to the deep cavity.<br>
Solution: wt_edge=3.0, wt_laplacian=0.01<br>
=> *Laptop*<br>
ico_sphere = 3, n_iters = 500<br>
ico_sphere = 4, n_iters = 2000<br>
=> *Camera*<br>
Normal settings will have holes.<br>
Solutions: (1) ico_sphere = 3. (2) wt_laplacian = 0.5 - 0.8. (3) wt_edge = 3.0 - 5.0.<br>
Choose wt_edge = 5.0<br>
=> *Can*<br>
Normal settings will have large holes in the bottom or top randomly.<br>
Solutions: (1) n_iters = 1000. Surface will retain some irregularities. (2) n_iters = 3000. Perfect cylinder.<br>
(3) wt_laplacian >= 0.2 < 0.8. Perfect cylinder. (4) wt_edge >= 3.0. Perfect Cylinder.<br>
Choose n_iters = 1000 to have the top flange and wt_laplacian = 0.3 to avoid holes.

# Batched Objects

```python
# source code for pytorch3d.datasets.collate_batched_meshes
# https://pytorch3d.readthedocs.io/en/v0.6.0/_modules/pytorch3d/datasets/utils.html#collate_batched_meshes
# comment out the two lines for texture loading to allow not loading textures
from typing import Dict, List
# from pytorch3d.renderer.mesh import TexturesAtlas
from pytorch3d.structures import Meshes
def collate_batched_meshes_notexture(batch: List[Dict]):
    if batch is None or len(batch) == 0:
        return None
    collated_dict = {}
    for k in batch[0].keys():
        collated_dict[k] = [d[k] for d in batch]

    collated_dict["mesh"] = None
    if {"verts", "faces"}.issubset(collated_dict.keys()):

        textures = None
#         if "textures" in collated_dict:
#             textures = TexturesAtlas(atlas=collated_dict["textures"])

        collated_dict["mesh"] = Meshes(
            verts=collated_dict["verts"],
            faces=collated_dict["faces"],
            textures=textures,
        )

    return collated_dict
```

```python
excluded_instances = {
    '02876657': ['3ba7dd61736e7a96270c0e719fe4ed97', # jug
                 '8d861e4406a9d3258534bca5562e21be', # jug
                 '5ca994bd512be485db3bac7ad77ae88e', # nozzle
                 '547fa0085800c5c3846564a8a219239b', # jug
                 '8e20da9c3f63ac2ce902acf138e668e2', # jug
                 '6c56bd467b8e6b7912c4d7e462c15a1a', # jug
                 'c5032e4be091aabb36fe7a88df793982', # art
                 '42f85b0eb5e9fd508f9c4ecc067067e9', # jug
                 'e23f62bb4794ee6a7fdd0518ed16e820', # nozzle
                 'd45bf1487b41d2f630612f5c0ef21eb8', # sprayer
                 '9b9a4bb5550f00ea586350d6e78ecc7',  # sprayer
                 'eded10bf44a2571a911cff0cb398f845', # jug
                 'cbc1cbc9cf65e9c2fd1d6016d24cc8d',  # sprayer
                 'ea200708a0779090eab2bd7322386ffd', # canteen
                 'd2e7e725aa6b39f0d333084c1357713e', # sprayer
                 '4a1fe3495565001d211b8bef58aae14f', # jug
                 '98f2bdf417a15ceaa5d6dc096343ac63', # wine plug
                 'c82bbd1b1a929c7ce8ed8cf0a077ddf7', # jug
                 '9dd18bbd88df283a6b36e3e6a5b248ba', # gas tank
                 'f1ae9b2eb551ba466dc523ea4bf73f1c', # vase                 
                 '41ab1142420916e374bb1e92fd01dc00', # jug
                 '77ad88568cb55237e01d7dc978602402', # jug
                 'be102516d65bcf23ff59f7e635b49cca', # jug
                 '64ad00b1757205a3f388b3201b32e1d0', # art
                 '3bf6160225cd15453a762233fffc49ea', # handle bottle
                 '5872e807edaf985878fe949fc1419876', # jug
                 '9a70dd116325eb2df6b14e57dce52ee1', # jug
                 '645cf2b7f442865ebf8d8d6022c8e3a',  # jug
                 '61847ebb20b88ce2a62e98cb10d773b0', # nozzle
                 'f6fa8cc6b08aaabb2aafb83519a0c8dc', # jug
                 '9fe7e6a7bf8ca964efad53eb3f0b36fa', # vase
                 '328ba04ff5119c74763cd81c1b401a16', # jug
                 'b2507913d631e8b8bcfad79fc308fa6d', # vase
                 '35ad544a3a4b25b122de9a89c84a71b7', # jug
                 '44dcea00fb1923051a4cb1c0c7bb0654', # nozzle
                 'adace8a468fd575037939c05190e283f', # nozzle
                 '495dea92a04b43e6be9fabb33de62010', # jug
                 '5ac30e855ea3faad8884a069d9619eaf', # jug
                 '347f9f4a04cc00841fba588fa05a3c38', # jug
                 '72a9f3c57e40416f8884a069d9619eaf', # jug
                 'f03f9704001d2055ab8a56962d772d13', # jug
                 '2cc108080535bc51e7a14f6d159dcd47', # twisted
                 '2efb9eef0ed832a89769b3a50399d097', # nozzle
                 '1f1e747bfce16fe0589ac2c115a083f1', # nozzle
                 'f47cbefc9aa5b6a918431871c8e05789', # nozzle
                 '807f781ec8310baaa2e58dbee8c64797', # handle bottle
                 '21dbbd17ae4c3fce6a90f735365ac29e', # jug
                 'c3bc0cd2648e76b9187964c89b3399f1', # big nozzle
                 'ac79fd7f10f274aba2c59a4d90e63212'  # jug
    ], # bottle (exclude asymmetric)
    '02880940': ['d28f7a7a8fbc5fc925b5a13384fa548b', # wine glass
                 '18529eba21e4be8b5cc4957a8e7226be', # double bowl
                 'd1addad5931dd337713f2e93cbeac35d', # wine glass
                 'e16e27d3670853b12d4e6b2984840098', # hollow wall
                 'a9ba34614bfd8ca9938afc5c0b5b182',  # grail
                 'e4c871d1d5e3c49844b2fa2cac0778f5', # fancy
                 'd98455f19960f99ed684faddec3c0090', # hanging bowl
                 'da5f13f4048dbd72fcb8d8c6d4df8143', # hanging bowl
                 'bd2ba805bf1739cdedd852e9640b8d4',  # double bowl
                 '1a0a2715462499fbf9029695a3277412', # eccentric
                 '446583d7bf857dced5cb6d178687b980', # hollow wall
                 'faa200741fa93abb47ec7417da5d353d', # wavy wall
                 '5019f979a6a360963a5e6305a3a7adee', # irregular
                 '4cf18594e8851a3d3a5e6305a3a7adee', # hollow wall
                 '2ffe06ee50ec1420adbe0813683fcfd0', # irregular
                 'e3095ecacc36080cb398a1cfd1079875', # legged bowl
                 '24907349b2323824664b3b9b23ddfcbc', # hanging bowl
                 '46d014209027ec188e74a3351a4d8b3a', # bottom outlet
    ], # bowl
    '02942699': ['89f0e0da4747aad1172ac9cfeff21994', # video tape
                 '3d81cebaa1226e9329fdbaf17f41d872', # surveillance
                 'fe669947912103aede650492e45fb14f', # surveillance
                 'a9408583f2c4d6acad8a06dbee1d115',  # lens only
                 'ff74c4d2e710df3401a67448bf8fe08',  # surveillance
                 'd1a3482c576f8c50592ecd319dfd8c5d', # surveillance
                 '7e677756898b40dc39513d756da531d0', # lens only
                 '97cd28c085e3754f22c69c86438afd28', # video recorder
                 '1ab3abb5c090d9b68e940c4e64a94e1e', # video recorder
                 '1cc93f96ad5e16a85d3f270c1c35f1c7', # video recorder
                 '7ef29f8a7a132c46e0afa7d1aded497',  # webcam
                 '7077395b60bf4aeb3cb44973ec1ffcf8', # webcam
                 '416ce9adf7ad40f4959c1c3d740c4f1',  # cap off
                 '2c0b4e318766e01723cd81bf29b64a1',  # video recorder
                 '4cd861035c740db5a33f3afcb8763f26', # video recorder
                 '1298634053ad50d36d07c55cf995503e', # webcam
                 'ce40b134b11e8c822bbc2c380e91dfe2', # stand video recorder
                 '290abe056b205c08240c46d333a693f',  # webcam
                 '550aea46c75351a387cfe978d99ba05d', # webcam
                 '39419e462f08dcbdc98cccf0d0f53d7',  # webcam
                 '87b8cec4d55b5f2d75d556067d060edf', # surveillance
                 '68f66f5ab594fd3cc2890daf3a9f7413', # surveillance
                 '147183af1ba4e97b8a94168388287ad5', # surveillance
                 'b42c73b391e14cb16f05a1f780f1cef',  # webcam
                 '6ca77b19e3f52c5031c1d3ccd72d7161', # no label = webcam
                 '3b56239e45828d2bb4521a835b1946c8',
                 'd9bb9c5a48c3afbfb84553e864d84802',
                 'ee58b922bd93d01be4f112f1b3124b84', # video tape
                 'cef23409b1a030d516fe5320f9715eef',
                 '2153bc743019671ae60635d9e388f801',
                 'fd058128095abf343ce579773d34f5bf',
                 '4700873107186c6e2203f435e9e6785', 
                 '45b01c604426a0a9c5c10e0ebb47766c',
                 'fc83047701a0e21b901ee6249a8d9beb',
                 '2fc6168fba3ef6953ada8db96f6d95a3',
                 '9db4b2c19c858a36eb34db531a289b8e',
                 'eb86c8c2a20066d0fb1468f5fc754e02',
                 'd680d61f934eaa163b211460f022e3d9',
                 '9cdaf68ed1e1daba9c21adf7cd249be9',
                 '73b02fe7b4b2e45f74cd541b34c6124',
                 'b27815a2bde54ad3ab3dfa44f5fab01',  # video recorder
                 'e3dc17dbde3087491a722bdf095986a4', # webcam
                 'c802792f388650428341191174307890',
                 'e57aa404a000df88d5d4532c6bb4bd2b', # webcam
                 'c3e6564fe7c8157ecedd967f62b864ab', # stand camera
                 '60923e8a6c785a8755a834a7aafb0236',
                 '22217d5660444eeeca93934e5f39869', 
                 '6ed69b00b4632b6e07718ee10b83e10',
                 'b22e56fdf18e2eb8968b65a7871de463'
    ], # camera
    '02946921': ['ac0eafed7aa1719b297d1fadd26a7035', # rectangle
                 'a8d4a3b32149d3ae4f0b8af4571b1440', # deformed can
                 '91a524cc9c9be4872999f92861cdea7a', # rectangle
                 '70172e6afe6aff7847f90c1ac631b97f', # no bottom
    ], # can
    '03642806': ['6ae14173f5bface824a85bcf396c330',  # wired laptop
                 'e555880523dc11425b440d1b0fe8ea5c', # wired laptop
                 'dc264eab83ca12b3da4c0d8596dff972', # wired laptop
                 'f9f484e4364663d61bb5a64b9a0f552b', # laptop accessories
                 '5085268208bb23ba7b037ded105df073', # disk drive extended
                 'b211cfb105e9f97e6436916a86a90ed7', # laptop accessories
                 '7bdac195f81588ff743bd44138e7d2e5', # disk drive extended
                 '7da524a31c9cfe22ea8725c845547c21', # usb drive
                 'c912c50b812268e8fc9e3f0d993c8121', # disk drive extended
                 'de9a39e8ef9c313bbba0ffaa383992dc', # disk drive extended
                 '471005df47f0b4a6625cbbe81bda8c5c', # usb drive
                 '4504a4d244d05ddbf5f79806bd65844f', # wrong orientation
                 '3b2db36aaa2546b99c7c402f274622c',  # tray
                 '8fefaa8179404c4ae165aa9cd54a0a29', # tablet
                 'ccf62b71e7a9aedfb162fa62b36ac9a4', # disk drive extended
                 '9fa387d7f442b96e75e60c00fabe2744', # wrong orientation
                 '6c1cba2e5d2ebb552209327b12ad3fd',  # disk drive extended
                 '97df7577c2690ab82290f0f0820b73a5', # disk drive extended
                 'caa4afd404f24d21275c1147a304ed86', # wrong orientation
                 '67e4cc8e13aa863cc418dd4d8b66af83', # wrong orientation
    ], # laptop
    '03797390': ['5b0c679eb8a2156c4314179664d18101', # spoon
                 '85d5e7be548357baee0fa456543b8166', # saucer
                 '659192a6ba300f1f4293529704725d98', # saucer
                 'b9be7cfe653740eb7633a2dd89cec754', # donut
                 'ef24c302911bcde6ea6ff2182dd34668', # saucer
                 'f42a9784d165ad2f5e723252788c3d6e', # spoon
                 '1dd8290a154f4b1534a8988fdcee4fde', # content
                 'bf2b5e941b43d030138af902bc222a59', # legged
                 '4f9f31db3c3873692a6f53dd95fd4468', # spoon
                 '24b17537bce40695b3207096ecd79542', # content
                 '6d2657c640e97c4dd4c0c1a5a5d9a6b8', # fancy
                 '83b41d719ea5af3f4dcd1df0d0a62a93', # content
                 '32e197b8118b3ff6a4bd4f46ba404890', # spoon
                 '3a7439cfaa9af51faf1af397e14a566d', # fancy
                 'b98fa11a567f644344b25d683fe71de',  # fancy
                 '642eb7c42ebedabd223d193f5a188983', # donut
                 '649a51c711dc7f3b32e150233fdd42e9', # fancy
                 '1038e4eac0e18dcce02ae6d2a21d494a', # donut
                 '127944b6dabee1c9e20e92c5b8147e4a', # straw
                 '8f550770386a6cf82f23d8349a133d2b', # saucer
                 '604fcae9d93201d9d7f470ee20dce9e0', # content
                 '1c3fccb84f1eeb97a3d0a41d6c77ec7c', # spoon
                 '5310945bb21d74a41fabf3cbd0fc77bc', # capped
                 '2d10421716b16580e45ef4135c266a12', # square
                 '6661c0b9b9b8450c4ee002d643e7b29e', # saucer
    ] # mug
}
for syn_cls in excluded_instances.keys():
    print(syn_cls)
    print('number excluded: ', len(excluded_instances[syn_cls]))

def exclude_weird_shapes(meshes_set):
    # input is a batch from iterating shapenet_loader
    feed_in_mdid = np.ones(len(meshes_set['model_id']), dtype=bool)
    feed_in_idx = []
    count_excluded = 0
    for mdid in range(len(meshes_set['model_id'])):
        if meshes_set['synset_id'][mdid] not in excluded_instances.keys():
            continue
        if meshes_set['model_id'][mdid] in excluded_instances[meshes_set['synset_id'][mdid]]:
            feed_in_mdid[mdid] = False
            count_excluded += 1
        else:
            feed_in_idx.append(mdid)
    if count_excluded == 0:
        return meshes_set
    meshes_set['synset_id'] = np.array(meshes_set['synset_id'])[feed_in_mdid]
    meshes_set['model_id'] = np.array(meshes_set['model_id'])[feed_in_mdid]
    meshes_set['verts'] = [meshes_set['verts'][inidx] for inidx in feed_in_idx]
    meshes_set['faces'] = [meshes_set['faces'][inidx] for inidx in feed_in_idx]
    meshes_set['label'] = [meshes_set['label'][inidx] for inidx in feed_in_idx]
    meshes_set['textures'] = None
    if 'mesh' in meshes_set.keys():
        meshes_set['mesh'] = pytorch3d.structures.Meshes(verts=meshes_set['verts'], faces=meshes_set['faces'])
    return meshes_set
```

# PCA Basis

```python
import numpy as np
import os
import glob
import torch
import pytorch3d.structures
import pytorch3d.io
from sklearn.decomposition import PCA
import timeit
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt

# user input
std_meshes_dir = '/home/yihao/Research/ShapeBasis/Records/20230723_001622_obj03797390_wtedge3_wtlap0.01'
K = 7
```

```python
# collect and prepare the mesh verts data
std_mesh_files = glob.glob(os.path.join(std_meshes_dir, '*_std.ply'))
all_flat_verts = []
obj_id_pca_idx_map = {}
count_objs = 0
for f in std_mesh_files:
    verts, faces = pytorch3d.io.load_ply(f)
    verts_np = verts.numpy()
    verts_flat = np.ndarray.flatten(verts_np)
    all_flat_verts.append(verts_flat)
    # find a mapping between pca data points idx to instance random id sequence
    obj_id_pca_idx_map[f.split('/')[-1].split('_')[0]] = count_objs
    count_objs += 1
all_flat_verts = np.array(all_flat_verts)
assert(count_objs == all_flat_verts.shape[0])
```

```python
# run pca
pca = PCA(n_components=K)
start = timeit.default_timer()
pca_basis_params = pca.fit_transform(all_flat_verts)
stop = timeit.default_timer()
print('Total Duration (sec): ', stop - start)
print(pca_basis_params.shape)
plt.figure()
plt.plot(pca.explained_variance_) # incremental amount of variance explained
plt.grid()
plt.figure()
plt.plot(pca.explained_variance_ratio_ * 100) # incremental percentage of variance explained
plt.grid()
plt.figure()
plt.plot(pca.singular_values_)
plt.grid()
```

```python
# save the pca results
import pickle

# file_name = '/home/yihao/Research/ShapeBasis/Records/20230204_pcaBasis_256.pickle' # [DONE] change
file_dir = os.path.join(std_meshes_dir, 'pca_indexed')
if not os.path.exists(file_dir):
    os.mkdir(file_dir)
file_name = os.path.join(file_dir, std_meshes_dir.split('/')[-1] + '_pca{:04d}.pickle'.format(K))
with open(file_name, 'wb') as f:
    pickle.dump({'pca_mean': pca.mean_,
                 'pca_bases': pca.components_,
                 'pca_params': pca_basis_params,
                 'faces': faces.detach().cpu().numpy(),
                 'obj_id_pca_idx_map': obj_id_pca_idx_map}, f)
```

```python
# save projected/compressed meshes
invpca_all_flat_verts = pca.inverse_transform(pca_basis_params)
loop = tqdm(range(len(std_mesh_files)))
for i in loop:
    verts_proj_np = np.reshape(invpca_all_flat_verts[i,:], (-1,3))
    verts_proj = torch.tensor(verts_proj_np)
    mesh_proj = pytorch3d.structures.Meshes(verts=[verts_proj], faces=[faces])
    mesh_save_path = std_mesh_files[i][:-8]+'_pca{:04d}.ply'.format(K)
    pytorch3d.io.IO().save_mesh(mesh_proj, mesh_save_path)
```

# Evaluation of Basis Reconstruction

I evaluate the shape reconstruction using Chamfer distance by sampling 5000 points on each mesh surface. I compare ground-truth vs. ground-truth (to see  how faithful the evaluation metric is), ground-truth vs. standardized mesh, and ground-truth vs. pca transformed mesh.

```python
import os
import glob
from tqdm.notebook import tqdm
import pickle
import warnings
import torch
import pytorch3d.io
import pytorch3d.ops
import pytorch3d.loss
import numpy as np

# user input
meshes_dir = '/home/yihao/Research/ShapeBasis/Records/20230722_181031_obj02946921_niters1000_wtlap0.3'
K = 7
chamfer_n_pts = 5000
eval_designated = True # designate a name to evaluate o.w. default name
designated_name = '_vaeNonlinear0032_latsz32_regwt00001_epc14994.ply' # designate a name to evaluate o.w. default name

# random seed
torch.manual_seed(667)
torch.cuda.manual_seed(667)
torch.cuda.manual_seed_all(667)

# device
twingpu = False
if twingpu:
    device = "cuda:0,1" if torch.cuda.is_available() else "cpu"
else:
    device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
```

```python
chamfer_trgtrg_all = []
chamfer_trgstd_all = []
chamfer_trgpca_all = []
trg_mesh_files = glob.glob(os.path.join(meshes_dir, '*_trg.ply'))
mesh_iter = tqdm(range(len(trg_mesh_files)))
for f in mesh_iter:
    trg_f = trg_mesh_files[f]
    mesh_id = os.path.basename(trg_f).split('_')[0]
    std_f = os.path.join(meshes_dir, mesh_id+'_std.ply')
    if not eval_designated:
        pca_f = os.path.join(meshes_dir, mesh_id+'_pca{:04d}.ply'.format(K))
    else:
        pca_f = os.path.join(meshes_dir, mesh_id + designated_name)
    trg_mesh = pytorch3d.io.IO().load_mesh(trg_f).to(device)
    std_mesh = pytorch3d.io.IO().load_mesh(std_f).to(device)
    pca_mesh = pytorch3d.io.IO().load_mesh(pca_f).to(device)
    trg_pts0 = pytorch3d.ops.sample_points_from_meshes(trg_mesh, chamfer_n_pts)
    trg_pts1 = pytorch3d.ops.sample_points_from_meshes(trg_mesh, chamfer_n_pts)
    std_pts = pytorch3d.ops.sample_points_from_meshes(std_mesh, chamfer_n_pts)
    pca_pts = pytorch3d.ops.sample_points_from_meshes(pca_mesh, chamfer_n_pts)
    chamfer_trgtrg, _ = pytorch3d.loss.chamfer_distance(trg_pts0, trg_pts1)
    chamfer_trgstd, _ = pytorch3d.loss.chamfer_distance(trg_pts0, std_pts)
    chamfer_trgpca, _ = pytorch3d.loss.chamfer_distance(trg_pts0, pca_pts)
    chamfer_trgtrg_all.append(chamfer_trgtrg.cpu().numpy())
    chamfer_trgstd_all.append(chamfer_trgstd.cpu().numpy())
    chamfer_trgpca_all.append(chamfer_trgpca.cpu().numpy())
```

```python
# show error statistics
def display_error_stats(errors, error_type, save_file=None):
    avgstd_err_msg = 'mean error: {:.10f} +- {:.10f}'.format(np.mean(errors), np.std(errors))
    minmax_err_msg = 'max min error: {:.10f}, {:.10f}'.format(np.max(errors), np.min(errors))
    print(error_type)
    print(avgstd_err_msg)
    print(minmax_err_msg)
    if save_file:
        with open(save_file, 'a') as f:
            f.write(error_type + '\n')
            f.write(avgstd_err_msg + '\n')
            f.write(minmax_err_msg + '\n')
# display and save error stats
eval_dir = os.path.join(meshes_dir, 'evaluation')
if not os.path.isdir(eval_dir):
    os.mkdir(eval_dir)
if not eval_designated:
    summary = os.path.join(eval_dir, 'pca{:04d}_summary.txt'.format(K))
else:
    designated_save_name = designated_name[1:-4]
    summary = os.path.join(eval_dir, designated_save_name + '_summary.txt') # assume name starting with '_'
if os.path.isfile(summary):
    warnings.warn('Summary file already exists. Current data are appended.')
display_error_stats(chamfer_trgtrg_all, 'trg vs. trg', save_file=summary)
display_error_stats(chamfer_trgstd_all, 'trg vs. std', save_file=summary)
display_error_stats(chamfer_trgpca_all, 'trg vs. pca', save_file=summary)
# save the evaluation results as back-up
if not eval_designated:
    error_file = os.path.join(eval_dir, 'pca{:04d}_chamfer_trgtrg.pickle'.format(K))
    with open(error_file, 'wb') as f:
        pickle.dump(chamfer_trgtrg_all, f)
    error_file = os.path.join(eval_dir, 'pca{:04d}_chamfer_trgstd.pickle'.format(K))
    with open(error_file, 'wb') as f:
        pickle.dump(chamfer_trgstd_all, f)
    error_file = os.path.join(eval_dir, 'pca{:04d}_chamfer_trgpca.pickle'.format(K))
    with open(error_file, 'wb') as f:
        pickle.dump(chamfer_trgpca_all, f)
else:
    error_file = os.path.join(eval_dir, designated_save_name + '_chamfer_trgpca.pickle'.format(K))
    with open(error_file, 'wb') as f:
        pickle.dump(chamfer_trgpca_all, f)
```

# Shape Estimation with PCA Basis given Full GT Point Cloud

Run the "PCA Basis" code block up until the last code cell (**excluding the last cell**). Currently, I do not save and reload pca results, so need to rerun it every time. Also run the "mesh_deform_loss" in the "Single Object" code block.

```python
import numpy as np
from tqdm.notebook import tqdm
import timeit
import pytorch3d.io
import pytorch3d.structures
import pytorch3d.loss
import pytorch3d.ops

# user input
meshes_dir = '/home/yihao/Research/ShapeBasis/Records/20221026_014123_obj03001627'
n_points_trg = 1000 # how many points sampled from target mesh (mimic depth image point cloud of the mesh)
n_points_src = 5000 # added 202300808
verts_as_points = False
init_method = 'mean' # avg: average pca basis weights over the dataset. mean: mean shape. mean and avg effectively no difference on recon
n_optim_iters = 2000
bs = 64
twingpu = False
wt_chamfer = 1.0 # default 1.0
wt_edge = 0.0 # defualt 1.0
wt_normal = 0.0 # default 0.01
wt_laplacian = 0.0 # default 0.1
wt_ptedge = 0.0
wt_ptface = 0.0
wt_pca_params_reg = 2e-3
```

```python
# get the device
if twingpu:
    device = "cuda:0,1" if torch.cuda.is_available() else "cpu"
else:
    device = "cuda" if torch.cuda.is_available() else "cpu"
print(device)
# load the pca results to torch
pca_mean = torch.tensor(pca.mean_).to(torch.float32).to(device) # (n_verts*3)
pca_bases = torch.tensor(pca.components_).to(torch.float32).to(device) # (K, n_verts*3)
pca_basis_params_torch = torch.tensor(pca_basis_params).to(torch.float32).to(device) # (n_obj,K)
uni_faces = faces.detach().to(device) # detach() returns a new tensor and no requires_grad
K = pca_bases.size(dim=0) # int
n_verts = int(pca_mean.size(dim=0) / 3)
print('n_verts: ', n_verts)
```

```python
# load saved target meshes and run fitting (alternative: load a ShapeNet dataset with pytorch3d)
trg_mesh_files = glob.glob(os.path.join(meshes_dir, '*_trg.ply'))
n_meshes = len(trg_mesh_files)
loop = tqdm(range(0, n_meshes, bs))
for i in loop:
    # load a batch of target meshes
    start = timeit.default_timer()
    trg_meshes = []
    trg_meshes_paths = []
    for j in range(i, i + bs):
        if j >= n_meshes:
            break
        trg_meshes.append(pytorch3d.io.IO().load_mesh(trg_mesh_files[j], include_textures=False, device=device))
        trg_meshes_paths.append(trg_mesh_files[j])
    trg_mesh_batch = pytorch3d.structures.join_meshes_as_batch(trg_meshes, include_textures=False)
    actual_bs = len(trg_meshes)
    pause0 = timeit.default_timer()
    # set up optimization variable and optimizer
    if init_method == 'mean':
        pca_est_params = torch.full((actual_bs, K), 0.0, device=device, requires_grad=True).to(torch.float32) # pca_mean is added in init 
    if init_method == 'avg':
        pca_est_params = torch.mean(pca_basis_params_torch, dim=0, keepdim=True).expand(actual_bs, -1).clone().requires_grad_() # -1 not change size = K. clone() as expand refers to single memory
    batch_faces = uni_faces.expand(actual_bs, -1, -1)
    optimizer = torch.optim.SGD([pca_est_params], lr=1.0, momentum=0.9)
    pause1 = timeit.default_timer()
    # run iterative optimization
    for m in range(n_optim_iters):
        optimizer.zero_grad()
        new_verts = torch.matmul(pca_est_params, pca_bases) + pca_mean # dim broadcastable
        new_verts = torch.reshape(new_verts, (actual_bs, -1, 3))
        recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
        [loss_chamfer, loss_edge, loss_normal, loss_laplacian, loss_ptedge, loss_ptface] \
                = mesh_deform_loss(recon_meshes, trg_mesh_batch, n_samples_trg=n_points_trg,
                                   use_temp_verts=verts_as_points, n_samples_temp=n_points_src,
                                   chamfer_on=wt_chamfer, edge_on=wt_edge, normal_on=wt_normal,
                                   laplacian_on=wt_laplacian, ptedge_on=wt_ptedge, ptface_on=wt_ptface)
        loss_pca_params_reg = torch.mean(torch.square(torch.norm(pca_est_params, 'fro', dim=1) - 5.8792543))
        total_loss = (wt_chamfer * loss_chamfer + wt_edge * loss_edge + \
                     wt_normal * loss_normal + wt_laplacian * loss_laplacian + \
                     wt_ptedge * loss_ptedge + wt_ptface * loss_ptface + \
                     wt_pca_params_reg * loss_pca_params_reg) * actual_bs
        total_loss.backward()
        optimizer.step()
    new_verts = torch.matmul(pca_est_params, pca_bases) + pca_mean # dim broadcastable
    new_verts = torch.reshape(new_verts, (actual_bs, -1, 3))
    recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
    pause2 = timeit.default_timer()
    # save the results
    for b in range(actual_bs):
        recon_save_path = trg_meshes_paths[b][:-8] + '_pca{:04d}_recon_{:06d}trgpts_{:.0e}cregsq5.88.ply'.format(K, n_points_trg, wt_pca_params_reg)
        pytorch3d.io.IO().save_mesh(recon_meshes[b], recon_save_path)
        # pytorch3d.io.IO().save_mesh(trg_mesh_batch[b], 'mesh_recon_trg_test.ply')
    torch.cuda.empty_cache()
    pause3 = timeit.default_timer()
```

# Pose and Scale Estimation between Two Close GT Meshes

Run the collate function cell under "`Batched Objects`".

```python
def apply_trsfms_to_pts(pts, rot, t, scale=1.0):
    # pts: [bs,n,3] batched padded point clouds, rot: [bs,3,3], t: [bs,3,1], scale: [bs]
    # output: [bs,n,3]
    pts_new = scale.unsqueeze(-1).unsqueeze(-1) * pts # [bs,n,3] scale around center (assumed origin before translating)
    pts_new = torch.matmul(rot.unsqueeze(1), pts_new.unsqueeze(-1)) # [bs,n,3,1]
    pts_new += t.unsqueeze(1)
    return pts_new.squeeze(-1)

def compose_poses(rot1, rot2, t1, t2, scl1, scl2):
    # rot: [bs,3,3], t: [bs,3,1], scl: [bs]
    # output: T1.dot(T2)
    rot_out = torch.matmul(rot1, rot2) # [bs,3,3]
    t_out = torch.matmul(rot1, t2) + t1 # [bs,3,1]
    scl_out = scl1 * scl2
    return rot_out, t_out, scl_out

def knn_umeyama(points_trg, points_src, est_scale, len_trg=None, len_src=None):
    # points: [bs,n,3] if padded with zeros, need the len argument
    # sim_trsfm: [0] R [bs,3,3] [1] t [bs,3] [2] s [bs] for sXR+t=Y
    # corresponding_points_alignment has no length argument or use PointClouds object (different lengths handled)
    knn_output = pytorch3d.ops.knn_points(points_trg, points_src, lengths1=len_trg, lengths2=len_src,
                                          K=1, return_nn=True, return_sorted=False) # associate points_src to each in points_trg
    sim_trsfm = pytorch3d.ops.corresponding_points_alignment(knn_output[2].squeeze(-2), points_trg, estimate_scale=est_scale)
    return sim_trsfm

def knn_umeyama_hetero(ptcobj_trg, points_src, est_scale, teaser=False, teaserpp_solver=None):
    # pass pytorch3d.structures.Pointclouds to handle point sets of different lengths (for trg)
    # teaserpp_solver is the teaserpp class object below
    points_trg_padded = ptcobj_trg.points_padded() # [bs,n,3]
    knn_output = pytorch3d.ops.knn_points(points_trg_padded, points_src,
                                          lengths1=ptcobj_trg.num_points_per_cloud(), lengths2=None,
                                          K=1, return_nn=True, return_sorted=False) # dists [bs,n,K], idx [bs,n,K], nn [bs,n,K,D]
    nnsrc = knn_output[2].squeeze(-2) # [bs,n,3]
    # as in doc, return a padded output for output[0] and [1] but not [2]. The padded zero index from [1] picks up points indexed 0.
    points_nnsrc = pytorch3d.structures.padded_to_list(nnsrc, ptcobj_trg.num_points_per_cloud().tolist())
    if not teaser:
        ptcobj_nnsrc = pytorch3d.structures.Pointclouds(points_nnsrc) # use padded_to_list to remove gathered points due to 0-padding
        sim_trsfm = pytorch3d.ops.corresponding_points_alignment(ptcobj_nnsrc, ptcobj_trg, estimate_scale=est_scale)
    else:
        trg_ptc_list = ptcobj_trg.points_list()
        sol_rots, sol_trsl, sol_scls = [], [], []
        for b in range(len(trg_ptc_list)):
            teapp_sol = teaserpp_solver.solve(points_nnsrc[b].transpose(-1,-2).detach().cpu().numpy(),
                                              trg_ptc_list[b].transpose(-1,-2).detach().cpu().numpy())
            sol_rots.append(teapp_sol.rotation)
            sol_trsl.append(teapp_sol.translation)
            sol_scls.append(teapp_sol.scale)
        sim_trsfm = (torch.tensor(sol_rots, dtype=nnsrc.dtype, device=nnsrc.device).transpose(-1,-2), # [bs,3,3] in pytorch3d format s*src*R+t
                     torch.tensor(sol_trsl, dtype=nnsrc.dtype, device=nnsrc.device), # [bs,3]
                     torch.tensor(sol_scls, dtype=nnsrc.dtype, device=nnsrc.device)) # [bs]
    return sim_trsfm

class teaserpp():
    def __init__(self, max_rot_iters=100, rot_cost_th=1e-12, noise_bound=0.01):
        # populate the parameters from https://teaser.readthedocs.io/en/master/quickstart.html
        solver_params = teapp.RobustRegistrationSolver.Params()
        solver_params.cbar2 = 1
        solver_params.noise_bound = noise_bound
        solver_params.estimate_scaling = True
        solver_params.rotation_estimation_algorithm = (
            teapp.RobustRegistrationSolver.ROTATION_ESTIMATION_ALGORITHM.GNC_TLS)
        solver_params.rotation_gnc_factor = 1.4
        solver_params.rotation_max_iterations = max_rot_iters
        solver_params.rotation_cost_threshold = rot_cost_th
        # print("TEASER++ Parameters are:", solver_params)
        self.solver = teapp.RobustRegistrationSolver(solver_params)
    
    def solve(self, src, dst):
        # src: [3,n] dst: [3,n]
        # solution is in the form of dst = sR*src + t accessible by solution.rotation [3,3] translation [3] scale []
        self.solver.solve(src, dst)
        return self.solver.getSolution()

# def knn_umeyama_hetero_ransac_b(ptcobj_trg, points_src, est_scale, ransac_n=10, ransac_th=0.1, ransac_inrate=0.5, ransac_maxiter=50):
#     # pass pytorch3d.structures.Pointclouds to handle point sets of different lengths (for trg)
#     # ransac_n: the size of subset to select randomly ransac_th: inlier threshold ransac_maxiter: max number of iterations
#     points_trg_padded = ptcobj_trg.points_padded() # [bs,n,3]
#     knn_output = pytorch3d.ops.knn_points(points_trg_padded, points_src,
#                                           lengths1=ptcobj_trg.num_points_per_cloud(), lengths2=None,
#                                           K=1, return_nn=True, return_sorted=False) # dists [bs,n,K], idx [bs,n,K], nn [bs,n,K,D]
#     nnsrc = knn_output[2].squeeze(-2) # [bs,n,3]
#     # ransac - set up multinomial probability vector [bs,n]
#     prob_one = torch.ones_like(knn_output[0][:,:,0]) # [bs,n]
#     prob_one = pytorch3d.structures.padded_to_list(prob_one, ptcobj_trg.num_points_per_cloud().tolist()) # remove the ones corresponding to zero paddings
#     prob_one = pytorch3d.structures.list_to_padded(prob_one) # [bs,n] with zero paddings having zero weighting
#     passed = torch.zeros_like(knn_output[0][:,0,0]).to(torch.bool) # [bs] to record pass of inlier test
#     for r in range(ransac_maxiter):
#         # select a random subset
#         subset_idx = torch.multinomial(prob_one, ransac_n, replacement=False) # [bs,n_r]
#         subset_idx = subset_idx.unsqueeze(-1).expand(-1,-1,3)
#         subset = torch.gather(nnsrc, dim=-2, index=subset_idx) # [bs,n_r,3]
#         subset_trg = torch.gather(points_trg_padded, dim=-2, index=subset_idx) # [bs,n_r,3]
#         # estimate the transformation
#         sim_trsfm = pytorch3d.ops.corresponding_points_alignment(subset, subset_trg, estimate_scale=est_scale) # [bs,3,3] [bs,3] [bs]
#         # check inliers (from doc sXR + T = Y)
#         approx_trg = (sim_trsfm[2].unsqueeze(-1).unsqueeze(-1) * nnsrc).matmul(sim_trsfm[0]) + sim_trsfm[1].unsqueeze(-2) # [bs,n,3]
#         residuals = torch.norm(approx_trg*prob_one.unsqueeze(-1)-points_trg_padded, dim=-1) # [bs,n] padding_mask to cancel the translation added to the zero paddings
#         inliers = residuals.le(ransac_th).to(torch.float32) * prob_one # [bs,n] prob_one turn off inliers due to zero paddings
#         inlier_rate = inliers.sum(-1) / prob_one.sum(-1) # [bs]
#         passed = inlier_rate.ge(ransac_inrate).logical_or(passed) # [bs]
#         # recompute transformation with the inlier set
#         # [TODO] what if there are not enough inliers for umeyama?
#     # as in doc, return a padded output for output[0] and [1] but not [2]. The padded zero index from [1] picks up points indexed 0.
#     points_nnsrc = pytorch3d.structures.padded_to_list(nnsrc, ptcobj_trg.num_points_per_cloud().tolist())
#     ptcobj_nnsrc = pytorch3d.structures.Pointclouds(points_nnsrc) # use padded_to_list to remove gathered points due to 0-padding
#     sim_trsfm = pytorch3d.ops.corresponding_points_alignment(ptcobj_nnsrc, ptcobj_trg, estimate_scale=est_scale)
#     return sim_trsfm

def knn_umeyama_hetero_ransac_s(ptcobj_trg, points_src, est_scale, ransac_n=10, ransac_th=0.1, ransac_inrate=0.5, ransac_maxiter=50):
    # pass pytorch3d.structures.Pointclouds to handle point sets of different lengths (for trg)
    # ransac_n: the size of subset to select randomly ransac_th: inlier threshold ransac_maxiter: max number of iterations
    points_trg_padded = ptcobj_trg.points_padded() # [bs,n,3]
    knn_output = pytorch3d.ops.knn_points(points_trg_padded, points_src,
                                          lengths1=ptcobj_trg.num_points_per_cloud(), lengths2=None,
                                          K=1, return_nn=True, return_sorted=False) # dists [bs,n,K], idx [bs,n,K], nn [bs,n,K,D]
    # ransac - set up multinomial probability vector
    prob_one = torch.ones_like(knn_output[0][:,:,0]) # [bs,n]
    # get list representation for looping over batch
    trg_len_list = ptcobj_trg.num_points_per_cloud().tolist()
    prob_one = pytorch3d.structures.padded_to_list(prob_one, trg_len_list) # list[n_i] remove the ones corresponding to zero paddings
    pts_trg = ptcobj_trg.points_list() # list[n_i,3]
    nnsrc_pad = knn_output[2].squeeze(-2) # [bs,n,3]
    nn_src = pytorch3d.structures.padded_to_list(nnsrc_pad, trg_len_list) # list[n_i,3]
    sim_trsfm = [torch.tensor([], dtype=points_src.dtype, device=points_src.device),
                 torch.tensor([], dtype=points_src.dtype, device=points_src.device),
                 torch.tensor([], dtype=points_src.dtype, device=points_src.device)]
    # record stats for ransac parameter tuning
    passing_mean_residual = torch.tensor([], dtype=points_src.dtype, device=points_src.device)
    passing_inlier_rate = torch.tensor([], dtype=points_src.dtype, device=points_src.device)
    final_nmb_iters = []
    mean_iters_mean_residual = torch.tensor([], dtype=points_src.dtype, device=points_src.device)
    mean_iters_mean_inlier_rate = torch.tensor([], dtype=points_src.dtype, device=points_src.device)
    std_iters_mean_residual = torch.tensor([], dtype=points_src.dtype, device=points_src.device)
    std_iters_mean_inlier_rate = torch.tensor([], dtype=points_src.dtype, device=points_src.device)
    batch_pass = False
    for b in range(len(trg_len_list)):
        # record stats
        iters_mean_residual = torch.tensor([], dtype=points_src.dtype, device=points_src.device)
        iters_mean_inlier_rate = torch.tensor([], dtype=points_src.dtype, device=points_src.device)
        final_nmb_iters.append(0)
        for r in range(ransac_maxiter):
            # select a random subset
            subset_idx = torch.multinomial(prob_one[b], ransac_n, replacement=False) # [n_r]
            subset_src = torch.index_select(nn_src[b], dim=0, index=subset_idx).unsqueeze(0) # [1,n_r,3]
            subset_trg = torch.index_select(pts_trg[b], dim=0, index=subset_idx).unsqueeze(0) # [1,n_r,3]
            # estimate the transformation
            sim_trsfm_sub = pytorch3d.ops.corresponding_points_alignment(subset_src, subset_trg, estimate_scale=est_scale) # [1,3,3] [1,3] [1]
            # check inliers (sXR + T = Y)
            approx_trg = (sim_trsfm_sub[2] * nn_src[b]).matmul(sim_trsfm_sub[0].squeeze(0)) + sim_trsfm_sub[1] # [n_i,3]
            residuals = torch.square(torch.norm(approx_trg - pts_trg[b], dim=-1)) # [n_i]
            inliers = residuals.le(ransac_th) # [n_i]
            inlier_rate = inliers.to(torch.float32).sum() / prob_one[b].sum() # []
            passed = (inlier_rate >= ransac_inrate and inliers.to(torch.float32).sum() >= ransac_n).detach().cpu().numpy()
            # record stats
            iters_mean_residual = torch.cat([iters_mean_residual, residuals.mean().unsqueeze(0)], dim=0)
            iters_mean_inlier_rate = torch.cat([iters_mean_inlier_rate, inlier_rate.unsqueeze(0)], dim=0)
            final_nmb_iters[-1] = r + 1
            # reestimate with inliers
            if passed: # somehow this works without cpu().numpy()
                inliers_src = nn_src[b][inliers,:].unsqueeze(0) # [1,n_in,3]
                inliers_trg = pts_trg[b][inliers,:].unsqueeze(0) # [1,n_in,3]
                sim_trsfm_b = pytorch3d.ops.corresponding_points_alignment(inliers_src, inliers_trg, estimate_scale=est_scale) # [1,3,3] [1,3] [1]
                # record stats for a batch
                passing_mean_residual = torch.cat([passing_mean_residual, residuals.mean().unsqueeze(0)], dim=0)
                passing_inlier_rate = torch.cat([passing_inlier_rate, inlier_rate.unsqueeze(0)], dim=0)
                batch_pass = True
                break
        # if maxiter exceeded just run on all correspondences
        if not passed:
            sim_trsfm_b = pytorch3d.ops.corresponding_points_alignment(nn_src[b].unsqueeze(0), pts_trg[b].unsqueeze(0), estimate_scale=est_scale) # [1,3,3] [1,3] [1]
        sim_trsfm[0] = torch.cat([sim_trsfm[0], sim_trsfm_b[0]], dim=0)
        sim_trsfm[1] = torch.cat([sim_trsfm[1], sim_trsfm_b[1]], dim=0)
        sim_trsfm[2] = torch.cat([sim_trsfm[2], sim_trsfm_b[2]], dim=0)
        # record stats for a batch
        mean_iters_mean_residual = torch.cat([mean_iters_mean_residual, iters_mean_residual.mean().unsqueeze(0)], dim=0) # [n_iters] -> []
        mean_iters_mean_inlier_rate = torch.cat([mean_iters_mean_inlier_rate, iters_mean_inlier_rate.mean().unsqueeze(0)], dim=0) # [n_iters] -> []
        std_iters_mean_residual = torch.cat([std_iters_mean_residual, torch.std(iters_mean_residual).unsqueeze(0)], dim=0)
        std_iters_mean_inlier_rate = torch.cat([std_iters_mean_inlier_rate, torch.std(iters_mean_inlier_rate).unsqueeze(0)], dim=0)
    # summarize stats
    if batch_pass: # run only when there is a pass in the batch o.w. mean([]) = nan
        passing_mean_residual = passing_mean_residual.mean() # mean over batch (may be less than batch size due to low passing rate)
        passing_inlier_rate = passing_inlier_rate.mean()
    mean_iters_mean_residual = mean_iters_mean_residual.mean() # over batch
    mean_iters_mean_inlier_rate = mean_iters_mean_inlier_rate.mean()
    std_iters_mean_residual = std_iters_mean_residual.mean() # variations from ransac iter to iter
    std_iters_mean_inlier_rate = std_iters_mean_inlier_rate.mean()
    ransac_stats = (passing_mean_residual, passing_inlier_rate, mean_iters_mean_residual, std_iters_mean_residual,
                    mean_iters_mean_inlier_rate, std_iters_mean_inlier_rate, np.mean(final_nmb_iters), batch_pass)
    return sim_trsfm, ransac_stats

def knn_umeyama_hetero_em_ransac_s(ptcobj_trg, points_src, est_scale, update_sigma, knn_k, sigma_sq_inv, bs, deg=3,
                                   ransac_n=10, ransac_th=0.1, ransac_inrate=0.5, ransac_maxiter=50, inlier_check_nn=True):
    # ptcobj_trg: Pointclouds object points_src: [bs,m,3] est_scale: bool, knn_k: int, sigma_sq_inv: [bs,n] with one/zero padding
    # ransac_n: the size of subset to select randomly ransac_th: inlier threshold ransac_maxiter: max number of iterations
    # ransac_inrate: ratio of inliers beyond which consider a valid transformation inlier_check_nn: nearest neighbor for inlier checking or avg of k nn's
    knn_output = pytorch3d.ops.knn_points(ptcobj_trg.points_padded(), points_src,
                                          lengths1=ptcobj_trg.num_points_per_cloud(), lengths2=None,
                                          K=knn_k, return_nn=True, return_sorted=True) # dists [bs,n,k], idx [bs,n,k], nn [bs,n,k,3] sorted need for inlier checking
    # update sigma using last estimates
    if update_sigma:
        post_prob = association_prob(knn_k) * gauss_prob(knn_output[0], sigma_sq_inv) # [bs,n,k] knn_output[0] has zero padding
        post_prob = post_prob / torch.sum(post_prob, dim=-1, keepdim=True) # [bs,n,k]
        knn_dists_onepad = pytorch3d.structures.padded_to_list(knn_output[0], ptcobj_trg.num_points_per_cloud().tolist())
        knn_dists_onepad = pytorch3d.structures.list_to_padded(knn_dists_onepad, pad_value=1.0) # switch to one padding to avoid zero denominator
        sigma_sq_inv = (3.0 * post_prob.sum(-1) / (post_prob * knn_dists_onepad).sum(-1)).detach() # [bs,n]
    # compute posterior prob. as weights
    post_prob = association_prob(knn_k) * gauss_prob(knn_output[0], sigma_sq_inv) # [bs,n,k] knn_output[0] has zero padding
    post_prob = post_prob / torch.sum(post_prob, dim=-1, keepdim=True) # [bs,n,k]
    # compute weights in EM (including the gaussian ln(p(X,Z)) cov discounting)
    weights = post_prob * sigma_sq_inv.unsqueeze(-1) # [bs,n,k]
    # ransac - set up multinomial probability vector
    prob_one = torch.ones_like(knn_output[0][:,:,0]) # [bs,n]
    # get list representation for looping over batch
    trg_len_list = ptcobj_trg.num_points_per_cloud().tolist()
    prob_one = pytorch3d.structures.padded_to_list(prob_one, trg_len_list) # list[n_i] remove zero paddings
    pts_trg_pad = ptcobj_trg.points_padded().unsqueeze(-2).expand(-1,-1,knn_k,-1).clone() # [bs,n,k,3]
    pts_trg = pytorch3d.structures.padded_to_list(pts_trg_pad, trg_len_list) # list[n_i,k,3]
    nn_src = pytorch3d.structures.padded_to_list(knn_output[2], trg_len_list) # list[n_i,k,3]
    em_wts = pytorch3d.structures.padded_to_list(weights, trg_len_list) # list[n_i,k]
    sim_trsfm = [torch.tensor([], dtype=points_src.dtype, device=points_src.device),
                 torch.tensor([], dtype=points_src.dtype, device=points_src.device),
                 torch.tensor([], dtype=points_src.dtype, device=points_src.device)]
    for b in range(len(trg_len_list)):
        for r in range(ransac_maxiter):
            # select a random subset
            subset_idx = torch.multinomial(prob_one[b], ransac_n, replacement=False) # [n_r]
            subset_src = torch.index_select(nn_src[b], dim=0, index=subset_idx).unsqueeze(0) # [1,n_r,k,3]
            subset_trg = torch.index_select(pts_trg[b], dim=0, index=subset_idx).unsqueeze(0) # [1,n_r,k,3]
            subset_wts = torch.index_select(em_wts[b], dim=0, index=subset_idx).unsqueeze(0) # [1,n_r,k]
            # estimate the transformation
            sim_trsfm_sub = pytorch3d.ops.corresponding_points_alignment(subset_src.reshape(1,-1,3), subset_trg.reshape(1,-1,3),
                                                                         weights=subset_wts.reshape(1,-1), estimate_scale=est_scale) # [1,3,3] [1,3] [1]
            # check inliers (sXR + T = Y)
            if inlier_check_nn:
                approx_trg = (sim_trsfm_sub[2] * nn_src[b][:,0,:]).matmul(sim_trsfm_sub[0].squeeze(0)) + sim_trsfm_sub[1] # [n_i,3]
                residuals = torch.square(torch.norm(approx_trg - pts_trg[b][:,0,:], dim=-1)) # [n_i]
            else:
                approx_trg = (sim_trsfm_sub[2] * nn_src[b]).matmul(sim_trsfm_sub[0]) + sim_trsfm_sub[1] # [n_i,k,3]
                residuals = torch.square(torch.norm(approx_trg - pts_trg[b], dim=-1)).mean(-1) # [n_i]
            inliers = residuals.le(ransac_th) # [n_i]
            inlier_rate = inliers.to(torch.float32).sum() / prob_one[b].sum() # []
            passed = (inlier_rate >= ransac_inrate and inliers.to(torch.float32).sum() >= ransac_n).detach().cpu().numpy()
            # reestimate with inliers
            if passed: # somehow this works without cpu().numpy()
                inliers_src = nn_src[b][inliers,:,:].unsqueeze(0) # [1,n_in,k,3]
                inliers_trg = pts_trg[b][inliers,:,:].unsqueeze(0) # [1,n_in,k,3]
                inliers_wts = em_wts[b][inliers,:].unsqueeze(0) # [1,n_in,k]
                sim_trsfm_b = pytorch3d.ops.corresponding_points_alignment(inliers_src.reshape(1,-1,3), inliers_trg.reshape(1,-1,3),
                                                                             weights=inliers_wts.reshape(1,-1), estimate_scale=est_scale) # [1,3,3] [1,3] [1]
                break
        # exceeding maxiter use all correspondences to estimate
        if not passed:
            sim_trsfm_b = pytorch3d.ops.corresponding_points_alignment(nn_src[b].reshape(1,-1,3), pts_trg[b].reshape(1,-1,3),
                                                                       weights=em_wts[b].reshape(1,-1), estimate_scale=est_scale) # [1,3,3] [1,3] [1]
        sim_trsfm[0] = torch.cat([sim_trsfm[0], sim_trsfm_b[0]], dim=0)
        sim_trsfm[1] = torch.cat([sim_trsfm[1], sim_trsfm_b[1]], dim=0)
        sim_trsfm[2] = torch.cat([sim_trsfm[2], sim_trsfm_b[2]], dim=0)
    return sim_trsfm, sigma_sq_inv

def knn_umeyama_hetero_em(ptcobj_trg, points_src, est_scale, update_sigma, knn_k, sigma_sq_inv, bs, deg=3):
    # ptcobj_trg: Pointclouds object points_src: [bs,m,3] est_scale: bool, knn_k: int, sigma_sq_inv: [bs,n] with one/zero padding
    knn_output = pytorch3d.ops.knn_points(ptcobj_trg.points_padded(), points_src,
                                          lengths1=ptcobj_trg.num_points_per_cloud(), lengths2=None,
                                          K=knn_k, return_nn=True, return_sorted=False)
    # update sigma using last estimates
    if update_sigma:
        post_prob = association_prob(knn_k) * gauss_prob(knn_output[0], sigma_sq_inv) # [bs,n,k] knn_output[0] has zero padding
        post_prob = post_prob / torch.sum(post_prob, dim=-1, keepdim=True) # [bs,n,k]
        knn_dists_onepad = pytorch3d.structures.padded_to_list(knn_output[0], ptcobj_trg.num_points_per_cloud().tolist())
        knn_dists_onepad = pytorch3d.structures.list_to_padded(knn_dists_onepad, pad_value=1.0) # switch to one padding to avoid zero denominator
        sigma_sq_inv = (3.0 * post_prob.sum(-1) / (post_prob * knn_dists_onepad).sum(-1)).detach() # [bs,n]
    # compute posterior prob. as weights
    post_prob = association_prob(knn_k) * gauss_prob(knn_output[0], sigma_sq_inv) # [bs,n,k] knn_output[0] has zero padding
    post_prob = post_prob / torch.sum(post_prob, dim=-1, keepdim=True) # [bs,n,k]
    # compute weights in EM (including the gaussian ln(p(X,Z)) cov discounting)
    weights = post_prob * sigma_sq_inv.unsqueeze(-1) # [bs,n,k]
    # prepare associated points
    points_nnsrc = pytorch3d.structures.padded_to_list(knn_output[2], ptcobj_trg.num_points_per_cloud().tolist()) # [bs,n,k,3]->list[n_i,k,3]
    points_nnsrc = pytorch3d.structures.list_to_padded(points_nnsrc) # replace extra gathered points with zero paddings (unnecessary but to be safe)
    weights = pytorch3d.structures.padded_to_list(weights, ptcobj_trg.num_points_per_cloud().tolist()) # [bs,n,k]->list[n_i,k]
    weights = pytorch3d.structures.list_to_padded(weights) # [bs,n,k] with zero padding. rely on this zero padding to remove zero paddings in points
    points_trg = ptcobj_trg.points_padded() # [bs,n,3]
    points_nntrg = points_trg.unsqueeze(-2).expand(-1,-1,knn_k,-1).clone() # [bs,n,k,3]
    sim_trsfm = pytorch3d.ops.corresponding_points_alignment(points_nnsrc.reshape((bs,-1,3)), points_nntrg.reshape((bs,-1,3)),
                                                             weights=weights.reshape((bs,-1)), estimate_scale=est_scale)
    return sim_trsfm, sigma_sq_inv

def association_prob(knn_k):
    # assume uniform distribution for all possible associations
    return 1.0 / knn_k

def gauss_prob(sq_dist, sigma_sq_inv, deg=3.0):
    # sq_dist: in the form of knn_points output[0] [bs,n,k] 
    # sigma_sq_inv: [bs,n] with one/zero padding for different trg point cloud len
    # compute probability for zero mean gaussian with a diagonal cov
    normalization = torch.sqrt(torch.pow(0.5*(1.0/np.pi)*sigma_sq_inv, deg)) # [bs,n]
    exponential = - 0.5 * sq_dist * sigma_sq_inv.unsqueeze(-1) # [bs,n,k]
    return normalization.unsqueeze(-1) * torch.exp(exponential) # [bs,n,k]

def icp(points_trg, points_src, est_scale):
    # no point cloud length argument. Assume same length for trg and src and no padding. Or use Pointclouds object.
    # points: [bs,n_trg/n_src,3] or Pointclouds. For each trg point, find nearest neighbor in points_src (according to src code)
    # namedtuple: converged, rmse, Xt, RTs (R [bs,3,3], T [bs,3], s [bs]), t_history (pose estimate at each step)
    # s*points_trg*R + t = points_src[nearest]
    icp_soln = pytorch3d.ops.iterative_closest_point(points_trg, points_src, estimate_scale=est_scale)
    if not icp_soln.converged:
        print('icp not converged')
    # flip/invert the pose. want the transformation applied to points_src
    # points_trg = (1/s)points_src[nearest]R.T - (1/s)tR.T
    sim_trsfm = (icp_soln.RTs.R.transpose(-1,-2), # [bs,3,3]
                 -((1.0/icp_soln.RTs.s).unsqueeze(-1) * icp_soln.RTs.T).unsqueeze(-2).matmul(
                 icp_soln.RTs.R.transpose(-1,-2)).squeeze(-2), # [bs,1,3] -> [bs,3]
                 1.0/icp_soln.RTs.s) # [bs]
    return sim_trsfm

def extract_cano_mesh_bbx(cano_meshes, compute_vol=False, scale=1.0):
    # assume the mesh in canonical orientation. extract axis-aligned bounding box.
    # cano_meshes: a batch of pytorch3d meshes object
    # scale: [bs] needed to compute bbox volume
    # output: [bs,8,3] eight vertices of the scaled bbx following ops.box3d_overlap order
    bboxes = cano_meshes.get_bounding_boxes() # [bs,3,2] [bs,xyz,minmax]
    if compute_vol: # default scale=1.0 does not work
        bboxes *= scale.unsqueeze(-1).unsqueeze(-1)
        edges_len = torch.abs(bboxes[:,:,1] - bboxes[:,:,0]) # [bs,3]
        volume = edges_len[:,0] * edges_len[:,1] * edges_len[:,2] # [bs]
    vert0, vert1 = bboxes[:,:,0], torch.cat([bboxes[:,0:1,1], bboxes[:,1:,0]], dim=-1) # [bs,3]
    vert2, vert3 = torch.cat([bboxes[:,0:2,1], bboxes[:,2:,0]], dim=-1), torch.cat([bboxes[:,0:1,0], bboxes[:,1:2,1], bboxes[:,2:,0]], dim=-1)
    vert4, vert5 = torch.cat([bboxes[:,0:2,0], bboxes[:,2:,1]], dim=-1), torch.cat([bboxes[:,0:1,1], bboxes[:,1:2,0], bboxes[:,2:,1]], dim=-1)
    vert6, vert7 = bboxes[:,:,1], torch.cat([bboxes[:,0:1,0], bboxes[:,1:,1]], dim=-1)
    bbox_verts = torch.stack([vert0,vert1,vert2,vert3,vert4,vert5,vert6,vert7], dim=-2) # [bs,8,3]
    if compute_vol:
        return bbox_verts, volume
    else:
        return bbox_verts

def iou_chamfer_loss(points1, points2, len1=None, len2=None, bbox1=None, bbox2=None, iou=True, chamfer=True, reduce_batch=True, reduce_points=True):
    # distance-iou loss cannot be implemented as it needs the smallest enclosing bounding box over the two boxes. Ok on 2D.
    # in pytorch3d, box3d_overlap and _box3d_overlap class, it says backward not supported.
    # chamfer loss is two-way from x to y and y to x (see the source code). I only need one-way.
    # bbox: [bs,8,3] following ops.box3d_overlap convention. vol: [bs]. points: [bs,n,3]. len: [bs]
    # assume 1 is the target and 2 is the source
    # compute iou loss
    iou_loss = 0.0
    chamfer_loss = 0.0
    if iou:
        _, iou_loss = pytorch3d.ops.box3d_overlap(bbox1, bbox2) # [bs,bs]
        iou_loss = torch.diagonal(iou_loss, dim1=-2, dim2=-1).mean()
    if chamfer:
        # chamfer loss in pytorch3d uses default mean point reduction
        knn_output = pytorch3d.ops.knn_points(points1, points2, lengths1=len1, lengths2=len2, K=1, return_nn=False, return_sorted=False)
        # chamfer_loss = knn_output[0].sum(dim=(-1,-2)).mean(dim=0) # sum point reduction [bs,n,1] -> scalar
        if len1 != None:
            chamfer_loss = knn_output.dists[...,0].sum(-1) # [bs] in pytorch3d chamfer loss, also just return the knn dists as loss
            chamfer_loss /= len1.clamp(min=1).to(torch.float32) # [bs] taking average
            if not reduce_batch:
                if reduce_points:
                    return chamfer_loss
                else:
                    return chamfer_loss, knn_outputs.dists[...,0] # [bs,n] with paddings
            chamfer_loss = chamfer_loss.mean(0)
        else:
            if not reduce_batch:
                if reduce_points:
                    return knn_output.dists[...,0].mean(-1) # [bs]
                else:
                    return knn_output.dists[...,0].mean(-1), knn_output.dists[...,0] # [bs,n]
            chamfer_loss = knn_output.dists[...,0].mean(-1).mean(0) # mean point reduction [bs,n,1] -> scalar (assume all same lengths)
    return iou_loss, chamfer_loss

def get_gt_pose_init_pose(bs_actual, device, cano_verts, cano_faces, cano_bboxes, rot_std, t_std, scl_std, scl_lbd):
    # obtain random poses as gt, transform canonical meshes to target poses, and perturb gt poses for initial poses
    # cano_verts: [bs,n,3] padded zero, cano_faces: [bs,n_f,3], cano_bboxes: [bs,8,3] in the order as in pytorch3d.ops.box3d_overlap
    # others are scalars
    # get random rotations, translations, and scales to place the target mesh (i.e. gt pose)
    random_rots = pytorch3d.transforms.random_rotations(bs_actual, torch.float32, device) # [bs,3,3]
    random_trsl = torch.rand((bs_actual, 3, 1), dtype=torch.float32, device=device) + 0.2 # (0.2,1.2) [bs,3,1] (-1,1)->(0.2,1.2) 230328 for in-bound rendering
    random_scls = torch.rand((bs_actual), dtype=torch.float32, device=device) * 1.5 + 0.5 # (0.5,2) [bs]
    # apply to the target meshes. May use pytorch3d.transforms.Transform3d but lack transparency
    trg_verts = apply_trsfms_to_pts(cano_verts, random_rots, random_trsl, random_scls) # [bs,n,3]
    trg_meshes = pytorch3d.structures.Meshes(verts=trg_verts, faces=cano_faces).to(device)
    # transform the canonical bbox
    trg_bboxes = apply_trsfms_to_pts(cano_bboxes, random_rots, random_trsl, random_scls) # [bs,8,3]
    # use small perturbations to the ground truth to get an initial guess
    axis = np.random.rand(bs_actual, 3) * 2.0 - 1.0 # vector in a 3D cube in [-1,1) [bs,3]
    axis = axis / np.linalg.norm(axis, axis=-1, keepdims=True)
    angle = np.random.randn(bs_actual, 1) * rot_std
    axis_angle_ptb = torch.tensor(axis * angle, dtype=torch.float32, device=device) # [bs,3]
    rot_ptb = pytorch3d.transforms.axis_angle_to_matrix(axis_angle_ptb) # [bs,3,3]
    trsl_ptb = torch.randn((bs_actual, 3, 1), dtype=torch.float32, device=device) * t_std # [bs,3,1]
    scale_ptb = torch.clamp(torch.randn((bs_actual), dtype=torch.float32, device=device) * scl_std + 1.0, min=scl_lbd) # [bs]
    # apply the small perturbations to ground truth and obtain an initial mesh
    init_rots = rot_ptb.matmul(random_rots) # [bs,3,3]
    init_trsl = trsl_ptb + random_trsl # [bs,3,1]
    init_scls = scale_ptb * random_scls # [bs]
    return (random_rots, random_trsl, random_scls), (init_rots, init_trsl, init_scls), (trg_meshes, trg_bboxes)
```

```python
import numpy as np
import json
import os
import glob
import datetime
import gc
import torch
import pytorch3d.structures
import pytorch3d.transforms
import pytorch3d.loss
import pytorch3d.datasets
import pytorch3d.ops
import pytorch3d.utils
from torch.utils.data import Dataset, TensorDataset, DataLoader
import timeit
from tqdm.notebook import tqdm
import matplotlib.pyplot as plt

# inputs
shapenet_dir = '/media/yihao/Samsung_T5/Datasets/ShapeNetCore.v2'
# shapenet_dir = '/data/Datasets/ShapeNetCore.v2'
save_meta_dir = '/home/yihao/Research/ShapeBasis/Records'
bs = 32
twingpu = False
init_rot_std = 10/180*np.pi # (rad) angle stddev perturbation to the ground truth to obtain the initial guess (zero mean)
init_t_std = 0.1 # translation stddev (zero mean)
init_scl_std = 0.1 # scale stddev (one mean)
init_scl_lbd = 0.5 # lower cap for the initial scale
n_sample_pts = 1200 # number of points to sample from mesh (cannot use verts as samples since src and trg are gt meshes whose verts are sparse)
mode1 = 0 # 0: one-step alignment. 1: full icp
mode2 = -1 # 0: iou chamfer optimize scale only. 1: scale + translation. 2: R,t,s, o.w.: off
mode2_iters = 0
iou_wt = 1.0
chamfer_wt = 1.0
n_iters = 30
lrate = 0.05
seed = 2147483647
save_freq = 20


obj_id = "03001627" # chair 6778
# obj_id = '03211117' # display 1095
# obj_id = '03797390' # mug 214
# obj_id = '03642806' # laptop 460
taxonomy = json.load(open(os.path.join(shapenet_dir, 'taxonomy.json')))
save_dir = os.path.join(save_meta_dir, datetime.datetime.now().strftime('%Y%m%d_%H%M%S')+'_gtMeshPoseEst'+'_obj'+obj_id)
if not os.path.exists(save_dir):
    os.mkdir(save_dir)
```

```python
iter_loader = iter(shapenet_loader)
data_loop = tqdm(range(len(iter_loader)))
# record results
pose_scl_gt = []
pose_scl_pred = []
labels = []
count_models = 0
total_time = 0
for i in data_loop:
    meshes_i = next(iter_loader) # [DONE] exclude weird shapes
    meshes_i = exclude_weird_shapes(meshes_i)
    bs_actual = len(meshes_i['model_id'])
    if bs_actual == 0:
        continue
    assert(bs_actual != 0)
    # if i == len(iter_loader) - 1:
    #     bs_actual = len(meshes_i['model_id']) # last iter may have a different bs
    faces_i = meshes_i['mesh'].faces_padded().to(device)
    verts_padded = meshes_i['mesh'].verts_padded().to(device) # [bs,n,3] padded with 0
    # find axis-aligned bounding box in canonical pose
    cano_bboxes = extract_cano_mesh_bbx(meshes_i['mesh'].to(device)).to(device) # [bs,8,3]
    # get gt and initial poses and target meshes
    ((random_rots, random_trsl, random_scls), (init_rots, init_trsl, init_scls),
    (trg_meshes, trg_bboxes)) = get_gt_pose_init_pose(bs_actual, device,
                                                      verts_padded, faces_i, cano_bboxes,
                                                      init_rot_std, init_t_std, init_scl_std, init_scl_lbd)
    est_rots, est_trsl, est_scls = init_rots, init_trsl, init_scls
    # apply transformation to vertices and then sample - inefficient
    # init_verts = apply_trsfms_to_pts(verts_padded, init_rots, init_trsl, init_scls)
    # src_meshes = pytorch3d.structures.Meshes(verts=init_verts, faces=faces_i).to(device)
    # apply transformation directly to samples
    src_meshes_0 = pytorch3d.structures.Meshes(verts=verts_padded, faces=faces_i).to(device)
    samples_src_0 = pytorch3d.ops.sample_points_from_meshes(src_meshes_0, n_sample_pts) # [bs,n_samples,3]
    samples_src = apply_trsfms_to_pts(samples_src_0, init_rots, init_trsl, init_scls)
    # solver
    # sample on the target mesh
    samples_trg = pytorch3d.ops.sample_points_from_meshes(trg_meshes, n_sample_pts) # [bs,n_samples,3]
    # set up optimizer if needed
    # if mode2 == 0:
    #     optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
    #     optimizer = torch.optim.SGD([optim_scls], lr=lrate, momentum=0.9)
    # if mode2 == 1:
    #     optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
    #     optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True)
    #     optimizer = torch.optim.SGD([optim_scls, optim_trsl], lr=lrate, momentum=0.9)
    # if mode2 == 2:
    #     optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
    #     optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
    #     optim_rots_so3 = torch.zeros_like(init_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
    #     optimizer = torch.optim.SGD([optim_scls, optim_trsl, optim_rots_so3], lr=lrate, momentum=0.9)
    # optimization loop
    start = timeit.default_timer()
    for j in range(n_iters):
        # sample on the source mesh - inefficient
        # samples_src = pytorch3d.ops.sample_points_from_meshes(src_meshes, n_sample_pts) # [bs,n_samples,3]
        sim_trsfm = None
        # knn_umeyama or icp
        if mode1 == 0:
            sim_trsfm = knn_umeyama(samples_trg, samples_src, est_scale=True)
        if mode1 == 1:
            sim_trsfm = icp(samples_trg, samples_src, est_scale=True)
        # accumulate increments on estimates if mode1 is run
        if sim_trsfm:
            est_scls = est_scls * sim_trsfm[2] # [bs]
            est_rots = sim_trsfm[0].transpose(-1,-2).matmul(est_rots) # [bs,3,3]
            est_trsl = sim_trsfm[0].transpose(-1,-2).matmul( # [bs,3,3]
                       (sim_trsfm[2].unsqueeze(-1).unsqueeze(-1) * est_trsl)) + sim_trsfm[1].unsqueeze(-1) # [bs,3,1]
        # set up optimizer
        if mode2 == 0:
            optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
            optimizer = torch.optim.SGD([optim_scls], lr=lrate, momentum=0.9)
        if mode2 == 1:
            optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
            optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True)
            optimizer = torch.optim.SGD([optim_scls, optim_trsl], lr=lrate, momentum=0.9)
        if mode2 == 2:
            optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
            optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
            optim_rots_so3 = torch.zeros_like(init_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
            optimizer = torch.optim.SGD([optim_scls, optim_trsl, optim_rots_so3], lr=lrate, momentum=0.9)
        if mode2 == 0 or mode2 == 1 or mode2 == 2:
            for k in range(mode2_iters):
                optimizer.zero_grad()
                # with torch.no_grad():
                #     optim_scls.zero_()
                #     optim_scls += 1.0
                #     optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
                #     optim_scls /= optim_scls.detach()
                # accumulate increments on estimates
                # est_scls_acc = est_scls * optim_scls # [bs]
                # alternative - keep updating the optimization variables
                est_scls_acc = optim_scls * est_scls # [bs]
                est_trsl_acc = est_trsl
                est_rots_acc = est_rots
                if mode2 == 1 or mode2 == 2:
                    # with torch.no_grad():
                    #     optim_trsl.zero_()
                    #     optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True)
                    #     optim_trsl -= optim_trsl.detach()
                    # est_trsl_acc = est_trsl + optim_trsl # [bs,3,1]
                    # alternative - keep updating the optimization variables
                    est_trsl_acc = optim_trsl + est_trsl # [bs,3,1]
                    if mode2 == 2:
                        # with torch.no_grad():
                        #     optim_rots_so3.zero_() # [bs,3]
                        #     optim_rots_so3 = torch.zeros_like(init_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True)
                        # est_rots_acc = pytorch3d.transforms.so3_exp_map(optim_rots_so3).matmul(est_rots) # [bs,3,3]
                        # alternative - keep updating the optimization variables
                        est_rots_acc = pytorch3d.transforms.so3_exp_map(optim_rots_so3).matmul(est_rots) # [bs,3,3]
                # apply transformation to vertices and then sample from mesh - inefficient
                # est_verts = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, est_scls)
                # src_meshes = pytorch3d.structures.Meshes(verts=est_verts, faces=faces_i).to(device)
                # samples_src = pytorch3d.ops.sample_points_from_meshes(src_meshes, n_sample_pts) # [bs,n_samples,3] resample needed if shape changed
                # apply transformation directly to samples
                samples_src = apply_trsfms_to_pts(samples_src_0, est_rots_acc, est_trsl_acc, est_scls_acc)
                src_bboxes = apply_trsfms_to_pts(cano_bboxes, est_rots_acc, est_trsl_acc, est_scls_acc)
                losses = iou_chamfer_loss(samples_trg, samples_src, bbox1=trg_bboxes, bbox2=src_bboxes, iou=False, chamfer=True)
                loss = losses[0] * iou_wt + losses[1] * chamfer_wt
                loss.backward()
                optimizer.step()
                # accumulate increments on estimates - pytorch seems not able to zero the variables without reinit the optimizer
                # est_scls *= optim_scls.detach()
                # if mode2 == 1 or mode2 == 2:
                #     est_trsl += optim_trsl.detach()
                #     if mode2 == 2:
                #         est_rots  = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
            # alternative - compose the estimates with the optimized quantities 
            est_scls *= optim_scls.detach()
            if mode2 == 1 or mode2 == 2:
                est_trsl += optim_trsl.detach()
                if mode2 == 2:
                    est_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
        # apply transformation to vertices and then sample from mesh - inefficient
        # est_verts = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, est_scls)
        # src_meshes = pytorch3d.structures.Meshes(verts=est_verts, faces=faces_i).to(device)
        # apply transformation directly to samples
        samples_src = apply_trsfms_to_pts(samples_src_0, est_rots, est_trsl, est_scls)
    # end of optimization loop
    total_time += (timeit.default_timer() - start)
    # [DONE] solve the inefficiency by swaping out meshes_i['faces'] which is a list
    # [DONE] add iou and chamfer loss weights
    # [DONE] Test point cloud transformation and transformation of vertices + sampling speed conparison
    # record results
    labels.append(meshes_i['label']) # bs
    pose_scl_gt.append((random_rots.detach().cpu().numpy(), # [bs,3,3]
                        random_trsl.detach().cpu().numpy(), # [bs,3,1]
                        random_scls.detach().cpu().numpy())) # [bs]
    pose_scl_pred.append((est_rots.detach().cpu().numpy(),
                          est_trsl.detach().cpu().numpy(),
                          est_scls.detach().cpu().numpy()))
    count_models += bs_actual
    if (i + 1) % save_freq == 0:
        # save the first one in a batch
        pytorch3d.io.IO().save_mesh(trg_meshes[0], os.path.join(save_dir, 'iter_{:d}_trg.ply'.format(i + 1)))
        src_verts_est = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, est_scls)
        src_meshes_est = pytorch3d.structures.Meshes(verts=src_verts_est, faces=faces_i)
        pytorch3d.io.IO().save_mesh(src_meshes_est[0], os.path.join(save_dir, 'iter_{:d}_src.ply'.format(i + 1)))
        init_verts = apply_trsfms_to_pts(verts_padded, init_rots, init_trsl, init_scls)
        init_meshes = pytorch3d.structures.Meshes(verts=init_verts, faces=faces_i)
        pytorch3d.io.IO().save_mesh(init_meshes[0], os.path.join(save_dir, 'iter_{:d}_init.ply'.format(i + 1)))
    # release memory
    del meshes_i, faces_i, verts_padded, trg_meshes, src_meshes_0, samples_src_0, samples_src, samples_trg
    torch.cuda.empty_cache()
    gc.collect() # does not help with memory exhaustion
# end of data loop
```

```python
class setCameras():
    def __init__(self, img_sz, bs, dev, input_camK=False, cam_K=None):
        # cam_K: [3,3] numpy assume same across a batch
        if type(img_sz) == tuple:
            min_imsz = min(img_sz)
            imw = img_sz[1]
            imh = img_sz[0]
        else: # int
            min_imsz, imw, imh = img_sz, img_sz, img_sz
        # get camera intrinsics
        # [DONE] if cam_K is inputted, how to scale focal_len to the normalized space maybe by img_size / 2
        if not input_camK:
            cam_K = torch.eye(3, dtype=torch.float32, device=dev).unsqueeze(0).expand(bs, -1, -1).clone()
        else:
            # transform a pytorch screen coord intrinsics to the ndc space intrinsics
            # but due to pytorch3d camera axis directions the one below gives different image coverage in the space
            # use the next one and flip the resulting rendered image left-right up-down
            cam_K_np = np.copy(cam_K)
            cam_K_np_cp = np.copy(cam_K_np)
            # cam_K_np[...,0,0] = cam_K_np_cp[...,0,0] * 2.0 / min_imsz
            # cam_K_np[...,1,1] = cam_K_np_cp[...,1,1] * 2.0 / min_imsz
            # cam_K_np[...,0,2] = (imw / 2.0 - cam_K_np_cp[...,0,2]) * 2.0 / min_imsz
            # cam_K_np[...,1,2] = (imh / 2.0 - cam_K_np_cp[...,1,2]) * 2.0 / min_imsz
            cam_K_np[...,0,0] = cam_K_np_cp[...,0,0] * 2.0 / min_imsz
            cam_K_np[...,1,1] = cam_K_np_cp[...,1,1] * 2.0 / min_imsz
            cam_K_np[...,0,2] = (cam_K_np_cp[...,0,2] - imw / 2.0) * 2.0 / min_imsz
            cam_K_np[...,1,2] = (cam_K_np_cp[...,1,2] - imh / 2.0) * 2.0 / min_imsz
            cam_K = torch.tensor(cam_K_np, dtype=torch.float32, device=dev).unsqueeze(0).expand(bs, -1, -1).clone()
        focal_len = cam_K[..., [0, 1], [0, 1]]
        pcp_point = cam_K[..., :2, 2]
        # set up cameras
        img_sz_torch = torch.tensor(img_sz, dtype=torch.float32, device=dev).unsqueeze(0).expand(bs, -1).clone()
        cameras = pytorch3d.renderer.cameras.PerspectiveCameras(focal_length=focal_len, principal_point=pcp_point,
                                                                image_size=img_sz_torch, in_ndc=True, device=dev)
        # return the camera intrinsics without pytorch3d normalization but the pcp point is still w.r.t. img center (not topleft corner) and with pytorch3d axis direction
        # using this cam_intr and meshgrid_abs(pyt3dformat=True) below can backproject rendered depth image as usual
        zero_fill = torch.zeros_like(cameras.focal_length[:,0], dtype=torch.float32)
        one_fill = torch.ones_like(cameras.focal_length[:,0], dtype=torch.float32)
        cam_intr = torch.stack([torch.stack([cameras.focal_length[:,0]*min_imsz/2.0, zero_fill.clone(), cameras.principal_point[:,0]*min_imsz/2.0], dim=-1),
                                torch.stack([zero_fill.clone(), cameras.focal_length[:,1]*min_imsz/2.0, cameras.principal_point[:,1]*min_imsz/2.0], dim=-1),
                                torch.stack([zero_fill.clone(), zero_fill.clone(), one_fill.clone()], dim=-1)], dim=-2).to(device=dev)
        self.cameras = cameras
        self.cam_intr = cam_intr # [bs,3,3]
        self.inv_intr = torch.inverse(cam_intr[0]).unsqueeze(0).expand(bs,-1,-1).clone() # assume same cam intrinsics
        self.bs = bs # used to check if the same bs
        self.dev = dev
    
    def zero_camera_pose(self):
        self.cameras.R = torch.eye(3, dtype=torch.float32, device=self.dev).unsqueeze(0).expand(self.bs, -1, -1).clone() # cam is world pose_c_w = identity
        self.cameras.T = torch.zeros((self.bs, 3), dtype=torch.float32, device=self.dev)
    
    def cameras_pointing_obj(self, t_w_o):
        # use the object position to get a camera at the origin to point at the object
        # the camera x-y in-plane rotation is arbitrary
        # t_w_o: [bs,3,1] object position in world frame
        dev = self.dev
        bs = self.bs
        rerun = True
        # compute camera pose
        t_w_o = t_w_o.squeeze(-1)
        zaxis_w = (t_w_o / torch.norm(t_w_o, dim=-1, keepdim=True))
        while rerun:
            random_u = torch.rand((bs, 3), dtype=torch.float32, device=dev)
            random_u = random_u / torch.norm(random_u, dim=-1, keepdim=True)
            xaxis_w = torch.cross(zaxis_w, random_u, dim=-1)
            xaxis_w = xaxis_w / torch.norm(xaxis_w, dim=-1, keepdim=True) # need normalize x since random_u is not perpendicular to z
            yaxis_w = torch.cross(zaxis_w, xaxis_w, dim=-1)
            yaxis_w = yaxis_w / torch.norm(yaxis_w, dim=-1, keepdim=True) # not needed but for safety
            rot_w_c = torch.stack((xaxis_w, yaxis_w, zaxis_w), dim=-1) # [bs,3,3]
            valid_ortho = torch.allclose(rot_w_c.bmm(rot_w_c.transpose(-1,-2)),
                                         torch.eye(3, dtype=torch.float32, device=dev).view(1,3,3).expand(bs,-1,-1),
                                         atol=1e-5) # 1e-5 from pytorch3d.transforms.transform3d.py/orthogonal_tol
            if valid_ortho:
                rerun = False
            else:
                print(datetime.datetime.now().strftime('%Y%m%d_%H%M%S'))
                print('rotation not valid')
        t_c_w = torch.zeros((bs, 3), dtype=torch.float32, device=dev)
        # set up camera poses
        self.cameras.R = rot_w_c # should be world-in-cam (rot_c_w) but rotation is transposed
        self.cameras.T = t_c_w
        self.rot_w_c = rot_w_c # [bs,3,3]

class rasterization():
    def __init__(self, img_sz):
        # img_sz: tuple or int
        raster_settings = pytorch3d.renderer.mesh.rasterizer.RasterizationSettings(image_size=img_sz, 
                                                                  blur_radius=0.0, faces_per_pixel=1)
        self.rasterizer = pytorch3d.renderer.mesh.rasterizer.MeshRasterizer(cameras=None, raster_settings=raster_settings)
    def rasterize(self, mesh_to_render, cameras):
        self.rasterizer.cameras = cameras
        fragments = self.rasterizer(mesh_to_render)
        return fragments.zbuf # [bs,h,w,1]

class processData():
# code from shape_fitting notebook
    def __init__(self, img_h, img_w, dev, pyt3dformat=False):
        self.device = dev
        self.dtype = torch.float32
        self.img_h = img_h
        self.img_w = img_w
        self.img_grid = self.meshgrid_abs(self.img_h, self.img_w, pyt3dformat)
        if img_h <= 10:
            print(self.img_grid)
    
    def meshgrid_abs(self, height, width, pyt3drender=False):
        # height and width: int
        # migrated from my VO_NN/selfsemantics/project.py
        if pyt3drender:
            start, end = 1.0, -1.0 # NOTE the pytorch3d convention x+ left y+ up
            center = 0.0
        else:
            start, end = -1.0, 1.0
            center = 1.0
        x_t = torch.matmul(
            torch.ones((height, 1), device=self.device, dtype=self.dtype, requires_grad=False),
            torch.transpose(torch.unsqueeze(
            torch.linspace(start, end, width, device=self.device, dtype=self.dtype, requires_grad=False),
            dim=1), dim0=0, dim1=1))
        y_t = torch.matmul(torch.unsqueeze(
            torch.linspace(start, end, height, device=self.device, dtype=self.dtype, requires_grad=False),
            dim=1), torch.ones((1, width), device=self.device, dtype=self.dtype, requires_grad=False))
        x_t = (x_t + center) * 0.5 * torch.tensor((width - 1), dtype=self.dtype) # + 0.5 if offset by 0.5 for the pixel center
        y_t = (y_t + center) * 0.5 * torch.tensor((height - 1), dtype=self.dtype) # w/o 0.5 seems better for pytorch3d rasterizer
        x_t_flat = torch.reshape(x_t, (1, -1)) # (1,h*w)
        y_t_flat = torch.reshape(y_t, (1, -1))
        ones = torch.ones_like(x_t_flat)
        grid = torch.cat((x_t_flat, y_t_flat, ones), dim=0) # (3,h*w)
        return grid
    
    @staticmethod
    def save_point_cloud(ptc_to_save, ptc_name='savedPointCloud.ply'):
        # save a point cloud ptc_to_save (...,3) torch tensor
        ptc_to_save = torch.reshape(ptc_to_save, (-1, 3))
        ptc = open3d.geometry.PointCloud()
        ptc.points = open3d.utility.Vector3dVector(ptc_to_save.detach().cpu().numpy())
        open3d.io.write_point_cloud(ptc_name, ptc)
        # pyt3d_ptc = pytorch3d.structures.Pointclouds(ptc_to_save.unsqueeze(0))
        # pytorch3d.io.IO().save_pointcloud(pyt3d_ptc[0], ptc_name[:-4]+'2.ply')
    
    def set_data(self, depth_img, inv_intrinsics, rotmats, tvectors):
        # set data to class attributes. Assume pytorch tensors.
        # rotmats and tvectors are camera-in-world
        self.depths = depth_img # [b,h,w,1] (not [b,h,w] may cause masking in project_to_3d to fail)
        self.inv_intrinsics = inv_intrinsics # [b,3,3]
        self.rotmats = rotmats # [b,3,3]
        self.tvectors = tvectors # [b,3]
    
    def project_to_3d(self, masking=True):
        # masking: project only valid depth points. This makes a batch has different point cloud lengths.
        # project pixels to 3D
        dims = list(self.depths.shape)
        batch_size, img_height, img_width = dims[0], dims[1], dims[2]
        depths = torch.reshape(self.depths, (batch_size, 1, img_height * img_width))
        # grid = torch.tile(torch.unsqueeze(self.img_grid, dim=0), (batch_size, 1, 1)) # (b,3,h*w) tile not existing in early torch versions
        grid = self.img_grid.unsqueeze(0).expand(batch_size, -1, -1).clone()
        cam_coords = torch.matmul(self.inv_intrinsics, grid) * depths # (b,3,h*w)
        # transform to world frame
        world_points = torch.matmul(self.rotmats, cam_coords) # (b,3,h*w)
        # tiled_tvectors = torch.tile(torch.unsqueeze(self.tvectors, dim=-1), (1,1,img_height*img_width)) # tile not existing in early torch versions
        tiled_tvectors = self.tvectors.unsqueeze(-1).expand(-1, -1, img_height*img_width).clone()
        world_points += tiled_tvectors
        world_points = torch.transpose(world_points, dim0=-1, dim1=-2) # (b,h*w,3)
        world_points = torch.reshape(world_points, (batch_size, img_height, img_width, 3))
        self.world_points = world_points # (b,h,w,3)
        if masking:
            mask = (self.depths > 0.0) # [b,h,w,1]
            self.world_points *= mask.to(torch.float32)
            # comment out below since not sure masked_select returns in the order of batch elements (should be)
            # self.unmasked_w_points = torch.reshape(torch.masked_select(self.world_points, mask), (-1, 3)) # (n,3)
            # batch_lengths = mask.squeeze(-1).sum(dim=(-1,-2)) # [bs]
            # batched_unmasked_pts = self.unmasked_w_points.split(batch_length.tolist(), dim=0) # aka structures.packed_to_list
            batched_unmasked_pts = [torch.masked_select(self.world_points[b], mask[b]).reshape(-1, 3) for b in range(batch_size)]
            self.unmasked_pts = pytorch3d.structures.Pointclouds(batched_unmasked_pts)
            
def normalize_center_vertices(vertices, normalizing=True, centering=True, bbox_center=False):
    # vertices: [bs,n,3] assuming vertices are not padded with zero because that may affect centroid and diagonal computation
    # following shapenet normalization. vNorm = (v - c) / norm, where c is mean of all vertices and norm is the norm of diagonal (max(x,y,z) - min(x,y,z))
    # bbox_center: instead of centering at vertices center, center at bbox center
    max_corner = torch.max(vertices, dim=-2, keepdim=True)[0] # [bs,1,3] [0] to pick max values in the tuple
    min_corner = torch.min(vertices, dim=-2, keepdim=True)[0] # [bs,1,3]
    if bbox_center:
        ctr = (max_corner + min_corner) / 2.0 # [bs,1,3]
    else:
        ctr = torch.mean(vertices, dim=-2, keepdim=True) # [bs,1,3]
    norm = torch.norm(max_corner - min_corner, dim=-1, keepdim=True) # [bs,1,1]
    if centering:
        updated_vertices = vertices - ctr
        if normalizing:
            updated_vertices /= norm
    if normalizing and not centering:
        updated_vertices = vertices / norm
    return updated_vertices.detach(), norm.detach(), ctr.detach() # [bs,n,3], [bs,1,1], [bs,1,3]
```

```python
iter_loader = iter(shapenet_loader)
data_loop = tqdm(range(len(iter_loader)))
nmb_batches = 0
# record results
pose_scl_gt = []
pose_scl_pred = []
shape_gt = []
shape_pred = []
posed_shape_gt = []
posed_shape_pred = []
labels = []
count_models = 0
total_time = 0
for i in data_loop:

#     if i == 3: # TESTING use 3 first 3 are jumped due to weird shapes for bowl
#         print('paused')
#         time.sleep(3600)
    
    meshes_i = next(iter_loader) # [DONE] exclude weird shapes
    meshes_i = exclude_weird_shapes(meshes_i)
    bs_actual = len(meshes_i['model_id'])
    if bs_actual == 0:
        continue
    assert(bs_actual != 0)
    # if i == len(iter_loader) - 1:
    #     bs_actual = len(meshes_i['model_id']) # last iter may have a different bs
    faces_i = meshes_i['mesh'].faces_padded().to(device)
    verts_padded = meshes_i['mesh'].verts_padded().to(device) # [bs,n,3] padded with 0
    # find axis-aligned bounding box in canonical pose
    cano_bboxes = extract_cano_mesh_bbx(meshes_i['mesh'].to(device)).to(device) # [bs,8,3]
    # get gt and initial poses and target meshes
    ((random_rots, random_trsl, random_scls), (init_rots, init_trsl, init_scls),
    (trg_meshes, trg_bboxes)) = get_gt_pose_init_pose(bs_actual, device,
                                                      verts_padded, faces_i, cano_bboxes,
                                                      init_rot_std, init_t_std, init_scl_std, init_scl_lbd)
    est_rots, est_trsl, est_scls = init_rots, init_trsl, init_scls
    # sample on the target mesh
    if not rendered_depth:
        samples_trg = pytorch3d.ops.sample_points_from_meshes(trg_meshes, n_sample_pts) # [bs,n_samples,3]
    else:
        if cameras_setup.bs != bs_actual:
            cameras_setup = setCameras(img_size, bs_actual, device)
        cameras_setup.cameras_pointing_obj(random_trsl)
        depth_img = rasterizer.rasterize(trg_meshes, cameras_setup.cameras).detach() # [bs,h,w,1]
        data_proc.set_data(depth_img, cameras_setup.inv_intr, cameras_setup.rot_w_c, cameras_setup.cameras.get_camera_center())
        data_proc.project_to_3d(masking=True)
        samples_trg = data_proc.unmasked_pts # structures.Pointcloud obj
    # solver
    # set up optimizer for pose and scale if needed
    # if mode2 == 0:
    #     optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
    #     optimizer = torch.optim.SGD([optim_scls], lr=lr_pose_scl, momentum=0.9)
    # if mode2 == 1:
    #     optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
    #     optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True)
    #     optimizer = torch.optim.SGD([optim_scls, optim_trsl], lr=lr_pose_scl, momentum=0.9)
    # if mode2 == 2:
    #     optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
    #     optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
    #     optim_rots_so3 = torch.zeros_like(init_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
    #     optimizer = torch.optim.SGD([optim_scls, optim_trsl, optim_rots_so3], lr=lr_pose_scl, momentum=0.9)
    # init pca basis shape
    if shape_init == 'mean':
        pca_est_params = torch.full((bs_actual, K), 0.0, device=device, requires_grad=True).to(torch.float32) # pca_mean is added in init 
    if shape_init == 'avg':
        pca_est_params = torch.mean(pca_basis_params_torch, dim=0, keepdim=True).expand(bs_actual, -1).clone().requires_grad_() # -1 not change size = K. clone() since expand refers to single memory
    batch_faces = uni_faces.expand(bs_actual, -1, -1)
    verts_init = torch.matmul(pca_est_params, pca_bases) + pca_mean # used for saving results
    # compute initial guess
    if rendered_depth and compute_init_guess:
        start = timeit.default_timer()
        # init_rots_guess, init_trsl, init_scls = pose_initer.initialize_old(samples_trg, depth_img, verts_init.reshape(bs_actual,-1,3),
        #     uni_faces, cameras_setup.cameras.R.transpose(-1,-2), cameras_setup.cameras.T, cameras_setup.cam_intr, bs_actual,
        #     pyt3d_img_coords=True, n_guess_rot=n_rot_guesses, n_samples=n_sample_pts, verts_as_samples=verts_as_sppts)
        init_rots_guess, init_trsl, init_scls = pose_initer.initialize(samples_trg, verts_init.reshape(bs_actual,-1,3),
            uni_faces, bs_actual, verts_as_samples=verts_as_sppts, n_samples=n_sample_pts, n_guess_rot=1)
        total_time += (timeit.default_timer() - start)
        if not use_good_rot_init:
            init_rots = init_rots_guess
        est_rots, est_trsl, est_scls = init_rots, init_trsl, init_scls
    # initialize em gaussian covariance
    if em and rendered_depth:
        sigma_sq_inv = sigma_sq_inv_init * torch.ones(samples_trg.points_padded()[:,:,0].shape, device=device, dtype=torch.float32) # [bs,n]
    else:
        sigma_sq_inv = None
    # apply transformation to vertices and then sample - inefficient - deprecated - no use now
    # init_verts = apply_trsfms_to_pts(verts_padded, init_rots, init_trsl, init_scls)
    # src_meshes = pytorch3d.structures.Meshes(verts=init_verts, faces=faces_i).to(device)
    # apply transformation directly to samples
    # src_meshes_0 = pytorch3d.structures.Meshes(verts=verts_padded, faces=faces_i).to(device)
    # samples_src_0 = pytorch3d.ops.sample_points_from_meshes(src_meshes_0, n_sample_pts) # [bs,n_samples,3]
    # samples_src = apply_trsfms_to_pts(samples_src_0, init_rots, init_trsl, init_scls)
    # shape optimization set-up
    if shape_mode == 0:
        optimizer_shape = torch.optim.SGD([pca_est_params], lr=lr_shape_m0, momentum=mm_shape_m0)
    if shape_mode == 1:
        optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
        optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
        optim_rots_so3 = torch.zeros_like(init_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
        optimizer_shape = torch.optim.SGD([pca_est_params, optim_scls, optim_trsl, optim_rots_so3], lr=lr_shape_m1, momentum=mm_shape_m1)
    nmb_batches += 1 # put here so that if it stops at the rendering, this does not increment (for ransac stats)
    # optimization loop
    start = timeit.default_timer()
    for j in range(n_iters):
        
#         print('OPTIM ITER', j) # TESTING
        
        # create meshes from bases and transform to estimated pose and scale
        new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # dim broadcastable
        new_verts_0 = torch.reshape(new_verts_0, (bs_actual, -1, 3))
        if normalize_for_pose or center_for_pose:
            new_verts_0, _, _ = normalize_center_vertices(new_verts_0, normalize_for_pose, center_for_pose) # [bs,n,3], [bs,1,1], [bs,1,3]
        if mode1 == 0 or mode1 == 1:
            new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
            # sample on the source mesh - inefficient
            if verts_as_sppts:
                samples_src = new_verts
            else:
                recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
                samples_src = pytorch3d.ops.sample_points_from_meshes(recon_meshes, n_sample_pts) # [bs,n_samples,3]
        # pose estimation
        sim_trsfm = None
        # knn_umeyama or icp
        if mode1 == 0:
            for aln_i in range(mode1_iters):
                # run umeyama
                if not rendered_depth:
                    sim_trsfm = knn_umeyama(samples_trg, samples_src, est_scale=True)
                else:
                    if not em:
                        if ransac_for_pose:
                            sim_trsfm, ransac_stats = knn_umeyama_hetero_ransac_s(samples_trg, samples_src, True,
                                                                    ransac_n, ransac_th, ransac_inrate, ransac_maxiter)
                            # record ransac stats
                            if ransac_stats[-1]: # batch pass
                                ransac_meta_stats['passing_residual'][j] += ransac_stats[0].detach().cpu().numpy() # a batch result in iter j
                                ransac_meta_stats['passing_inlier_rate'][j] += ransac_stats[1].detach().cpu().numpy()
                                ransac_meta_stats['nmb_passes'][j] += 1
                            ransac_meta_stats['iters_mean_residual'][j] += ransac_stats[2].detach().cpu().numpy()
                            ransac_meta_stats['iters_std_residual'][j] += ransac_stats[3].detach().cpu().numpy()
                            ransac_meta_stats['iters_mean_inlier_ratio'][j] += ransac_stats[4].detach().cpu().numpy()
                            ransac_meta_stats['iters_std_inlier_ratio'][j] += ransac_stats[5].detach().cpu().numpy()
                            ransac_meta_stats['final_nmb_iters'][j] += ransac_stats[6]
                        else:
                            sim_trsfm = knn_umeyama_hetero(samples_trg, samples_src, est_scale=True, teaser=teaserpp_pose, teaserpp_solver=teapp_solver)
                    else:
                        if ransac_for_pose:
                            sim_trsfm, sigma_sq_inv = knn_umeyama_hetero_em_ransac_s(samples_trg, samples_src, True,
                                                                                  (j > 0 or aln_i > 0) and not fix_sigma, knn_k,
                                                                                  sigma_sq_inv, bs_actual, 3, ransac_n, ransac_th,
                                                                                  ransac_inrate, ransac_maxiter, inlier_check_nn)
                        else:
                            sim_trsfm, sigma_sq_inv = knn_umeyama_hetero_em(samples_trg, samples_src, est_scale=True,
                                                                         update_sigma=(j > 0 or aln_i > 0) and not fix_sigma, knn_k=knn_k,
                                                                         sigma_sq_inv=sigma_sq_inv, bs=bs_actual, deg=3)
                if mode1_iters > 1: # if == 1, do this after mode1==1
                    # [DONE] Possibly wrong. Causing cuda runtime error. [20230805] simple bug? I was updating with est_scls not sim_trsfm[2]
                    est_scls = est_scls * sim_trsfm[2].detach() # [bs] accumuate increments
                    est_rots = sim_trsfm[0].detach().transpose(-1,-2).matmul(est_rots) # [bs,3,3]
                    est_trsl = sim_trsfm[0].detach().transpose(-1,-2).matmul( # [bs,3,3]
                       (sim_trsfm[2].detach().unsqueeze(-1).unsqueeze(-1) * est_trsl)) + sim_trsfm[1].detach().unsqueeze(-1) # [bs,3,1]
                    samples_src = apply_trsfms_to_pts(samples_src, sim_trsfm[0].detach().transpose(-1,-2),
                                                      sim_trsfm[1].detach().unsqueeze(-1), sim_trsfm[2].detach()) # transform with the increments
                    sim_trsfm = None
        if mode1 == 1:
            sim_trsfm = icp(samples_trg, samples_src, est_scale=True)
        # accumulate increments on estimates if mode1 is run (double checked 20230807)
        if sim_trsfm:
            est_scls = est_scls * sim_trsfm[2].detach() # [bs]
            est_rots = sim_trsfm[0].detach().transpose(-1,-2).matmul(est_rots) # [bs,3,3]
            est_trsl = sim_trsfm[0].detach().transpose(-1,-2).matmul( # [bs,3,3]
                       (sim_trsfm[2].detach().unsqueeze(-1).unsqueeze(-1) * est_trsl)) + sim_trsfm[1].detach().unsqueeze(-1) # [bs,3,1]
            samples_src = apply_trsfms_to_pts(samples_src, sim_trsfm[0].detach().transpose(-1,-2),
                                              sim_trsfm[1].detach().unsqueeze(-1), sim_trsfm[2].detach()) # transform with the increments
            sim_trsfm = None
        # mode2 - set up optimizer
        if mode2 == 0:
            optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
            optimizer = torch.optim.SGD([optim_scls], lr=lr_pose_scl, momentum=0.9)
        if mode2 == 1:
            optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True)
            optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True)
            optimizer = torch.optim.SGD([optim_scls, optim_trsl], lr=lr_pose_scl, momentum=0.9)
        if mode2 == 2:
            optim_scls = torch.ones_like(init_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
            optim_trsl = torch.zeros_like(init_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
            optim_rots_so3 = torch.zeros_like(init_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
            optimizer = torch.optim.SGD([optim_scls, optim_trsl, optim_rots_so3], lr=lr_pose_scl, momentum=0.9)
        if mode2 == 0 or mode2 == 1 or mode2 == 2:
            # samples_src_start = samples_src.detach().clone() # added 20230807 for applying increments on the transformed points
            if verts_as_sppts:
                samples_src_start = new_verts_0.detach().clone()
            else:
                recon_meshes_0 = pytorch3d.structures.Meshes(verts=new_verts_0, faces=batch_faces) # for applying directly on R,t,s
                samples_src_start = pytorch3d.ops.sample_points_from_meshes(recon_meshes_0, n_sample_pts)
            for k in range(mode2_iters):
                optimizer.zero_grad()
                optim_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3) # fine actaully only mode2 == 2 works
                # apply transformation to vertices and then sample from mesh - inefficient - depracated. use only after updating
                # est_verts = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, est_scls)
                # src_meshes = pytorch3d.structures.Meshes(verts=est_verts, faces=faces_i).to(device)
                # samples_src = pytorch3d.ops.sample_points_from_meshes(src_meshes, n_sample_pts) # [bs,n_samples,3] resample needed if shape changed
                # apply transformation directly to samples
                # samples_src = apply_trsfms_to_pts(samples_src_start, optim_rots, optim_trsl, optim_scls) # transform with increments see why below [WRONG][corrected 20230807] samples_src_start was samples_src which causes samples_src to be double transformed by previous estimates
                samples_src = apply_trsfms_to_pts(samples_src_start, optim_rots.matmul(est_rots), optim_trsl+est_trsl, optim_scls*est_scls)
                src_bboxes = apply_trsfms_to_pts(cano_bboxes, optim_rots, optim_trsl, optim_scls) # [WRONG] need bbox in estimated pose but iou loss not used
                if not rendered_depth:
                    losses = iou_chamfer_loss(samples_trg, samples_src, bbox1=trg_bboxes, bbox2=src_bboxes, iou=False, chamfer=True)
                else:
                    losses = iou_chamfer_loss(samples_trg.points_padded(), samples_src, len1=samples_trg.num_points_per_cloud(),
                                              bbox1=trg_bboxes, bbox2=src_bboxes, iou=False, chamfer=True)
                loss = losses[0] * iou_wt + losses[1] * chamfer_wt
                loss.backward()
                optimizer.step()
                # accumulate increments on estimates - pytorch seems not able to zero the variables without reinit the optimizer
                # est_scls *= optim_scls.detach()
                # if mode2 == 1 or mode2 == 2:
                #     est_trsl += optim_trsl.detach()
                #     if mode2 == 2:
                #         est_rots  = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
            # alternative - compose the estimates with the optimized quantities 
            est_scls *= optim_scls.detach()
            if mode2 == 1 or mode2 == 2:
                est_trsl += optim_trsl.detach() 
                # [WRONG][corrected 20230807] was est_trsl+=optim_trsl. this is different than the pose only estimation where I increment directly on R,t,s. Here increment is applied on the transformed verts.
                # not directly increment on R,t,s (as in pose-only estimation) because I want to reuse samples_src which is for incremental transformation estimation in mode1 for code succinctness
                # est_trsl = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(optim_scls.detach().unsqueeze(-1).unsqueeze(-1) * est_trsl) + optim_trsl.detach()
                if mode2 == 2:
                    est_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
        # apply transformation to vertices and then sample from mesh - inefficient
        # est_verts = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, est_scls)
        # src_meshes = pytorch3d.structures.Meshes(verts=est_verts, faces=faces_i).to(device)
        # apply transformation directly to samples
        # samples_src = apply_trsfms_to_pts(samples_src_0, est_rots, est_trsl, est_scls)
        
        
#         for errb in range(samples_trg.points_padded().shape[0]): # TESTING
#             pytorch3d.io.IO().save_pointcloud(samples_trg[errb], './teaser_debug/error_trg1_{:03d}{:03d}.ply'.format(j,errb))
#             pytorch3d.io.IO().save_pointcloud(pytorch3d.structures.Pointclouds(samples_src)[errb], './teaser_debug/error_src1_{:03d}{:03d}.ply'.format(j,errb))
#             # pytorch3d.io.IO().save_pointcloud(trg_points_normal_ctr[errb], './teaser_debug/error_trg1_{:03d}{:03d}.ply'.format(j,errb))
#             new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # dim broadcastable
#             new_verts_0 = torch.reshape(new_verts_0, (bs_actual, -1, 3))
#             new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
#             recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
#             pytorch3d.io.IO().save_mesh(recon_meshes[errb], './teaser_debug/error_src1m_{:03d}{:03d}.ply'.format(j,errb))

        
        if (j+1) >= shape_early_stop_iter:
            continue
        # shape estimation
        for k_sh in range(n_shape_iters):
            optimizer_shape.zero_grad()
            new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # dim broadcastable
            new_verts_0 = torch.reshape(new_verts_0, (bs_actual, -1, 3))
            if shape_mode == 0:
                new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
            if shape_mode == 1:
                optim_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3)
                new_verts = apply_trsfms_to_pts(new_verts_0, optim_rots.matmul(est_rots), optim_trsl+est_trsl, optim_scls*est_scls) # 20230807 to increment on R,t,s directly
            recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
            if not rendered_depth:
                [loss_chamfer, loss_edge, loss_normal, loss_laplacian, loss_ptedge, loss_ptface] \
                    = mesh_deform_loss(recon_meshes, samples_trg, n_samples_trg=n_sample_pts,
                                       use_temp_verts=verts_as_sppts, n_samples_temp=n_sample_pts,
                                       chamfer_on=wt_chamfer, edge_on=wt_edge, normal_on=wt_normal,
                                       laplacian_on=wt_laplacian, ptedge_on=wt_ptedge, ptface_on=wt_ptface,
                                       given_trg_samples=True)
            else:
                [loss_chamfer, loss_edge, loss_normal, loss_laplacian, loss_ptedge, loss_ptface] \
                    = mesh_deform_loss(recon_meshes, samples_trg, n_samples_trg=n_sample_pts,
                                       use_temp_verts=verts_as_sppts, n_samples_temp=n_sample_pts,
                                       chamfer_on=wt_chamfer, edge_on=wt_edge, normal_on=wt_normal,
                                       laplacian_on=wt_laplacian, ptedge_on=wt_ptedge, ptface_on=wt_ptface,
                                       given_trg_samples=True, double_chamfer=False, em_weights=em, knn_k=knn_k, sigma_sq_inv=sigma_sq_inv)
            total_loss = (wt_chamfer * loss_chamfer + wt_edge * loss_edge + \
                     wt_normal * loss_normal + wt_laplacian * loss_laplacian + \
                     wt_ptedge * loss_ptedge + wt_ptface * loss_ptface) * bs_actual
            if wt_modsink > 0:
                loss_modsink = torch.sum(torch.min(torch.sum(torch.abs(
                    pca_est_params.unsqueeze(-2) - pca_basis_params_torch.unsqueeze(0)), dim=-1), dim=-1)[0])
                if wt_modsink_schd == 0:
                    wt_modsink_ac = wt_modsink
                if wt_modsink_schd == 1:
                    wt_modsink_ac = wt_modsink / (n_iters * n_shape_iters) * (n_shape_iters * j + k_sh + 1)
                if wt_modsink_schd == 2:
                    if n_shape_iters * j + k_sh + 1 <= n_iters * n_shape_iters * wt_modsink_thpct:
                        wt_modsink_ac = 0
                    else:
                        wt_modsink_th = np.floor(n_iters * n_shape_iters * wt_modsink_thpct)
                        wt_modsink_ac = wt_modsink / (n_iters * n_shape_iters - wt_modsink_th) *\
                                                     (n_shape_iters * j + k_sh + 1 - wt_modsink_th)
                if wt_modsink_schd == 3:
                    wt_modsink_ac = wt_modsink / (n_iters * n_shape_iters)**2 * (n_shape_iters * j + k_sh + 1)**2
                total_loss += wt_modsink_ac * loss_modsink
            total_loss.backward()
            optimizer_shape.step()
        if shape_mode == 1:
            est_scls *= optim_scls.detach()
            est_trsl += optim_trsl.detach() # [WRONG][corrected 20230807 above] just like the pose step above since increments were applied on already transformed points
            est_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
    # end of optimization loop
    total_time += (timeit.default_timer() - start)
    # [DONE] solve the inefficiency by swaping out meshes_i['faces'] which is a list
    # [DONE] add iou and chamfer loss weights
    # [DONE] Test point cloud transformation and transformation of vertices + sampling speed conparison
    # record results
    labels.append(meshes_i['label']) # bs
    count_models += bs_actual
    # record shape (no pose transformation) results
    new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # dim broadcastable
    new_verts_0 = torch.reshape(new_verts_0, (bs_actual, -1, 3))
    if normalize_for_eval or center_for_eval:
        new_verts_0, norm_factor, centroid = normalize_center_vertices(new_verts_0, normalize_for_eval, center_for_eval) # [bs,n,3], [bs,1,1], [bs,1,3]
    if center_for_eval:
        est_trsl += est_rots.matmul((est_scls.unsqueeze(-1) * centroid.squeeze(-2)).unsqueeze(-1)) # [bs,3,1] t' = t + Rsc
    if normalize_for_eval:
        est_scls *= norm_factor.squeeze() # [bs] s' = s*norm
    recon_meshes_0 = pytorch3d.structures.Meshes(verts=new_verts_0, faces=batch_faces)
    gt_meshes = pytorch3d.structures.Meshes(verts=verts_padded, faces=faces_i)
    shape_gt.append(gt_meshes.detach().cpu())
    shape_pred.append(recon_meshes_0.detach().cpu())
    # record transformed shape results
    new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
    recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
    posed_shape_gt.append(trg_meshes.detach().cpu())
    posed_shape_pred.append(recon_meshes.detach().cpu())
    # record pose results
    pose_scl_gt.append((random_rots.detach().cpu().numpy(), # [bs,3,3]
                        random_trsl.detach().cpu().numpy(), # [bs,3,1]
                        random_scls.detach().cpu().numpy())) # [bs]
    pose_scl_pred.append((est_rots.detach().cpu().numpy(),
                          est_trsl.detach().cpu().numpy(),
                          est_scls.detach().cpu().numpy()))
    # save some results to disk
    if (i + 1) % save_freq == 0:
        # save the first one in a batch
        pytorch3d.io.IO().save_mesh(trg_meshes[0], os.path.join(save_dir, 'iter_{:d}_trg.ply'.format(i + 1)))
        trg_verts_src_pose = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, est_scls)
        trg_meshes_src_pose = pytorch3d.structures.Meshes(verts=trg_verts_src_pose, faces=faces_i)
        pytorch3d.io.IO().save_mesh(trg_meshes_src_pose[0], os.path.join(save_dir, 'iter_{:d}_trg_src_pose.ply'.format(i + 1)))
        trg_verts_src_pose_scl1 = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, torch.ones_like(est_scls))
        trg_meshes_src_pose_scl1 = pytorch3d.structures.Meshes(verts=trg_verts_src_pose_scl1, faces=faces_i)
        pytorch3d.io.IO().save_mesh(trg_meshes_src_pose_scl1[0], os.path.join(save_dir, 'iter_{:d}_trg_src_pose_scl1.ply'.format(i + 1)))
        # src_verts_est = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
        # src_meshes_est = pytorch3d.structures.Meshes(verts=src_verts_est, faces=batch_faces)
        pytorch3d.io.IO().save_mesh(recon_meshes[0], os.path.join(save_dir, 'iter_{:d}_src.ply'.format(i + 1)))
        pytorch3d.io.IO().save_mesh(recon_meshes_0[0], os.path.join(save_dir, 'iter_{:d}_src_shape.ply'.format(i + 1)))
        pytorch3d.io.IO().save_mesh(meshes_i['mesh'][0], os.path.join(save_dir, 'iter_{:d}_trg_shape.ply'.format(i + 1)))
        verts_init = torch.reshape(verts_init, (bs_actual, -1, 3))
        verts_init = apply_trsfms_to_pts(verts_init, init_rots, init_trsl, init_scls)
        init_meshes = pytorch3d.structures.Meshes(verts=verts_init, faces=batch_faces)
        pytorch3d.io.IO().save_mesh(init_meshes[0], os.path.join(save_dir, 'iter_{:d}_init.ply'.format(i + 1)))
        if rendered_depth:
            pytorch3d.io.IO().save_pointcloud(samples_trg[0], os.path.join(save_dir, 'iter_{:d}_trgdepth.ply'.format(i + 1)))
    # release memory
    del meshes_i, faces_i, verts_padded, trg_meshes, samples_src, samples_trg
    torch.cuda.empty_cache()
    gc.collect() # does not help with memory exhaustion
# end of data loop
```

```python
import sys
sys.path.insert(1, './third_party')
import third_party.sdfest_so3grid
import torchvision.ops

def masks_to_boxes(masks):
    n = masks.shape[0]
    bounding_boxes = torch.zeros((n, 4), device=masks.device, dtype=torch.float32)
    for index, mask in enumerate(masks):
        y, x = torch.where(mask != 0)
        bounding_boxes[index, 0] = torch.min(x)
        bounding_boxes[index, 1] = torch.min(y)
        bounding_boxes[index, 2] = torch.max(x)
        bounding_boxes[index, 3] = torch.max(y)
    return bounding_boxes

class pose_initializer(): # [TODO] Check the class methods
    def __init__(self, dev, so3gridreso=1, init_by_render=False, img_sz=512, split_bs = 256):
        gridso3 = third_party.sdfest_so3grid.SO3Grid(so3gridreso)
        n = gridso3.num_cells()
        self.n_cells = int(n/2) # the other half repeats
        # generate rotation grid samples
        quat = gridso3.index_to_quat(0) # x y z w
        quat = np.concatenate((quat[3:4], quat[:3])) # w x y z
        batch_quat = np.expand_dims(quat, 0) # (1,4)
        for i in range(1, self.n_cells): # [WARN] the other half repeats (see above)
            quat = gridso3.index_to_quat(i) # x y z w
            quat = np.concatenate((quat[3:4], quat[:3])) # w x y z (4)
            batch_quat = np.concatenate([batch_quat, np.expand_dims(quat,0)], axis=0) # (n,4)
        self.grid_rotmats = pytorch3d.transforms.quaternion_to_matrix(torch.tensor(batch_quat)).to(dev).to(torch.float32) # (n,3,3)
        print('rotation guesses: ', self.grid_rotmats.shape)
        # find the best initial rotation by compare render
        if init_by_render:
            self.cam_setup = setCameras(img_sz, split_bs, dev)
        self.img_sz = img_sz
        self.dev = dev
        self.eps = 1e-8 # for division
        self.split_bs = split_bs
        
    def initialize_old(self, trg_ptc, trg_depth2d, src_v, src_f, rot_c_w, t_c_w, K_cam, bs, 
                   pyt3d_img_coords=True, n_guess_rot=20, outlier_th=(0.0,1.0), n_samples=1000, verts_as_samples=False):
        # [DONE]: test this code with rendering. Tested projecting to the image place. Notice the different pytorch3d img coords
        # trg_ptc: structures.Pointcloud obj, trg_depth2d: 2d depth image [bs,h,w,1]
        # src_v: [bs,n_v,3], src_f: src uni faces [1,n_f,3], outlier_th: to extract max difference and robust to outliers
        # rot_c_w: [bs,3,3], t_c_w: [bs,3], K_cam: [bs,3,3]
        # use the 2D bounding box aspect ratio to filter rotation guesses for efficiency
        t_w_o_init, cloud_sizes = self.compute_centers_scales(trg_ptc, outlier_th) # [bs,3] [bs]
        # get trg bounding box aspect ratio
        trg_aspect = self.compute_target_aspect_ratios(trg_depth2d)
        # each object in a batch generates a batch of rotation guesses
        src_f_multi = src_f.expand(n_guess_rot, -1, -1)
        rotmat_init = torch.tensor([], device=self.dev)
        for b in range(bs): # for each object in the batch
            aspect_ratios = self.compute_projected_aspect_ratios(src_v[b], self.grid_rotmats, t_w_o_init[b], 
                                                                 rot_c_w[b], t_c_w[b], K_cam[b], pyt3d_img_coords) # [n]
            # gather rotations of the closest aspect ratios
            aspect_diff = torch.abs(trg_aspect[b] - aspect_ratios) # [n]
            top_indices = torch.argsort(aspect_diff, dim=-1, descending=False) # from low to high
            top_rotmats = torch.index_select(self.grid_rotmats, 0, top_indices[:n_guess_rot]) # [n_guess,3,3]
            # compute chamfer distances
            chamfer_dist, _ = self.compute_chamfer_distances(src_v[b], src_f_multi, top_rotmats, t_w_o_init[b], cloud_sizes[b],
                                                             verts_as_samples, n_samples,
                                                             trg_ptc.points_padded()[b].unsqueeze(0).expand(n_guess_rot,-1,-1),
                                                             trg_ptc.num_points_per_cloud()[b].unsqueeze(0).expand(n_guess_rot))
            # select the rotation with the smallest chamfer
            best_rotmat_idx = torch.argmin(chamfer_dist, dim=0)
            best_rotmat = torch.index_select(top_rotmats, 0, best_rotmat_idx) # [1,3,3]
            rotmat_init = torch.cat([rotmat_init, best_rotmat], dim=0)
        return rotmat_init, t_w_o_init.unsqueeze(-1), cloud_sizes # [bs,3,3], [bs,3,1], [bs]
    
    def initialize(self, trg_ptc, src_v, src_f, bs, verts_as_samples=False, n_samples=100, outlier_th=(0.0,1.0), n_guess_rot=1):
        # trg_ptc: structures.Pointcloud obj src_v: [bs,n_v,3], src_f: src uni faces [1,n_f,3] n_guess_rot: number of rotations in final output
        # outlier_th: to extract max difference and robust to outliers
        t_w_o_init, cloud_sizes = self.compute_centers_scales(trg_ptc, outlier_th) # [bs,3] [bs]
        src_f_multi = src_f.expand(self.n_cells, -1, -1)
        rotmat_init = torch.tensor([], device=self.dev)
        for b in range(bs):
            chamfer_dists, per_point_chamfer = self.compute_chamfer_distances(src_v[b], src_f_multi, self.grid_rotmats,
                                                                              t_w_o_init[b], cloud_sizes[b], verts_as_samples, n_samples,
                                                                              samples_trg.points_list()[b].unsqueeze(0).expand(self.n_cells,-1,-1),None,False)
            chamfer_dists_meanstd = chamfer_dists + torch.std(per_point_chamfer, dim=-1) # [n_g]
            if n_guess_rot > 1:
                top_rot_idx = torch.argsort(chamfer_dists_meanstd, dim=0, descending=False)
            else:
                top_rot_idx = torch.argmin(chamfer_dists_meanstd, dim=0)
            best_rotmat = torch.index_select(self.grid_rotmats, 0, top_rot_idx) # [n_g,3,3]
            rotmat_init = torch.cat([rotmat_init, best_rotmat.unsqueeze(0)], dim=0) # [bs,n_g,3,3]
        if n_guess_rot == 1:
            rotmat_init = rotmat_init.squeeze(1) # if n_g == 1, [bs,3,3], else, [bs,n_g,3,3]
        return rotmat_init.detach(), t_w_o_init.unsqueeze(-1).detach(), cloud_sizes.detach() # [bs,3,3] or [bs,n_g,3,3], [bs,3,1], [bs]
    
    def compute_target_aspect_ratios(self, trg_depth2d):
        # trg_depth2d: [bs,h,w,1]
        trg_mask2d = (trg_depth2d >= 0).squeeze(-1) # [bs,h,w]
        # trg_boxes2d = torchvision.ops.masks_to_boxes(trg_mask2d) # [bs,4] (x1,y1,x2,y2) from upper left corner using pixel indexing. update version
        trg_boxes2d = masks_to_boxes(trg_mask2d)
        trg_aspect = (trg_boxes2d[:,2] - trg_boxes2d[:,0]) / (trg_boxes2d[:,3] - trg_boxes2d[:,1] + self.eps) # [bs] x/y w/h
        return trg_aspect
    
    def compute_centers_scales(self, trg_ptc, outlier_th=(0.0,1.0)):
        # trg_ptc: structures.Pointcloud object
        # get the trg points center as the init translation t_w_o
        trg_centers = trg_ptc.points_padded().sum(-2) / trg_ptc.num_points_per_cloud().unsqueeze(-1).to(torch.float32) # [bs,3] avoid zero padding
        # get the axis-aligned max difference as scale
        q = torch.tensor(outlier_th, dtype=torch.float32, device=self.dev)
        cloud_sizes = []
        pt_i = 0
        for pt in trg_ptc.points_list():
            # quant_points = torch.quantile(pt, q, dim=0, keepdim=False) # [n_pt,3]->[2,3] quantile not available in early pytorch
            # quant_diffs = quant_points[-1] - quant_points[0] # [3]
#             quant_diffs = torch.max(pt, dim=0).values - torch.min(pt, dim=0).values # [3]
#             cloud_sizes.append(torch.max(quant_diffs)) # in torch.max if dim is not specified only values are returned no indices
#             cloud_sizes.append(torch.norm(quant_diffs) / 2.0) # tried on 20230724 worse
            dists_to_ctrs = torch.norm(pt - trg_centers[pt_i], dim=-1) # [n] cannot work in batch due to padded zeros (these zeros may get large dist)
            cloud_sizes.append(torch.mean(dists_to_ctrs)) # tried on 20230725: farthest point from center times two
            pt_i += 1
        cloud_sizes = torch.tensor(cloud_sizes, dtype=torch.float32, device=self.dev) # [bs]
        return trg_centers, cloud_sizes
    
    def compute_projected_aspect_ratios(self, vertices, rot_wo_guesses, t_w_o_init, rot_c_w, t_c_w, K_cam, pyt3d_img_coords): # visually checked code
        # only handles one object (not a batch of objects)
        # vertices: [n_v,3] rot_wo_guesses: [n,3,3] t_w_o_init: [3] rot_c_w: [3,3] t_c_w: [3] K_cam: [3,3]
        # compute relative pose T_c_o
        aspect_ratios = torch.tensor([], dtype=vertices.dtype, device=vertices.device)
        for bs0 in range(0, self.n_cells, self.split_bs):
            bs1 = bs0 + self.split_bs # last incomplete batch will get over indexed (idx >= len) but this is fine
            rot_wo_guesses_b = rot_wo_guesses[bs0:bs1]
            rot_c_o = rot_c_w.matmul(rot_wo_guesses_b) # (n,3,3)
            t_c_o = rot_c_w.matmul(t_w_o_init.unsqueeze(-1)) + t_c_w.unsqueeze(-1) # (3,1)
            # cast object vertices into camera frame [TODO] need scale (maybe ignore)
            v_c = rot_c_o.matmul(vertices.transpose(-1,-2)) + t_c_o # (n,3,n_v)
            # project to image plane and convert to pytorch3d convention
            xyd_c = K_cam.matmul(v_c) # (n,3,n_v) u v depth [TODO] negative depth. [TODO] image boundary clip
            uv_c = torch.stack([xyd_c[:,0,:] / (xyd_c[:,2,:] + self.eps),
                                xyd_c[:,1,:] / (xyd_c[:,2,:] + self.eps)], dim=-2) # (n,2,n_v)
            if pyt3d_img_coords: # pytorch3d image x+ left y+ up (i.e. upside-down leftside-right camera for the same rotation)
                uv_c = -uv_c # do u+w/2 v+h/2 since pytorch3d coords are at img ctr not top-left. here uv_c may have negative values.
            # compute aspect ratio
            u_diff = torch.max(uv_c[:,0,:], dim=-1).values - torch.min(uv_c[:,0,:], dim=-1).values # [n]
            v_diff = torch.max(uv_c[:,1,:], dim=-1).values - torch.min(xyd_c[:,1,:], dim=-1).values
            aspect_ratios_b = u_diff / (v_diff + self.eps) # [n]
            aspect_ratios = torch.cat([aspect_ratios, aspect_ratios_b], dim=0)
        return aspect_ratios
    
    @staticmethod
    def compute_chamfer_distances(vertices, faces, rot_wo_guesses, t_wo_init, scl_init, verts_as_samples, n_samples, trg_points,
                                  trg_points_len, return_meshes=False): # visually checked code
        # only handles one object (not a batch of objects)
        # vertices: [n_v,3] faces: [n,n_f,3] rot_wo_guesses: [n,3,3] t_wo_init: [3] scl_init: [] verts_as_samples: use vertices as samples
        # trg_points: [n,n_p,3] trg_points_len: [n] or None
        # removing the batch dim for inputs faces, trg_points, trg_points_len is not easy as they may have batch dim != self.n_cells. Follow rot_wo_guesses.
        # apply the initial guesses to the source
        # src_v_init = apply_trsfms_to_pts(vertices, rot_wo_guesses, t_wo_init, scl_init) # [n_v,3]->[n_guess,1,3,3][n_v,3,1]->[n_guess,n_v,3,1][3,1]->[n_guess,n_v,3]
        src_v_init = rot_wo_guesses.matmul((vertices * scl_init).transpose(-1,-2)) + t_wo_init.unsqueeze(-1) # [n_guess,3,n_v]
        if return_meshes or not verts_as_samples:
            meshes_init = pytorch3d.structures.Meshes(verts=src_v_init.transpose(-1,-2), faces=faces)
        if verts_as_samples:
            samples_init = src_v_init.detach().clone().transpose(-1,-2) # [n_guess,n_v,3]
        else:
            samples_init = pytorch3d.ops.sample_points_from_meshes(meshes_init, n_samples) # [n_guess,n_samp,3]
        # compute chamfer distance for each rotation guesses
        chamfer_dist, chamfer_per_point = iou_chamfer_loss(trg_points, samples_init, len1=trg_points_len,
                                              iou=False, chamfer=True, reduce_batch=False, reduce_points=False) # [n_guess], [n_guess,n_p]
        if return_meshes:
            return chamfer_dist, chamfer_per_point, meshes_init
        else:
            return chamfer_dist, chamfer_per_point
    
    def compute_depth_img_diffs(self, trg_depth_img, meshes_to_render, cam_rot_wc, cam_t_cw, rasterizer):
        # trg_depth_img: [h,w,1] meshes_to_render: structures.Meshes with bs = n_guesses
        # cam_rot_wc: [3,3] cam_t_cw: [3] rasterizer: use the rasterizer in the main code
        depth_img_diff = torch.tensor([], dtype=trg_depth_img.dtype, device=trg_depth_img.device)
        for bs0 in range(0, self.n_cells, self.split_bs):
            bs1 = bs0 + self.split_bs # last incomplete batch will get over indexed (idx >= len) but this is fine
            if bs1 > self.n_cells:
                bs1 = self.n_cells
                actual_split_bs = bs1 - bs0
                cam_setup = setCameras(self.img_sz, actual_split_bs, self.dev)
            else:
                actual_split_bs = self.split_bs
                cam_setup = self.cam_setup
            meshes_to_render_b = meshes_to_render[bs0:bs1]
            cam_setup.cameras.R = cam_rot_wc.unsqueeze(0).expand(actual_split_bs,-1,-1)
            cam_setup.cameras.T = cam_t_cw.unsqueeze(0).expand(actual_split_bs,-1)
            rendered_depth_img = rasterizer.rasterize(meshes_to_render_b, cam_setup.cameras).detach() # [n_guesses,h,w,1]
            # make invalid depth 0 instead of -1
            trg_depth_img = trg_depth_img.clamp(min=0)
            rendered_depth_img = rendered_depth_img.clamp(min=0)
            # compute differences
            depth_img_diff_b = torch.abs(rendered_depth_img - trg_depth_img).squeeze(-1).mean((-1,-2)) # [n_guesses]
            depth_img_diff = torch.cat([depth_img_diff, depth_img_diff_b], dim=0)
            del meshes_to_render_b, rendered_depth_img
        return depth_img_diff
```

```python
iter_loader = iter(shapenet_loader)
data_loop = tqdm(range(len(iter_loader)))
best_rot_ar_place = [] # the place of the best rotation in aspect ratio diffs
best_rot_cd_place = [] # the place of the best rotation in chamfer distances
best_rot_cdstd_place = [] # the place of the best rotation in mean chamfer distance + std chamfer distance of all target spoints
best_rot_cdmed_place = [] # median chamfer distance
best_rot_cdmedstd_place = [] # median + stddev
best_rot_ddiff_place = []
min_theta_diffs = []
min_theta_diffs_check = []
best_rot_ar_theta = []
best_rot_cd_theta = []
best_rot_cdstd_theta = []
best_rot_cdmed_theta = []
best_rot_cdmedstd_theta = []
best_rot_ddiff_theta = []
best_rot_ar_intheta = []
best_rot_cd_inthtea = []
best_rot_cdstd_intheta = []
best_rot_cdmed_intheta = []
best_rot_cdmedstd_intheta = []
best_rot_ddiff_intheta = []
for i in data_loop:
    meshes_i = next(iter_loader)
    meshes_i = exclude_weird_shapes(meshes_i)
    bs_actual = len(meshes_i['model_id'])
    if bs_actual == 0:
        continue
    assert(bs_actual != 0)
    # if i == len(iter_loader) - 1:
    #     bs_actual = len(meshes_i['model_id']) # last iter may have a different bs
    faces_i = meshes_i['mesh'].faces_padded().to(device)
    verts_padded = meshes_i['mesh'].verts_padded().to(device) # [bs,n,3] padded with 0
    # find axis-aligned bounding box in canonical pose
    cano_bboxes = extract_cano_mesh_bbx(meshes_i['mesh'].to(device)).to(device) # [bs,8,3]
    # get gt and initial poses and target meshes
    ((random_rots, random_trsl, random_scls), (init_rots, init_trsl, init_scls),
    (trg_meshes, trg_bboxes)) = get_gt_pose_init_pose(bs_actual, device,
                                                      verts_padded, faces_i, cano_bboxes,
                                                      init_rot_std, init_t_std, init_scl_std, init_scl_lbd) # [bs,3,3] [bs,3,1] [bs]
    # render the object to depth image
    if cameras_setup.bs != bs_actual:
        cameras_setup = setCameras(img_size, bs_actual, device)
    cameras_setup.cameras_pointing_obj(random_trsl)
    depth_img = rasterizer.rasterize(trg_meshes, cameras_setup.cameras) # [bs,h,w,1]
    data_proc.set_data(depth_img, cameras_setup.inv_intr, cameras_setup.rot_w_c, cameras_setup.cameras.get_camera_center())
    data_proc.project_to_3d(masking=True)
    samples_trg = data_proc.unmasked_pts # structures.Pointcloud obj
    # initialize shape
    if shape_init == 'mean':
        pca_est_params = torch.full((bs_actual, K), 0.0, device=device, requires_grad=True).to(torch.float32) # pca_mean is added in init 
    if shape_init == 'avg':
        pca_est_params = torch.mean(pca_basis_params_torch, dim=0, keepdim=True).expand(bs_actual, -1).clone().requires_grad_() # -1 not change size = K. clone() since expand refers to single memory
    verts_init = torch.matmul(pca_est_params, pca_bases) + pca_mean
    verts_init = verts_init.reshape(bs_actual,-1,3)
    # initialize translation and scale
    init_trsl, init_scls = pose_initer.compute_centers_scales(samples_trg, (0.0,1.0)) # [bs,3] [bs]
    # compute target aspect ratios
    trg_aspect = pose_initer.compute_target_aspect_ratios(depth_img)
    # test rotation hypotheses for each object
    batch_faces = uni_faces.expand(pose_initer.n_cells, -1, -1)
    for b in range(bs_actual): # per object
        # compute scores w.r.t. gt
        theta_diffs = [] # rotation guesses for each object
        Rt_gt = torch.cat((random_rots[b], random_trsl[b]), dim=-1)
        Rt_gt = torch.cat((Rt_gt, torch.tensor([[0,0,0,1]], dtype=torch.float32, device=device)), dim=0).detach().cpu().numpy()
        for r in range(pose_initer.n_cells):
            Rt_guess = torch.cat((pose_initer.grid_rotmats[r], init_trsl[b].unsqueeze(-1)), dim=-1)
            Rt_guess = torch.cat((Rt_guess, torch.tensor([[0,0,0,1]], dtype=torch.float32, device=device)), dim=0).detach().cpu().numpy()
            theta_t = third_party.nocs_utils.compute_RT_degree_cm_symmetry(Rt_guess, Rt_gt, 0, False, [meshes_i['label'][b]])
            theta_diffs.append(theta_t[0]) # [n]
        # compute aspect ratio differences
        aspect_ratios = pose_initer.compute_projected_aspect_ratios(verts_init[b], pose_initer.grid_rotmats, init_trsl[b],
                                                                    cameras_setup.cameras.R.transpose(-1,-2)[b],
                                                                    cameras_setup.cameras.T[b], cameras_setup.cam_intr[b], True) # [n]
        aspect_diffs = torch.abs(trg_aspect[b] - aspect_ratios) # [n]
        # compute chamfer distances
        chamfer_dists, per_point_chamfer, init_meshes = pose_initer.compute_chamfer_distances(
                                                                                 verts_init[b], batch_faces, pose_initer.grid_rotmats, init_trsl[b],
                                                                                 init_scls[b], verts_as_sppts, n_sample_pts,
                                                                                 samples_trg.points_list()[b].unsqueeze(0).expand(pose_initer.n_cells,-1,-1),
                                                                                 None, return_meshes=True) # [n] [n,n_p]
                                                                                 # samples_trg.num_points_per_cloud()[b].unsqueeze(0).expand(pose_initer.n_cells)) # [n]
        # process per-point chamfer
        chamfer_dists_meanstd = chamfer_dists + torch.std(per_point_chamfer, dim=-1)
        chamfer_dists_median = torch.median(per_point_chamfer, dim=-1)[0] # [n]
        chamfer_dists_medstd = torch.median(per_point_chamfer, dim=-1)[0] + torch.std(per_point_chamfer, dim=-1)
        # compute depth img differences
        if init_by_render:
            depth_img_diff = pose_initer.compute_depth_img_diffs(depth_img[b], init_meshes, cameras_setup.cameras.R[b],
                                                                 cameras_setup.cameras.T[b], rasterizer) # [n]
        else:
            depth_img_diff = torch.zeros_like(chamfer_dists)
        # order with the gt angle differences
        top_indices_gt = np.argsort(theta_diffs, axis=-1) # low to high i.e. good to bad. each is an idx to theta_diffs
        top_indices_ar = torch.argsort(aspect_diffs, dim=-1, descending=False).detach().cpu().numpy()
        top_indices_cd = torch.argsort(chamfer_dists, dim=-1, descending=False).detach().cpu().numpy()
        top_indices_cdstd = torch.argsort(chamfer_dists_meanstd, dim=-1, descending=False).detach().cpu().numpy()
        top_indices_cdmed = torch.argsort(chamfer_dists_median, dim=-1, descending=False).detach().cpu().numpy()
        top_indices_cdmedstd = torch.argsort(chamfer_dists_medstd, dim=-1, descending=False).detach().cpu().numpy()
        top_indices_ddiff = torch.argsort(depth_img_diff, dim=-1, descending=False).detach().cpu().numpy()
        best_rot_ar_place.append(np.where(top_indices_ar == top_indices_gt[0])[0] + 1) # +1 to convert 0-indexed to 1-indexed
        best_rot_cd_place.append(np.where(top_indices_cd == top_indices_gt[0])[0] + 1) # last [0] because np.where returns a tuple with only one element here
        best_rot_cdstd_place.append(np.where(top_indices_cdstd == top_indices_gt[0])[0] + 1)
        best_rot_cdmed_place.append(np.where(top_indices_cdmed == top_indices_gt[0])[0] + 1)
        best_rot_cdmedstd_place.append(np.where(top_indices_cdmedstd == top_indices_gt[0])[0] + 1)
        best_rot_ddiff_place.append(np.where(top_indices_ddiff == top_indices_gt[0])[0] + 1)
        # see the minimum angle difference to see if the grid is fine enough
        min_theta_diffs.append(theta_diffs[top_indices_gt[0]])
        min_theta_diffs_check.append(np.min(theta_diffs))
        # first place rotation hypothesis theta_diff
        best_rot_ar_theta.append(theta_diffs[top_indices_ar[0]])
        best_rot_cd_theta.append(theta_diffs[top_indices_cd[0]])
        best_rot_cdstd_theta.append(theta_diffs[top_indices_cdstd[0]])
        best_rot_cdmed_theta.append(theta_diffs[top_indices_cdmed[0]])
        best_rot_cdmedstd_theta.append(theta_diffs[top_indices_cdmedstd[0]])
        best_rot_ddiff_theta.append(theta_diffs[top_indices_ddiff[0]])
        # first place rotation theta_diff place
        best_rot_ar_intheta.append(np.where(top_indices_gt == top_indices_ar[0])[0] + 1)
        best_rot_cd_inthtea.append(np.where(top_indices_gt == top_indices_cd[0])[0] + 1)
        best_rot_cdstd_intheta.append(np.where(top_indices_gt == top_indices_cdstd[0])[0] + 1)
        best_rot_cdmed_intheta.append(np.where(top_indices_gt == top_indices_cdmed[0])[0] + 1)
        best_rot_cdmedstd_intheta.append(np.where(top_indices_gt == top_indices_cdmedstd[0])[0] + 1)
        best_rot_ddiff_intheta.append(np.where(top_indices_gt == top_indices_ddiff[0])[0] + 1)
    # release memory
    del meshes_i, faces_i, verts_padded, trg_meshes, samples_trg
    torch.cuda.empty_cache()
    gc.collect() # does not help with memory exhaustion
# end of data loop
```

```python
iter_loader = iter(shapenet_loader)
data_loop = tqdm(range(len(iter_loader)))
# record results
pose_scl_gt = []
pose_scl_pred = []
shape_gt = []
shape_pred = []
posed_shape_gt = []
posed_shape_pred = []
labels = []
count_models = 0
total_time = 0
for i in data_loop:
    meshes_i = next(iter_loader) # [DONE] exclude weird shapes
    meshes_i = exclude_weird_shapes(meshes_i)
    bs_actual = len(meshes_i['model_id'])
    if bs_actual == 0:
        continue
    assert(bs_actual != 0)
    # if i == len(iter_loader) - 1:
    #     bs_actual = len(meshes_i['model_id']) # last iter may have a different bs
    faces_i = meshes_i['mesh'].faces_padded().to(device)
    verts_padded = meshes_i['mesh'].verts_padded().to(device) # [bs,n,3] padded with 0
    # find axis-aligned bounding box in canonical pose
    cano_bboxes = extract_cano_mesh_bbx(meshes_i['mesh'].to(device)).to(device) # [bs,8,3]
    # get gt and initial poses and target meshes
    ((random_rots, random_trsl, random_scls), (init_rots, init_trsl, init_scls),
    (trg_meshes, trg_bboxes)) = get_gt_pose_init_pose(bs_actual, device,
                                                      verts_padded, faces_i, cano_bboxes,
                                                      init_rot_std, init_t_std, init_scl_std, init_scl_lbd) # [bs,3,3][bs,3,1][bs]
    # sample on the target mesh
    if not rendered_depth:
        samples_trg_batch = pytorch3d.ops.sample_points_from_meshes(trg_meshes, n_sample_pts) # [bs,n_samples,3]
    else:
        if cameras_setup.bs != bs_actual:
            cameras_setup = setCameras(img_size, bs_actual, device)
        cameras_setup.cameras_pointing_obj(random_trsl)
        depth_img = rasterizer.rasterize(trg_meshes, cameras_setup.cameras).detach() # [bs,h,w,1]
        data_proc.set_data(depth_img, cameras_setup.inv_intr, cameras_setup.rot_w_c, cameras_setup.cameras.get_camera_center())
        data_proc.project_to_3d(masking=True)
        samples_trg_batch = data_proc.unmasked_pts.detach() # structures.Pointcloud obj 
    # initialize translation and scale
    if rendered_depth and compute_init_guess:
        start = timeit.default_timer()
        init_trsl, init_scls = pose_initer.compute_centers_scales(samples_trg_batch, (0.0,1.0)) # [bs,3] [bs]
        init_trsl = init_trsl.unsqueeze(-1) # [bs,3,1] to be consistent with the init_trsl above
        total_time += (timeit.default_timer() - start)
    for b in range(bs_actual):
        # set up for multi-rotation hypotheses
        n_grids = pose_initer.n_cells
        batch_faces = uni_faces.expand(n_grids, -1, -1)
        est_trsl = init_trsl[b].unsqueeze(0).expand(n_grids,-1,-1).clone() # [n_g,3,1]
        est_scls = init_scls[b].unsqueeze(0).expand(n_grids).clone() # [n_g]
        est_rots = pose_initer.grid_rotmats.clone() # [n_g,3,3]
        if not rendered_depth:
            samples_trg = samples_trg_batch[b:b+1].expand(n_grids,-1,-1).clone() # [n_g,n_samples,3]
        else:
            samples_trg = samples_trg_batch[b].extend(n_grids) # [n_g] of Pointclouds object but now same length
        # initialize em gaussian covariance
        if em and rendered_depth:
            sigma_sq_inv = sigma_sq_inv_init * torch.ones(samples_trg.points_padded()[:,:,0].shape, device=device, dtype=torch.float32) # [n_g,n]
        else:
            sigma_sq_inv = None
        # initialize pca basis shape
        if shape_init == 'mean':
            pca_est_params = torch.full((n_grids, K), 0.0, device=device, requires_grad=True).to(torch.float32) # pca_mean is added in init 
        if shape_init == 'avg':
            pca_est_params = torch.mean(pca_basis_params_torch, dim=0, keepdim=True).expand(n_grids, -1).clone().requires_grad_() # [n_g,K] -1 not change size = K. clone() since expand refers to single memory
        verts_init = torch.matmul(pca_est_params, pca_bases) + pca_mean # [n_g=bs,n_verts*3] used for saving results
        verts_init = torch.reshape(verts_init, (n_grids, -1, 3))
        # shape optimization set-up
        if shape_mode == 0:
            optimizer_shape = torch.optim.SGD([pca_est_params], lr=lr_shape_m0, momentum=mm_shape_m0)
        if shape_mode == 1:
            optim_scls = torch.ones_like(est_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
            optim_trsl = torch.zeros_like(est_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
            optim_rots_so3 = torch.zeros_like(est_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
            optimizer_shape = torch.optim.SGD([pca_est_params, optim_scls, optim_trsl, optim_rots_so3], lr=lr_shape_m1, momentum=mm_shape_m1)
        # optimization loop
        resample_count = 0
        start = timeit.default_timer()
        for j in range(n_iters):
            # check if want to resample
            if (j+1) in resample_iters: # reserve for more complex conditions
                resample = True
                resample_count += 1
            else:
                resample = False
            if resample:
                # compute src mesh and samples
                new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # [n_g,n_verts*3] dim broadcastable
                new_verts_0 = torch.reshape(new_verts_0, (n_grids, -1, 3)) # [n_g,n_verts,3]
                new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
                if verts_as_sppts:
                    samples_src = new_verts
                else:
                    recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
                    samples_src = pytorch3d.ops.sample_points_from_meshes(recon_meshes, n_sample_pts)
                # compute a score with chamfer distance and resample using the score
                if not rendered_depth: # [DONE] 20230718 scoring method subject to changes after pose initialization tested
                    chamfer_dists, chamfer_per_point = iou_chamfer_loss(samples_trg, samples_src, iou=False, chamfer=True, reduce_batch=False, reduce_points=False)
                else:
                    chamfer_dists, chamfer_per_point = iou_chamfer_loss(samples_trg.points_padded(), samples_src, len1=None,
                                              iou=False, chamfer=True, reduce_batch=False, reduce_points=False) # [n_grids] [DONE] remove length argument for samples_trg later
                chamfer_dists_meanstd = chamfer_dists.detach().clone() + torch.std(chamfer_per_point, dim=-1).detach().clone()
                n_grids = resample_n[resample_count-1]
                if resample_method == 0: # resample with softmax prob
                    softmax_prob = 1.0 - torch.softmax(chamfer_dists_meanstd, dim=0) # small distance means better so 1.0 minus
                    selected = torch.multinomial(softmax_prob, n_grids, replacement=False)
                if resample_method == 1 or resample_method == 2:
                    selected = torch.argsort(chamfer_dists_meanstd, dim=0, descending=False) # low to high, good to bad
                    if resample_method == 1:
                        selected = selected[:n_grids]
                    if resample_method == 2:
                        dummy_trsl = np.zeros((3,1)).astype(np.float32)
                        selected_list = list(selected.detach().cpu().numpy())
                        est_rots_np = est_rots.detach().cpu().numpy()
                        idx1 = selected_list.pop(0)
                        spaced_selected = [idx1]
                        rot1 = np.concatenate([est_rots_np[idx1,:,:], dummy_trsl], axis=-1) # [3,4]
                        rot1 = np.concatenate([rot1, [[0,0,0,1]]], axis=0) # [4,4]
                        while (n_grids - len(spaced_selected)) < len(selected_list) and len(spaced_selected) < n_grids:
                            idx2 = selected_list.pop(0)
                            rot2 = np.concatenate([est_rots_np[idx2,:,:], dummy_trsl], axis=-1) # [3,4]
                            rot2 = np.concatenate([rot2, [[0,0,0,1]]], axis=0) # [4,4]
                            theta_diff = third_party.nocs_utils.compute_RT_degree_cm_symmetry(rot2, rot1, 0, False, [meshes_i['label'][b]])[0] # [0] to get theta
                            if theta_diff >= theta_spacing:
                                spaced_selected.append(idx2)
                                rot1 = rot2
                                idx1 = idx2
                        if len(spaced_selected) < n_grids:
                            spaced_selected += selected_list[:n_grids-len(spaced_selected)]
                        selected = torch.tensor(spaced_selected, device=device)
                # pick the selected ones
                est_rots = torch.index_select(est_rots, 0, selected).clone() # [n_g,3,3]
                est_trsl = torch.index_select(est_trsl, 0, selected).clone() # [n_g,3,1]
                est_scls = torch.index_select(est_scls, 0, selected).clone() # [n_g]
                samples_trg = samples_trg[:n_grids]
                batch_faces = uni_faces.expand(n_grids, -1, -1)
                pca_est_params = torch.index_select(pca_est_params, 0, selected).detach().clone().requires_grad_() # [n_g,K]
                if em and rendered_depth:
                    sigma_sq_inv = torch.index_select(sigma_sq_inv, 0, selected).clone()
                # shape optimization set-up
                if shape_mode == 0:
                    optimizer_shape = torch.optim.SGD([pca_est_params], lr=lr_shape_m0, momentum=mm_shape_m0) # SGD object not subscritable cannot pick
                if shape_mode == 1:
                    optim_scls = torch.ones_like(est_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
                    optim_trsl = torch.zeros_like(est_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
                    optim_rots_so3 = torch.zeros_like(est_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
                    optimizer_shape = torch.optim.SGD([pca_est_params, optim_scls, optim_trsl, optim_rots_so3], lr=lr_shape_m1, momentum=mm_shape_m1)
            # create meshes from bases and transform to estimated pose and scale
            new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # [n_g,n_verts*3] dim broadcastable
            new_verts_0 = torch.reshape(new_verts_0, (n_grids, -1, 3)) # [n_g,n_verts,3]
            if normalize_for_pose or center_for_pose:
                new_verts_0, _, _ = normalize_center_vertices(new_verts_0, normalize_for_pose, center_for_pose) # [n_g,n,3], [n_g,1,1], [n_g,1,3]
            if mode1 == 0 or mode1 == 1:
                new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
                if verts_as_sppts:
                    samples_src = new_verts
                else:
                    recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
                    # sample on the source mesh - inefficient
                    samples_src = pytorch3d.ops.sample_points_from_meshes(recon_meshes, n_sample_pts) # [n_g,n_samples,3]
            # pose estimation
            sim_trsfm = None
            # knn_umeyama or icp
            if mode1 == 0:
                for aln_i in range(mode1_iters): # [DONE] Possibly wrong. Causing cuda runtime error
                    if not rendered_depth:
                        sim_trsfm = knn_umeyama(samples_trg, samples_src, est_scale=True)
                    else:
                        if not em:
                            sim_trsfm = knn_umeyama_hetero(samples_trg, samples_src, est_scale=True) # [DONE] update to EM
                        else:
                            sim_trsfm, sigma_sq_inv = knn_umeyama_hetero_em(samples_trg, samples_src, est_scale=True,
                                                                         update_sigma=(j > 0 or aln_i > 0) and not fix_sigma, knn_k=knn_k,
                                                                         sigma_sq_inv=sigma_sq_inv, bs=n_grids, deg=3)
                    if mode1_iters > 1:
                        est_scls = est_scls * sim_trsfm[2].detach() # [bs] accumuate increments
                        est_rots = sim_trsfm[0].detach().transpose(-1,-2).matmul(est_rots) # [bs,3,3]
                        est_trsl = sim_trsfm[0].detach().transpose(-1,-2).matmul( # [bs,3,3]
                           (sim_trsfm[2].detach().unsqueeze(-1).unsqueeze(-1) * est_trsl)) + sim_trsfm[1].detach().unsqueeze(-1) # [bs,3,1]
                        samples_src = apply_trsfms_to_pts(samples_src, sim_trsfm[0].detach().transpose(-1,-2),
                                                          sim_trsfm[1].detach().unsqueeze(-1), sim_trsfm[2].detach()) # transform with the increments
                        sim_trsfm = None
            if mode1 == 1:
                sim_trsfm = icp(samples_trg, samples_src, est_scale=True)
            # accumulate increments on estimates if mode1 is run
            if sim_trsfm:
                est_scls = est_scls * sim_trsfm[2].detach() # [bs]
                est_rots = sim_trsfm[0].detach().transpose(-1,-2).matmul(est_rots) # [bs,3,3]
                est_trsl = sim_trsfm[0].detach().transpose(-1,-2).matmul( # [bs,3,3]
                    (sim_trsfm[2].detach().unsqueeze(-1).unsqueeze(-1) * est_trsl)) + sim_trsfm[1].detach().unsqueeze(-1) # [bs,3,1]
                samples_src = apply_trsfms_to_pts(samples_src, sim_trsfm[0].detach().transpose(-1,-2),
                                                  sim_trsfm[1].detach().unsqueeze(-1), sim_trsfm[2].detach()) # transform with the increments
                sim_trsfm = None
            # mode2 - set up optimizer
            if mode2 == 0:
                optim_scls = torch.ones_like(est_scls, dtype=torch.float32, device=device, requires_grad=True)
                optimizer = torch.optim.SGD([optim_scls], lr=lr_pose_scl, momentum=0.9)
            if mode2 == 1:
                optim_scls = torch.ones_like(est_scls, dtype=torch.float32, device=device, requires_grad=True)
                optim_trsl = torch.zeros_like(est_trsl, dtype=torch.float32, device=device, requires_grad=True)
                optimizer = torch.optim.SGD([optim_scls, optim_trsl], lr=lr_pose_scl, momentum=0.9)
            if mode2 == 2:
                optim_scls = torch.ones_like(est_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
                optim_trsl = torch.zeros_like(est_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
                optim_rots_so3 = torch.zeros_like(est_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
                optimizer = torch.optim.SGD([optim_scls, optim_trsl, optim_rots_so3], lr=lr_pose_scl, momentum=0.9)
            if mode2 == 0 or mode2 == 1 or mode2 == 2:
                # samples_src_start = samples_src.detach().clone()
                if verts_as_sppts:
                    samples_src_start = new_verts_0.detach().clone()
                else:
                    recon_meshes_0 = pytorch3d.structures.Meshes(verts=new_verts_0, faces=batch_faces) # for applying directly on R,t,s
                    samples_src_start = pytorch3d.ops.sample_points_from_meshes(recon_meshes_0, n_sample_pts)
                for k in range(mode2_iters):
                    optimizer.zero_grad()
                    optim_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3)
                    # apply transformation to vertices and then sample from mesh - inefficient - depracated. use only after updating
                    # est_verts = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, est_scls)
                    # src_meshes = pytorch3d.structures.Meshes(verts=est_verts, faces=faces_i).to(device)
                    # samples_src = pytorch3d.ops.sample_points_from_meshes(src_meshes, n_sample_pts) # [bs,n_samples,3] resample needed if shape changed
                    # apply transformation directly to samples
                    # samples_src = apply_trsfms_to_pts(samples_src_start, optim_rots, optim_trsl, optim_scls) # transform with increments
                    samples_src = apply_trsfms_to_pts(samples_src_start, optim_rots.matmul(est_rots), optim_trsl+est_trsl, optim_scls*est_scls)
                    src_bboxes = apply_trsfms_to_pts(cano_bboxes[b].unsqueeze(0).expand(n_grids,-1,-1).clone(), 
                                                     optim_rots, optim_trsl, optim_scls) # [WRONG] need bbox in estimated pose but iou loss not used
                    if not rendered_depth:
                        losses = iou_chamfer_loss(samples_trg, samples_src, bbox1=trg_bboxes, bbox2=src_bboxes, iou=False, chamfer=True)
                    else:
                        losses = iou_chamfer_loss(samples_trg.points_padded(), samples_src, len1=samples_trg.num_points_per_cloud(),
                                                  bbox1=trg_bboxes, bbox2=src_bboxes, iou=False, chamfer=True)
                    loss = losses[0] * iou_wt + losses[1] * chamfer_wt
                    loss.backward()
                    optimizer.step()
                    # accumulate increments on estimates - pytorch seems not able to zero the variables without reinit the optimizer
                    # est_scls *= optim_scls.detach()
                    # if mode2 == 1 or mode2 == 2:
                    #     est_trsl += optim_trsl.detach()
                    #     if mode2 == 2:
                    #         est_rots  = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
                # alternative - compose the estimates with the optimized quantities 
                est_scls *= optim_scls.detach()
                if mode2 == 1 or mode2 == 2:
                    est_trsl += optim_trsl.detach()
                    # est_trsl = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(optim_scls.detach().unsqueeze(-1).unsqueeze(-1) * est_trsl) + optim_trsl.detach()
                    if mode2 == 2:
                        est_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
            # apply transformation to vertices and then sample from mesh - inefficient
            # est_verts = apply_trsfms_to_pts(verts_padded, est_rots, est_trsl, est_scls)
            # src_meshes = pytorch3d.structures.Meshes(verts=est_verts, faces=faces_i).to(device)
            # apply transformation directly to samples
            # samples_src = apply_trsfms_to_pts(samples_src_0, est_rots, est_trsl, est_scls)
            if j+1 >= shape_early_stop_iter:
                continue
            # shape estimation
            for k_sh in range(n_shape_iters):
                optimizer_shape.zero_grad()
                new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # [n_g,n_verts*3] dim broadcastable
                new_verts_0 = torch.reshape(new_verts_0, (n_grids, -1, 3)) # [n_g,n_verts,3]
                if shape_mode == 0:
                    new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
                if shape_mode == 1:
                    optim_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3)
                    new_verts = apply_trsfms_to_pts(new_verts_0, optim_rots.matmul(est_rots), optim_trsl+est_trsl, optim_scls*est_scls)
                recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
                if not rendered_depth:
                    [loss_chamfer, loss_edge, loss_normal, loss_laplacian, loss_ptedge, loss_ptface] \
                        = mesh_deform_loss(recon_meshes, samples_trg, n_samples_trg=n_sample_pts,
                                           use_temp_verts=verts_as_sppts, n_samples_temp=n_sample_pts,
                                           chamfer_on=wt_chamfer, edge_on=wt_edge, normal_on=wt_normal,
                                           laplacian_on=wt_laplacian, ptedge_on=wt_ptedge, ptface_on=wt_ptface,
                                           given_trg_samples=True)
                else:
                    [loss_chamfer, loss_edge, loss_normal, loss_laplacian, loss_ptedge, loss_ptface] \
                        = mesh_deform_loss(recon_meshes, samples_trg, n_samples_trg=n_sample_pts,
                                           use_temp_verts=verts_as_sppts, n_samples_temp=n_sample_pts,
                                           chamfer_on=wt_chamfer, edge_on=wt_edge, normal_on=wt_normal,
                                           laplacian_on=wt_laplacian, ptedge_on=wt_ptedge, ptface_on=wt_ptface,
                                           given_trg_samples=True, double_chamfer=False, em_weights=em, knn_k=knn_k, sigma_sq_inv=sigma_sq_inv) # [DONE] update to EM
                total_loss = (wt_chamfer * loss_chamfer + wt_edge * loss_edge + \
                         wt_normal * loss_normal + wt_laplacian * loss_laplacian + \
                         wt_ptedge * loss_ptedge + wt_ptface * loss_ptface) * bs_actual
                if wt_modsink > 0:
                    loss_modsink = torch.sum(torch.min(torch.sum(torch.abs(
                        pca_est_params.unsqueeze(-2) - pca_basis_params_torch.unsqueeze(0)), dim=-1), dim=-1)[0])
                    if wt_modsink_schd == 0:
                        wt_modsink_ac = wt_modsink
                    if wt_modsink_schd == 1:
                        wt_modsink_ac = wt_modsink / (n_iters * n_shape_iters) * (n_shape_iters * j + k_sh + 1)
                    if wt_modsink_schd == 2:
                        if n_shape_iters * j + k_sh + 1 <= n_iters * n_shape_iters * wt_modsink_thpct:
                            wt_modsink_ac = 0
                        else:
                            wt_modsink_th = np.floor(n_iters * n_shape_iters * wt_modsink_thpct)
                            wt_modsink_ac = wt_modsink / (n_iters * n_shape_iters - wt_modsink_th) *\
                                                         (n_shape_iters * j + k_sh + 1 - wt_modsink_th)
                    if wt_modsink_schd == 3:
                        wt_modsink_ac = wt_modsink / (n_iters * n_shape_iters)**2 * (n_shape_iters * j + k_sh + 1)**2
                    total_loss += wt_modsink_ac * loss_modsink
                total_loss.backward()
                optimizer_shape.step()
            if shape_mode == 1:
                est_scls *= optim_scls.detach()
                est_trsl += optim_trsl.detach()
                est_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
        # end of optimization loop -- n_grids = bs = 1 should reduce to one
        assert(n_grids == 1)
        total_time += (timeit.default_timer() - start)
        # [DONE] solve the inefficiency by swaping out meshes_i['faces'] which is a list
        # [DONE] add iou and chamfer loss weights
        # [DONE] Test point cloud transformation and transformation of vertices + sampling speed conparison
        # record results
        labels.append(meshes_i['label'][b:b+1]) # bs
        count_models += 1
        # record shape (no pose transformation) results
        new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # dim broadcastable
        new_verts_0 = torch.reshape(new_verts_0, (n_grids, -1, 3)) # n_grids = 1
        if normalize_for_eval or center_for_eval:
            new_verts_0, norm_factor, centroid = normalize_center_vertices(new_verts_0, normalize_for_eval, center_for_eval) # [bs,n,3], [bs,1,1], [bs,1,3]
        if center_for_eval:
            est_trsl += est_rots.matmul((est_scls.unsqueeze(-1) * centroid.squeeze(-2)).unsqueeze(-1)) # [bs,3,1] t' = t + Rsc
        if normalize_for_eval:
            est_scls *= norm_factor.squeeze() # [bs] s' = s*norm
        recon_meshes_0 = pytorch3d.structures.Meshes(verts=new_verts_0, faces=batch_faces)
        gt_meshes = pytorch3d.structures.Meshes(verts=verts_padded[b:b+1], faces=faces_i[b:b+1])
        shape_gt.append(gt_meshes.detach().cpu())
        shape_pred.append(recon_meshes_0.detach().cpu())
        # record transformed shape results
        new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
        recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
        posed_shape_gt.append(trg_meshes.detach().cpu())
        posed_shape_pred.append(recon_meshes.detach().cpu())
        # record pose results
        pose_scl_gt.append((random_rots[b:b+1].detach().cpu().numpy(), # [bs,3,3]
                            random_trsl[b:b+1].detach().cpu().numpy(), # [bs,3,1]
                            random_scls[b:b+1].detach().cpu().numpy())) # [bs]
        pose_scl_pred.append((est_rots.detach().cpu().numpy(),
                              est_trsl.detach().cpu().numpy(),
                              est_scls.detach().cpu().numpy()))
        # [DONE] more results to record
        # save some results to disk
        if (i + 1) % save_freq == 0 and b == 0:
            # save the first one in a batch
            pytorch3d.io.IO().save_mesh(trg_meshes[b], os.path.join(save_dir, 'iter_{:d}_trg.ply'.format(i + 1)))
            trg_verts_src_pose = apply_trsfms_to_pts(verts_padded[b:b+1], est_rots, est_trsl, est_scls)
            trg_meshes_src_pose = pytorch3d.structures.Meshes(verts=trg_verts_src_pose, faces=faces_i[b:b+1])
            pytorch3d.io.IO().save_mesh(trg_meshes_src_pose[0], os.path.join(save_dir, 'iter_{:d}_trg_src_pose.ply'.format(i + 1)))
            trg_verts_src_pose_scl1 = apply_trsfms_to_pts(verts_padded[b:b+1], est_rots, est_trsl, torch.ones_like(est_scls))
            trg_meshes_src_pose_scl1 = pytorch3d.structures.Meshes(verts=trg_verts_src_pose_scl1, faces=faces_i[b:b+1])
            pytorch3d.io.IO().save_mesh(trg_meshes_src_pose_scl1[0], os.path.join(save_dir, 'iter_{:d}_trg_src_pose_scl1.ply'.format(i + 1)))
            # src_verts_est = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
            # src_meshes_est = pytorch3d.structures.Meshes(verts=src_verts_est, faces=batch_faces)
            pytorch3d.io.IO().save_mesh(recon_meshes[0], os.path.join(save_dir, 'iter_{:d}_src.ply'.format(i + 1)))
            pytorch3d.io.IO().save_mesh(recon_meshes_0[0], os.path.join(save_dir, 'iter_{:d}_src_shape.ply'.format(i + 1)))
            pytorch3d.io.IO().save_mesh(meshes_i['mesh'][b], os.path.join(save_dir, 'iter_{:d}_trg_shape.ply'.format(i + 1)))
            verts_init = apply_trsfms_to_pts(verts_init[b:b+1], est_rots, init_trsl[b:b+1], init_scls[b:b+1])
            init_meshes = pytorch3d.structures.Meshes(verts=verts_init, faces=batch_faces)
            pytorch3d.io.IO().save_mesh(init_meshes[0], os.path.join(save_dir, 'iter_{:d}_initscltrslshape_estrot.ply'.format(i + 1)))
            if rendered_depth:
                pytorch3d.io.IO().save_pointcloud(samples_trg[0], os.path.join(save_dir, 'iter_{:d}_trgdepth.ply'.format(i + 1)))
        # release memory
        del samples_src, samples_trg
    # end of batch loop
    # release memory
    del meshes_i, faces_i, verts_padded, trg_meshes
    torch.cuda.empty_cache()
    gc.collect() # does not help with memory exhaustion
# end of data loop
```

```python
# record image_id, inst_id, partial ptc for objects in each class
partial_ptc_all, image_id_all, inst_id_all, inst_idx_all, gt_mesh_all, init_pose_all = {}, {}, {}, {}, {}, {}
depth_imgs_all, masks_all, gt_RTs_all, fr_results_save, gt_mesh_all_alt = {}, {}, {}, {}, {}
for syn_name in synset_names[1:]: # no BG
    partial_ptc_all[syn_name] = []
    image_id_all[syn_name] = []
    inst_id_all[syn_name] = []
    inst_idx_all[syn_name] = []
    gt_mesh_all[syn_name] = []
    gt_mesh_all_alt[syn_name] = []
    init_pose_all[syn_name] = []
    depth_imgs_all[syn_name] = []
    masks_all[syn_name] = []
    gt_RTs_all[syn_name] = [] # for testing depth rendering (pytorch3d weird conventions)
    fr_results_save[syn_name] = [] # whether the results of the instances in a frame are saved
inst_count = 0
skipped_images_nomrcnn = []
all_image_results = {}
# iterating all images and batch up partial point cloud data
images_loop = tqdm(range(len(dataset.image_ids))) ##################################TESTING no [:10]
for i in images_loop:
    image_id = dataset.image_ids[i] # for i, image_id in enumerate(dataset.image_ids):
    result = {}
    # load image, depth, and mask
    image_path = dataset.image_info[image_id]["path"] # without color, mask, depth
    image = dataset.load_image(image_id) # run the load_image in dataset.py read color image
    depth = dataset.load_depth(image_id) # uint16 image
    if depth_factor != 1.0:
        depth = depth.astype(np.float64) / depth_factor # see utils.backproject this will cause the xyz of ptc all divided by 1000
    gt_mask, gt_coord, gt_class_ids, gt_scales, gt_domain_label = dataset.load_mask(image_id) # indexed in the same order as gt_sorted_insts
    gt_bbox = utils.extract_bboxes(gt_mask) # [4] gt_bbox indices follow gt_mask so follows gt_class_ids and gt_sorted_insts (contains inst id)
    inst_dict =  dataset.image_info[image_id]["inst_dict"] # updated with deleted instances during dataset.load_mask/process_data
    gt_sorted_insts = sorted(inst_dict.keys()) # make sure sorted since instance_ids in dataset.py/process_data is sorted so things like gt_class_ids are also sorted by instance
    assert(len(gt_sorted_insts) == len(gt_class_ids))
    result['image_id'] = image_id
    result['image_path'] = image_path
    result['gt_class_ids'] = gt_class_ids # 0-indexed but 0 is BG
    result['gt_inst_ids'] = np.array(gt_sorted_insts) # was np.arange(len(gt_class_ids)) + 1 # 1-indexed
    result['gt_bboxes'] = gt_bbox          
    result['gt_scales'] = gt_scales
    # load gt pose
    image_path_parsing = image_path.split('/')
    gt_pkl_path = os.path.join(gt_dir, 'results_{}_{}_{}.pkl'.format(data, image_path_parsing[-2], image_path_parsing[-1]))
    assert(os.path.exists(gt_pkl_path)) # I think all gt files for REAL and CAMERA val should exist
    if (os.path.exists(gt_pkl_path)):
        with open(gt_pkl_path, 'rb') as f:
            gt = cPickle.load(f)
        result['gt_RTs'] = gt['gt_RTs']
        assert(gt['gt_RTs'].shape[0] == len(gt_class_ids)) # make sure gt_RTs has no deleted instances
        if 'handle_visibility' in gt:
            result['gt_handle_visibility'] = gt['handle_visibility']
            assert len(gt['handle_visibility']) == len(gt_class_ids)
            print('got handle visibiity.')
        else: 
            result['gt_handle_visibility'] = np.ones_like(gt_class_ids)
    # finish loading gt
    # prepare mrcnn results
    if use_mrcnn_mask:
        mrcnn_results_path = os.path.join(mrcnn_results_dir, data, 'results_{}_{}_{}.pkl'.format(data.split('_')[-1], image_path_parsing[-2], image_path_parsing[-1]))
        if not os.path.exists(mrcnn_results_path):
            skipped_images_nomrcnn.append(mrcnn_results_path.split('/')[-1])
            continue
        with open(mrcnn_results_path, 'rb') as f:
            mrcnn_content = cPickle.load(f)
        pred_mask = mrcnn_content['masks'] # [h,w,n_insts]
        if np.shape(pred_mask)[-1] <= 0:
            skipped_images_nomrcnn.append(mrcnn_results_path.split('/')[-1])
            continue
        pred_class_ids = mrcnn_content['class_ids']
        sorted_insts = list(range(len(pred_class_ids))) # sorting
        pred_bbox = utils.extract_bboxes(pred_mask) # [n_pred_inst, 4]
        result['pred_class_ids_origidx'] = np.array(pred_class_ids) # [n_pred_inst] save these for repacking the results later after estimation
        result['pred_bboxes_origidx'] = pred_bbox
        result['pred_scores_origidx'] = np.array(mrcnn_content['scores']) # [n_pred_inst]
        # mask overlap with gt mask for gt mesh matching
        moverlaps_pred_gt = utils.compute_overlaps_masks(pred_mask, gt_mask) # [n_inst_pred, n_inst_gt] entry is iou
    else:
        sorted_insts = gt_sorted_insts
    all_image_results[image_id] = result
    # prepare semantic primitives results as initial pose
    if use_sprim_init_pose:
        sprim_results_path = os.path.join(sprim_results_dir, 'results_{}_{}_{}.pkl'.format(data.split('_')[-1], image_path_parsing[-2], image_path_parsing[-1]))
        with open(sprim_results_path, 'rb') as f:
            sprim_results = cPickle.load(f)
    # prepare the meta.txt file for object mesh
    meta_path = dataset.image_info[image_id]["path"] + '_meta.txt'
    with open(meta_path, 'r') as f:
        lines = f.readlines()
    # iterate over instances to prepare batch data for each class
    for j in range(len(sorted_insts)):
        if use_mrcnn_mask:
            if pred_class_ids[j] == 0 or pred_class_ids[j] > 6: # copied from shape classification loading
                print('found invalid class ids at i = {:d} j = {:d}'.format(i, j))
                print('image path: ' + image_path)
                continue
        else:
            if gt_class_ids[j] == 0 or gt_class_ids[j] > 6:
                print('found invalid class ids at i = {:d} j = {:d}'.format(i, j))
                print('image path: ' + image_path)
                continue
        if not use_mrcnn_mask:
            assert(inst_dict[sorted_insts[j]] == gt_class_ids[j])
            inst_count += 1
            image_id_all[synset_names[gt_class_ids[j]]].append(image_id) # int
            inst_id_all[synset_names[gt_class_ids[j]]].append(sorted_insts[j]) # 1-indexed and skipping deleted instances
            inst_idx_all[synset_names[gt_class_ids[j]]].append(j) # int
            depth_imgs_all[synset_names[gt_class_ids[j]]].append(depth) # [h,w]
            masks_all[synset_names[gt_class_ids[j]]].append(gt_mask[:,:,j]) # [h,w]
            # backproject to get point cloud [DONE] filter out outlier points
            backproj_pts, _ = utils.backproject(depth, intrinsics, gt_mask[:,:,j]) # [n_ij,3] need flip x and y due to nocs convention
        else:
            backproj_pts, _ = utils.backproject(depth, intrinsics, pred_mask[:,:,j])
            if backproj_pts.shape[0] < ignore_inst_th: # ignore instances that do not have many points
                continue
            inst_count += 1
            image_id_all[synset_names[pred_class_ids[j]]].append(image_id)
            inst_id_all[synset_names[pred_class_ids[j]]].append(sorted_insts[j])
            inst_idx_all[synset_names[pred_class_ids[j]]].append(j)
            depth_imgs_all[synset_names[pred_class_ids[j]]].append(depth)
            masks_all[synset_names[pred_class_ids[j]]].append(pred_mask[:,:,j])
        backproj_pts[:, 0] = -backproj_pts[:, 0] # flipping x and y since it is flipped in utils.backproject
        backproj_pts[:, 1] = -backproj_pts[:, 1]
        # remove background outlier points
        ptc_bp = open3d.geometry.PointCloud()
        ptc_bp.points = open3d.utility.Vector3dVector(backproj_pts.reshape(-1,3))
        clean_ptc_bp, _ = ptc_bp.remove_statistical_outlier(nb_neighbors=filter_nmb_neighbors, std_ratio=filter_nmb_stddev)
        backproj_pts_clean = np.asarray(clean_ptc_bp.points) # [n_p,3]
        # append to a list
        if not use_mrcnn_mask:
            partial_ptc_all[synset_names[gt_class_ids[j]]].append(backproj_pts_clean)
        else:
            partial_ptc_all[synset_names[pred_class_ids[j]]].append(backproj_pts_clean)
        # append initial pose
        if use_mrcnn_mask and use_sprim_init_pose:
            sprim_rot = sprim_results['pred_RTs'][j][:3,:3] # [3,3]
            sprim_trsl = sprim_results['pred_RTs'][j][:3,-1:] # [3,1]
            sprim_xyz_scales = sprim_results['pred_scales'][j] # [3]
            # factor out the scale in rotation
            sprim_rot_scale = np.cbrt(np.linalg.det(sprim_rot))
            sprim_rot_valid = sprim_rot / sprim_rot_scale
            assert(np.isclose(np.linalg.det(sprim_rot_valid), 1.0)) # check valid rotation
            assert(np.allclose(sprim_rot_valid.T@sprim_rot_valid, np.eye(3)))
            assert(np.allclose(sprim_rot_valid@sprim_rot_valid.T, np.eye(3)))
            # factor out the scale in bbox xyz scales
            norm_xyz_scales = np.linalg.norm(sprim_xyz_scales)
            # make the scalar scale
            sprim_scalar_scale = sprim_rot_scale * norm_xyz_scales
            init_pose_all[synset_names[pred_class_ids[j]]].append((sprim_rot_valid, # [3,3]
                                                                   sprim_trsl, # [3,1] not yet applied the 1000 factor apply in the estimation part
                                                                   sprim_scalar_scale)) # [] not yet applied the 1000 factor
        # find the gt mesh file for the current instance 
        if use_mrcnn_mask:
            if sprior_gtshape_match:
                gt_inst_idx = -1
                # original shape prior matching
                for k in range(len(gt_class_ids)):
                    if gt_class_ids[k] != pred_class_ids[j]:
                        continue
                    if moverlaps_pred_gt[j,k] < 0.4: # only match by greedily over a threshold
                        continue
                    gt_inst_idx = k
                    break
                # alternative: use the max iou match that is greater than the threshold
                matched_class_id_gt_insts = np.equal(np.array(gt_class_ids), pred_class_ids[j]) # array of true false
                matched_iou = np.greater_equal(moverlaps_pred_gt[j], 0.4) # array of true false
                matched_id_iou = np.logical_and(matched_class_id_gt_insts, matched_iou)
                if np.sum(matched_id_iou) <= 0:
                    gt_inst_idx_alt = -1
                else:
                    gt_inst_idx_alt = np.argmax(moverlaps_pred_gt[j] * matched_id_iou)
            else:
                # greedy match to gt instance using mask iou
                gt_inst_idx = np.argmax(moverlaps_pred_gt[j]) # idx should follow gt_mask so also can index gt_sorted_insts, gt_class_ids
                # try to avoid a gt instance assigned to two pred instances (can also try without this)
                if moverlaps_pred_gt.shape[0] <= moverlaps_pred_gt.shape[1]: # n_pred_inst <= n_gt_inst
                    moverlaps_pred_gt[:, gt_inst_idx] = -1.0 # set to -1 so never be the max again (np is okay with a scalar assigned to a column)
        else:
            gt_inst_idx = j
        # follow dataset.py/load_objs to save path to each object mesh and use the path as key to a dict to extract mesh
        if gt_inst_idx != -1:
            words = lines[gt_sorted_insts[gt_inst_idx] - 1][:-1].split(' ') # insts are 1-indexed while lines are 0-indexed
            inst_id = int(words[0])
            assert(inst_id == gt_sorted_insts[gt_inst_idx])
            if len(words) == 3: ## real data # see process_data comments
                if words[2][-3:] == 'npz':
                    obj_name = words[2].replace('.npz', '_norm.obj')
                    mesh_file = os.path.join(config.OBJ_MODEL_DIR, 'real_val', obj_name) # not my case
                else:
                    mesh_file = os.path.join(config.OBJ_MODEL_DIR, data, words[2] + '.obj')
            else:
                assert len(words) == 4 ## synthetic data
                mesh_file = os.path.join(config.OBJ_MODEL_DIR, data, words[2], words[3], 'model.obj')
        else:
            mesh_file = -1
        if not use_mrcnn_mask:
            gt_mesh_all[synset_names[gt_class_ids[j]]].append(mesh_file)
        else:
            gt_mesh_all[synset_names[pred_class_ids[j]]].append(mesh_file) # this is directly used for evaluation with shape results later
        # alternative of computing gt shape matches similar to sprior method but with max iou. copy above code for ease
        if use_mrcnn_mask and sprior_gtshape_match:
            if gt_inst_idx_alt != -1:
                words = lines[gt_sorted_insts[gt_inst_idx_alt] - 1][:-1].split(' ') # insts are 1-indexed while lines are 0-indexed
                inst_id = int(words[0])
                assert(inst_id == gt_sorted_insts[gt_inst_idx_alt])
                if len(words) == 3: ## real data # see process_data comments
                    if words[2][-3:] == 'npz':
                        obj_name = words[2].replace('.npz', '_norm.obj')
                        mesh_file = os.path.join(config.OBJ_MODEL_DIR, 'real_val', obj_name) # not my case
                    else:
                        mesh_file = os.path.join(config.OBJ_MODEL_DIR, data, words[2] + '.obj')
                else:
                    assert len(words) == 4 ## synthetic data
                    mesh_file = os.path.join(config.OBJ_MODEL_DIR, data, words[2], words[3], 'model.obj')
            else:
                mesh_file = -1
            gt_mesh_all_alt[synset_names[pred_class_ids[j]]].append(mesh_file) # this is directly used for evaluation with shape results later
        # for testing depth rendering
        if not use_mrcnn_mask:
            gt_RTs_all[synset_names[gt_class_ids[j]]].append(gt['gt_RTs'][j]) # assume the gt poses are arranged in the same order as gt_masks gt_class_ids
        # to determine whether the results of a frame is saved
        if (i + 1) % fr_results_save_freq == 0:
            fr_save_choice = True
        else:
            fr_save_choice = False
        if not use_mrcnn_mask:
            fr_results_save[synset_names[gt_class_ids[j]]].append(fr_save_choice)
        else:
            fr_results_save[synset_names[pred_class_ids[j]]].append(fr_save_choice)
print('total number of instances: ', inst_count)
print('number of skipped images due no mrcnn results: ', len(skipped_images_nomrcnn))
```

```python
class shape_latent_vae(nn.Module):
    def __init__(self, input_dim, lat_dim, dev, hidden_dim=(1024, 512), linear_dec=True):
        super(shape_latent_vae, self).__init__()
        self.dev = dev
        self.encoder = nn.Sequential(
            nn.Linear(input_dim, hidden_dim[0]),
            nn.LeakyReLU(0.2),
            nn.Linear(hidden_dim[0], hidden_dim[1]),
            nn.LeakyReLU(0.2))
        self.mean_layer = nn.Linear(hidden_dim[1], lat_dim)
        self.logvar_layer = nn.Linear(hidden_dim[1], lat_dim)
        if linear_dec:
            self.decoder = nn.Linear(lat_dim, input_dim) # y = xA.T + b
        else:
            self.decoder = nn.Sequential(
                nn.Linear(lat_dim, hidden_dim[1]),
                nn.LeakyReLU(0.2),
                nn.Linear(hidden_dim[1], hidden_dim[0]),
                nn.LeakyReLU(0.2),
                nn.Linear(hidden_dim[0], input_dim))
    
    def set_sampling(self, on_off):
        self.sampling = on_off
    
    def reparam(self, mean, sigma): # [DONE] remove reparam when testing
        epsilon = torch.randn_like(sigma).to(self.dev)
        z = mean + sigma * epsilon
        return z
    
    def encoding(self, x):
        pre_lat = self.encoder(x)
        mean = self.mean_layer(pre_lat)
        logvar = self.logvar_layer(pre_lat)
        return mean, logvar
    
    def forward(self, x):
        mean, log_var = self.encoding(x)
        if self.sampling:
            lat_z = self.reparam(mean, torch.exp(0.5 * log_var)) # [TODO] double check with the KL reg loss
        else:
            lat_z = mean
        x_hat = self.decoder(lat_z)
        return x_hat, mean, log_var
    
def recon_loss(x, x_hat):
    return torch.square(x_hat - x).mean() # mean across vertices and batch

def kldiv_loss(mean, log_var):
    return -0.5 * torch.sum(1 + log_var - mean.pow(2) - log_var.exp(), -1).mean() # mean across batch [TODO] double check
```

```python
# pose_initer only for computing init translation and scale
pose_initer = pose_initializer(dev=device, so3gridreso=so3grid_resol, img_sz=0) # img_sz only for init_by_render
# conversion matrix between shapenet v1 and v2
rot_v2_to_v1 = torch.tensor([[0,0,-1], [0,1,0], [1,0,0]], dtype=torch.float32, device=device) # [3,3]
rot_v1_to_v2 = torch.tensor([[0,0,1], [0,1,0], [-1,0,0]], dtype=torch.float32, device=device) # [3,3]
# reading shape classification files and pre-checking
if shape_init == 'classify':
    if data == 'real_test':
        if use_mrcnn_mask:
            with open(os.path.join(nocs_shape_cls_dir, 'real_test_mrcnn', 'synname_packed_cls_image_paths.pickle'), 'rb') as f:
                repacked_data_to_shape_cls_images = pickle.load(f)
        else:
            with open(os.path.join(nocs_shape_cls_dir, 'real_test', 'synname_packed_cls_image_paths.pickle'), 'rb') as f:
                repacked_data_to_shape_cls_images = pickle.load(f)
    else:
        print('shape classification for the data type not done!')
    # check lens matched to partial_ptc_all
    for syn_cls in synset_names[1:]: ########################### Commented out for testing
        assert(len(partial_ptc_all[syn_cls]) == len(repacked_data_to_shape_cls_images[syn_cls]))
# load gt shape vertices and faces
if use_gt_shape:
    gt_shape_points = {}
    gt_shape_meshes = {}
    gt_shape_verts = {}
    gt_shape_faces = {}
    for obj in gt_shapes.keys(): # code from the evaluation part below
        verts_torch = torch.tensor(gt_shapes[obj][0], dtype=torch.float32, device=device).unsqueeze(0) # [1,n_v,3]
        if convert_to_snv1_rot:
            verts_torch = torch.matmul(verts_torch, rot_v1_to_v2.transpose(-1,-2)) # should be pre-multiply transpose to post-multiply
        faces_torch = torch.tensor(gt_shapes[obj][1], dtype=torch.long, device=device).unsqueeze(0) # [1,n_f,3]
        gt_shape_verts[obj] = verts_torch
        gt_shape_faces[obj] = faces_torch
# results recording
total_time = {}
shape_time = {}
shape_time_names = ['model recon', 'trg centering model trsfm and meshing', 'em wts init',\
                    'mesh deform losses total time', 'mesh deform losses summation',\
                    'sink loss', 'render loss', 'backprop', 'shape param update']
inst_count_per_cls = {}
pose_scl_pred = {}
shape_pred = {}
viz_fr_dir = os.path.join(save_dir, 'viz_fr')
if not os.path.exists(viz_fr_dir):
    os.mkdir(viz_fr_dir)
# iterate over classes
for syn_cls in synset_names[1:]: # 1 to exclude BG ##################################TESTING should be 1
    print(syn_cls)
    viz_dir = os.path.join(save_dir, 'viz_' + syn_cls)
    if not os.path.exists(viz_dir):
        os.mkdir(viz_dir)
    total_time[syn_cls] = 0
    shape_time[syn_cls] = np.zeros(17)
    inst_count_per_cls[syn_cls] = 0
    pose_scl_pred[syn_cls] = []
    shape_pred[syn_cls] = []
    # load the pca results to torch
    pca_file = glob.glob(os.path.join(pca_dir, '*' + syn_cls + '.pickle'))[0]
    with open(pca_file, 'rb') as f:
        pca_dict = pickle.load(f)
    if linear_shape:
        pca_mean = torch.tensor(pca_dict['pca_mean']).to(torch.float32).to(device) # (n_verts*3)
        pca_bases = torch.tensor(pca_dict['pca_bases']).to(torch.float32).to(device) # (K, n_verts*3)
        pca_basis_params_torch = torch.tensor(pca_dict['pca_params']).to(torch.float32).to(device) # (n_obj,K)
        uni_faces = torch.tensor(pca_dict['faces']).to(torch.float32).to(device) # detach() returns a new tensor and no requires_grad
        K = pca_bases.size(dim=0) # int
        n_verts = int(pca_mean.size(dim=0) / 3)
        print('pca n_verts: ', n_verts)
    else:
        vae_model_file = glob.glob(os.path.join(vae_model_dir, '*' + syn_cls + '.pth'))[0]
        vae_model = torch.load(vae_model_file)
        vae_model.eval()
        print('shape vae on device: ', vae_model.dev)
        vae_model.set_sampling(False)
        pca_basis_params_torch = torch.tensor(pca_dict['pca_params']).to(torch.float32).to(device) # (n_obj,K)
        uni_faces = torch.tensor(pca_dict['faces']).to(torch.float32).to(device) # detach() returns a new tensor and no requires_grad
        K = pca_basis_params_torch.size(dim=1) # int
    # load shape classification results if needed
    if shape_init == 'classify':
        if data == 'real_test':
            if use_mrcnn_mask:
                with open(os.path.join(shape_cls_results_dir, syn_cls, 'real_test_mrcnn_results.pickle'), 'rb') as f:
                    shape_cls_results = pickle.load(f)
            else:
                with open(os.path.join(shape_cls_results_dir, syn_cls, 'real_test_test_results.pickle'), 'rb') as f:
                    shape_cls_results = pickle.load(f)
    # load closest shapenet shapes (various versions/modes)
    if use_closest_shape:
        shape_dir_timeid = '_'.join(pca_file.split('/')[-1].split('_')[0:3])
        closest_shape_dir = glob.glob(os.path.join(closest_shape_meta_dir, shape_dir_timeid + '*'))[0]
        # get all the matched closest shape object ids
        closest_shape_ids = []
        for roi_img in repacked_data_to_shape_cls_images[syn_cls]:
            if shape_cls_results[roi_img] not in closest_shape_ids:
                closest_shape_ids.append(shape_cls_results[roi_img])
        # load closest shapes
        if closest_shape_mode == 0:
            shape_file_suffix = '_trg'
        if closest_shape_mode == 1:
            shape_file_suffix = '_std'
        if closest_shape_mode == 2:
            shape_file_suffix = '_' + pca_file.split('_')[-2]
        closest_shape_verts = {}
        closest_shape_faces = {}
        for shapeid in closest_shape_ids:
            cl_shape_file = glob.glob(os.path.join(closest_shape_dir, shapeid + shape_file_suffix + '*'))
            assert(len(cl_shape_file) == 1)
            cl_shape_file = cl_shape_file[0]
            cl_verts, cl_faces = pytorch3d.io.load_ply(cl_shape_file) # [n_v,3] [n_f,3] should be already in shapenet v2 convention
            closest_shape_verts[shapeid] = cl_verts.to(torch.float32).to(device).unsqueeze(0) # [1,n_v,3] load verts separately to be transformed
            closest_shape_faces[shapeid] = cl_faces.to(torch.long).to(device).unsqueeze(0) # [1,n_f,3]
    # use class specific shape optimization weights
    wt_chamfer = wts_chamfer[syn_cls]
    wt_edge = wts_edge[syn_cls]
    wt_normal = wts_normal[syn_cls]
    wt_laplacian = wts_laplacian[syn_cls]
    # class specific symmetry and compute rotation symmetry matrices
    rot_symm_on = rot_symm_obj[syn_cls]
    mirror_symm_on = mirror_symm_obj[syn_cls]
    if (symmetry_check or symm_cplt) and rot_symm_on:
        symm_test_angles = np.linspace(0.0, 2*np.pi, num=n_rot_symm+1, endpoint=False)[1:] # exclude 0 360 as it is the same as original
        symm_test_rots = []
        for symm_test_angle in symm_test_angles:
            symm_test_rot_i = pytorch3d.transforms.euler_angles_to_matrix(torch.tensor([0, symm_test_angle, 0]), 'XYZ').to(torch.float32).to(device) # [3,3]
            symm_test_rots.append(symm_test_rot_i)
        symm_cplt_rots = []
        for symm_cplt_angle in symm_cplt_rot_angles: # in deg
            sc_angle = symm_cplt_angle / 180.0 * np.pi
            symm_cplt_rot_i = pytorch3d.transforms.euler_angles_to_matrix(torch.tensor([0, sc_angle, 0]), 'XYZ').to(torch.float32).to(device) # [3,3]
            symm_cplt_rots.append(symm_cplt_rot_i)
    if (symmetry_check or symm_cplt) and mirror_symm_on:
        if symm_plane == 'v2':
            mirr_test_rot = torch.tensor([[-1,0,0],[0,1,0],[0,0,1]], dtype=torch.float32, device=device) # [3,3]
        elif symm_plane == 'v1':
            mirr_test_rot = torch.tensor([[1,0,0],[0,1,0],[0,0,-1]], dtype=torch.float32, device=device) # [3,3]
        else:
            print('unrecognized symm_plane type')
            assert(False)
    # iterate over instances in batch for the current class
    data_loop = tqdm(range(0, len(partial_ptc_all[syn_cls]), bs)) ################################
    i_for_save = 0
    for i in data_loop: # [TODO] right now batching is for pose initer compute centers and scales (not sure if this makes it faster)
        i_for_save += 1
        partial_ptc_batch = partial_ptc_all[syn_cls][i:i+bs]
        bs_actual = len(partial_ptc_batch) # at the end may not have a batch
        partial_ptc_batch_torch = [torch.tensor(pptc, dtype=torch.float32, device=device) for pptc in partial_ptc_batch]
        samples_trg_batch = pytorch3d.structures.Pointclouds(partial_ptc_batch_torch)
        inst_count_per_cls[syn_cls] += bs_actual
        # batch up the masked depth images
        if wt_render > 0 or render_criterion:
            depth_imgs_batch = depth_imgs_all[syn_cls][i:i+bs]
            masks_batch = masks_all[syn_cls][i:i+bs]
            depth_imgs_batch_torch = torch.tensor(depth_imgs_batch, dtype=torch.float32, device=device) # [bs,h,w]
            masks_batch_torch = torch.tensor(masks_batch, dtype=torch.float32, device=device) # [bs,h,w]
            masked_depth_batch_torch = depth_imgs_batch_torch * masks_batch_torch + (1.0 - masks_batch_torch.to(torch.float32)) * render_bgval # off-mask pixels get -1.0
        # compute initial translation scale
        if init_by_given:
            if use_gt_pose:
                given_init_pose = gt_RTs_all[syn_cls][i:i+bs] # list([4,4])
                init_rots = torch.stack([torch.tensor(init_p[:3,:3], dtype=torch.float32, device=device) for init_p in given_init_pose], dim=0) # [bs,3,3]
                init_trsl = torch.stack([torch.tensor(init_p[:3,3:], dtype=torch.float32, device=device) for init_p in given_init_pose], dim=0) # [bs,3,1]
                init_scls = torch.ones(bs, dtype=torch.float32, device=device) # [bs] in gt pose, single scale from Umeyama is factored into ratation. there is also gt_scales (x,y,z normalized) probably from shape.
            else:
                given_init_pose = init_pose_all[syn_cls][i:i+bs] # list[([3,3],[3,1],[])]
                init_rots = torch.stack([torch.tensor(init_p[0], dtype=torch.float32, device=device) for init_p in given_init_pose], dim=0) # [bs,3,3]
                init_trsl = torch.stack([torch.tensor(init_p[1], dtype=torch.float32, device=device) for init_p in given_init_pose], dim=0) # [bs,3,1]
                init_scls = torch.stack([torch.tensor(init_p[2], dtype=torch.float32, device=device) for init_p in given_init_pose], dim=0) # [bs]
            if nocs_1000factor:
                init_trsl *= 1000.0
                init_scls *= 1000.0
            if convert_to_snv1_rot:
                init_rots = torch.matmul(init_rots, rot_v2_to_v1) # [bs,3,3] convert to rot with shapenet v2 shape
        else:
            start = timeit.default_timer()
            init_trsl, init_scls = pose_initer.compute_centers_scales(samples_trg_batch, (0.0,1.0)) # [bs,3] [bs]
            init_trsl = init_trsl.unsqueeze(-1) # [bs,3,1] to be consistent with the init_trsl
            total_time[syn_cls] += (timeit.default_timer() - start)
        # get shape classification results
        if shape_init == 'classify':
            repackedd2clsimg = repacked_data_to_shape_cls_images[syn_cls][i:i+bs]
        
#         if i == 1: # TESTING
#             time.sleep(3600)
        
        # iter over instances
        for b in range(bs_actual):
            # set up for multi-rotation hypotheses
            n_grids = pose_initer.n_cells
            if not init_by_given:
                est_trsl = init_trsl[b].unsqueeze(0).expand(n_grids,-1,-1).clone() # [n_g,3,1]
                est_scls = init_scls[b].unsqueeze(0).expand(n_grids).clone() # [n_g]
                est_rots = pose_initer.grid_rotmats.clone() # [n_g,3,3]
            else:
                n_grids = 1
                est_rots = init_rots[b:b+1].clone() # [1,3,3]
                est_trsl = init_trsl[b:b+1].clone() # [1,3,1]
                est_scls = init_scls[b:b+1].clone() # [1]
            if not given_shape:
                batch_faces = uni_faces.expand(n_grids, -1, -1)
            elif use_gt_shape:
                # extract gt for this instance
                batch_faces = gt_shape_faces[gt_mesh_all[syn_cls][i+b]].clone().expand(n_grids, -1, -1)
                batch_given_verts = gt_shape_verts[gt_mesh_all[syn_cls][i+b]].clone().expand(n_grids, -1, -1)
            elif use_closest_shape:
                batch_faces = closest_shape_faces[shape_cls_results[repackedd2clsimg[b]]].clone().expand(n_grids, -1, -1)
                batch_given_verts = closest_shape_verts[shape_cls_results[repackedd2clsimg[b]]].clone().expand(n_grids, -1, -1)
            samples_trg = samples_trg_batch[b].extend(n_grids) # [n_g] of Pointclouds object but now same length
            # samples_trg = samples_trg_batch[b].points_padded().expand(n_grids,-1,-1).clone() # [n_g,n_p,3] (pose estimation uses Pointclouds as input)
            # center and normalize the trg and apply the same to src
#             trg_points_ctr_orig = torch.mean(samples_trg.points_padded(), dim=-2, keepdim=True) # [n_g,1,3] [DONE] use bbox center instead of centroid. may not need since here is just rough normalization not to gaurantee source model bbox center is at origin
#             trg_points_maxcorner_orig = torch.max(samples_trg.points_padded(), dim=-2, keepdim=True).values # [n_g,1,3]
#             trg_points_mincorner_orig = torch.min(samples_trg.points_padded(), dim=-2, keepdim=True).values # [n_g,1,3]
#             trg_points_bboxdiag_orig = trg_points_maxcorner_orig - trg_points_mincorner_orig
#             trg_points_bboxdiag_orig = torch.norm(trg_points_bboxdiag_orig, dim=-1, keepdim=True) # [n_g,1,1]
#             # trg_points_ctr_orig = (trg_points_maxcorner_orig + trg_points_mincorner_orig) / 2.0 # [n_g,1,3] bbx ctr
#             trg_points_normal_ctr_orig = (samples_trg.points_padded() - trg_points_ctr_orig) / trg_points_bboxdiag_orig
#             trg_points_normal_ctr_orig = pytorch3d.structures.Pointclouds(trg_points_normal_ctr_orig).detach().cpu() # ptc same len but for mesh_deform_loss compatibility use Pointclouds
            # initialize em gaussian covariance
            if em:
                sigma_sq_inv = sigma_sq_inv_init * torch.ones(samples_trg.points_padded()[:,:,0].shape, device=device, dtype=torch.float32) # [n_g,n]
#                 sigma_sq_inv_shape = sigma_sq_inv * torch.square(trg_points_bboxdiag_orig.squeeze(-1)) # [n_g,n]
            else:
                sigma_sq_inv = None
                sigma_sq_inv_shape = None
            # initialize pca basis shape
            if shape_init == 'mean':
                pca_est_params = torch.full((n_grids, K), 0.0, device=device, requires_grad=True).to(torch.float32) # pca_mean is added in init. pca_mean is data mean (i.e. verts mean)
            if shape_init == 'avg':
                pca_est_params = torch.mean(pca_basis_params_torch, dim=0, keepdim=True).expand(n_grids, -1).clone().requires_grad_() # [n_g,K] -1 not change size = K. clone() since expand refers to single memory
            if shape_init == 'classify':
                cls_pca_idx = pca_dict['obj_id_pca_idx_map'][shape_cls_results[repackedd2clsimg[b]]] # repacked data -> classification image -> classification result as obj id -> idx
                pca_est_params = pca_basis_params_torch[cls_pca_idx,:].unsqueeze(0).expand(n_grids, -1).clone().requires_grad_()
            if linear_shape and not given_shape:
                verts_init = torch.matmul(pca_est_params, pca_bases) + pca_mean # [n_g=bs,n_verts*3] used for saving results
            elif not linear_shape and not given_shape:
                verts_init = vae_model.decoder(pca_est_params.detach())
            else:
                verts_init = batch_given_verts
            verts_init = torch.reshape(verts_init, (n_grids, -1, 3))
            # shape optimization set-up
            if shape_mode == 0:
                optimizer_shape = torch.optim.SGD([pca_est_params], lr=lr_shape_m0, momentum=mm_shape_m0)
            if shape_mode == 1:
                optim_scls = torch.ones_like(est_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
                optim_trsl = torch.zeros_like(est_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
                optim_rots_so3 = torch.zeros_like(est_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
                optimizer_shape = torch.optim.SGD([pca_est_params, optim_scls, optim_trsl, optim_rots_so3], lr=lr_shape_m1, momentum=mm_shape_m1)
            # rendering camera set-up
            if wt_render > 0 or render_criterion:
                cameras_setup = setCameras(image_shape, n_grids, device, True, intrinsics)
                cameras_setup.zero_camera_pose() # cam is world pose_c_w = identity
                rasterizer.rasterizer.cameras = cameras_setup.cameras
            # optimization loop
            resample_count = 0
            start = timeit.default_timer()
            for j in range(n_iters):
                # check if want to resample
                if (j+1) in resample_iters: # reserve for more complex conditions
                    resample = True
                    resample_count += 1
                else:
                    resample = False
                if resample:
                    # compute src mesh and samples
                    if linear_shape and not given_shape:
                        new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # [n_g,n_verts*3] dim broadcastable
                    elif not linear_shape and not given_shape:
                        new_verts_0 = vae_model.decoder(pca_est_params.detach()) # [DONE] check all detach() ok for vae
                    else:
                        new_verts_0 = batch_given_verts
                    new_verts_0 = torch.reshape(new_verts_0, (n_grids, -1, 3)) # [n_g,n_verts,3]
                    new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
                    if verts_as_sppts:
                        samples_src = new_verts
                        if render_criterion and j+1 >= render_criterion_start_iter:
                            recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
                    else:
                        recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
                        samples_src = pytorch3d.ops.sample_points_from_meshes(recon_meshes, n_sample_pts) # [n_g,n_samples_3]
                    # compute a score with chamfer distance and resample using the score
                    chamfer_dists, chamfer_per_point = iou_chamfer_loss(samples_trg.points_padded(), samples_src, iou=False, chamfer=True,
                                                                        reduce_batch=False, reduce_points=False) # [n_grids]
                    chamfer_dists_meanstd = chamfer_dists.detach() + torch.std(chamfer_per_point, dim=-1).detach()
                    if render_criterion and j+1 >= render_criterion_start_iter:
                        rendered_depth_resample = rasterizer.rasterizer(recon_meshes).zbuf.squeeze(-1) # [n_g,h,w]
                        rendered_depth_resample = torch.flip(rendered_depth_resample, (-1,-2)) # to normal image coord convention
                        rendered_mask_resample = torch.ge(rendered_depth_resample, 0)
                        rendered_depth_resample = rendered_depth_resample * rendered_mask_resample + (1.0 - rendered_mask_resample.to(torch.float32)) * render_bgval
                        render_score = torch.square(rendered_depth_resample - masked_depth_batch_torch[b]).mean((-1,-2)) # [n_g]
                        render_score = render_criterion_wt * render_score
                        chamfer_dists_meanstd += render_score.detach() # [n_g]
                    if symmetry_check and j+1 >= symm_check_start_iter:
                        inv_est_rots = est_rots.transpose(-1,-2)
                        inv_est_trsl = - torch.matmul(inv_est_rots, est_trsl)
                        inv_est_scls = 1.0 / est_scls
                        tot_symm_score = 0.0
                        if rot_symm_on:
                            rot_symm_score = 0.0
                            for symm_test_rot in symm_test_rots:
                                symm_test_rot_batch = symm_test_rot.unsqueeze(0).expand(n_grids, -1, -1) # [n_g,3,3] rotations around axis of symm
                                symm_rots, symm_trsl, symm_scls = compose_poses(symm_test_rot_batch, inv_est_rots, torch.zeros_like(inv_est_trsl), inv_est_trsl,
                                                                                torch.ones_like(inv_est_scls), inv_est_scls)
                                symm_rots, symm_trsl, symm_scls = compose_poses(est_rots, symm_rots, est_trsl, symm_trsl, est_scls, symm_scls) # symm rotation in world fr
                                symm_rots = symm_rots.detach()
                                symm_trsl = symm_trsl.detach()
                                symm_scls = symm_scls.detach()
                                symm_samples_trg = apply_trsfms_to_pts(
                                                    samples_trg.points_padded(), # [WARN] if samples_trg not same len, padded zeros will be translated, here same len ok
                                                    symm_rots, symm_trsl, symm_scls) # [n_g,n_p,3]
                                symm_chamfer_dists, symm_chamfer_per_point = iou_chamfer_loss(symm_samples_trg, samples_src,
                                                                                iou=False, chamfer=True, reduce_batch=False, reduce_points=False) # [n_grids]
                                symm_chamfer_dists_meanstd = symm_chamfer_dists.detach() + torch.std(symm_chamfer_per_point, dim=-1).detach() # [n_g]
                                rot_symm_score += symm_chamfer_dists_meanstd
                                if symmetry_per_point:
                                    symm_per_point_diff = torch.abs(chamfer_per_point - symm_chamfer_per_point).mean(-1) # [n_g]
                                    rot_symm_score += symm_per_point_wt * symm_per_point_diff
                            rot_symm_score /= n_rot_symm # take average o.w. gets bigger with larger n_rot_symm
                            tot_symm_score += rot_symm_score
                            del symm_samples_trg
                        if mirror_symm_on:
                            mirr_test_rot_batch = mirr_test_rot.unsqueeze(0).expand(n_grids, -1, -1) # [n_g,3,3] rotations around axis of symm
                            symm_rots, symm_trsl, symm_scls = compose_poses(mirr_test_rot_batch, inv_est_rots, torch.zeros_like(inv_est_trsl), inv_est_trsl,
                                                                            torch.ones_like(inv_est_scls), inv_est_scls)
                            symm_rots, symm_trsl, symm_scls = compose_poses(est_rots, symm_rots, est_trsl, symm_trsl, est_scls, symm_scls) # symm rotation in world fr
                            symm_rots = symm_rots.detach()
                            symm_trsl = symm_trsl.detach()
                            symm_scls = symm_scls.detach()
                            symm_samples_trg = apply_trsfms_to_pts(
                                                samples_trg.points_padded(), # [WARN] if samples_trg not same len, padded zeros will be translated, here same len ok
                                                symm_rots, symm_trsl, symm_scls) # [n_g,n_p,3]
                            symm_chamfer_dists, symm_chamfer_per_point = iou_chamfer_loss(symm_samples_trg, samples_src,
                                                                            iou=False, chamfer=True, reduce_batch=False, reduce_points=False) # [n_grids]
                            mirr_symm_score = symm_chamfer_dists.detach() + torch.std(symm_chamfer_per_point, dim=-1).detach() # [n_g]
                            tot_symm_score += mirr_symm_score
                            if symmetry_per_point:
                                mirr_symm_per_point_diff = torch.abs(chamfer_per_point - symm_chamfer_per_point).mean(-1) # [n_g]
                                tot_symm_score += symm_per_point_wt * mirr_symm_per_point_diff
                            del symm_samples_trg
                        if mirror_symm_on and rot_symm_on:
                            tot_symm_score /= 2.0 # average both symm scores
                        chamfer_dists_meanstd += symmetry_check_wt * tot_symm_score
                    n_grids = resample_n[resample_count-1]
                    if resample_method == 0: # resample with softmax prob
                        # softmax_prob = 1.0 - torch.softmax(chamfer_dists_meanstd, dim=0) # small distance means better so 1.0 minus causing sum not to 1
                        softmax_prob = torch.softmax(1.0 / (chamfer_dists_meanstd + 1e-10), dim=0) # inverse first so smaller distance becomes larger so larger prob.
                        selected = torch.multinomial(softmax_prob, n_grids, replacement=False)
                    if resample_method == 1 or resample_method == 2:
                        selected = torch.argsort(chamfer_dists_meanstd, dim=0, descending=False) # low to high, good to bad
                        if resample_method == 1:
                            selected = selected[:n_grids]
                        if resample_method == 2:
                            dummy_trsl = np.zeros((3,1)).astype(np.float32)
                            selected_list = list(selected.detach().cpu().numpy())
                            est_rots_np = est_rots.detach().cpu().numpy()
                            idx1 = selected_list.pop(0)
                            spaced_selected = [idx1]
                            rot1 = np.concatenate([est_rots_np[idx1,:,:], dummy_trsl], axis=-1) # [3,4]
                            rot1 = np.concatenate([rot1, [[0,0,0,1]]], axis=0) # [4,4]
                            while (n_grids - len(spaced_selected)) < len(selected_list) and len(spaced_selected) < n_grids:
                                idx2 = selected_list.pop(0)
                                rot2 = np.concatenate([est_rots_np[idx2,:,:], dummy_trsl], axis=-1) # [3,4]
                                rot2 = np.concatenate([rot2, [[0,0,0,1]]], axis=0) # [4,4]
                                theta_diff = third_party.nocs_utils.compute_RT_degree_cm_symmetry(rot2, rot1, 0, False, [syn_cls])[0] # [0] to get theta
                                if theta_diff >= theta_spacing: # this only checks 1d rotation deg diff but there can be 3d difference
                                    spaced_selected.append(idx2)
                                    rot1 = rot2
                                    idx1 = idx2
                            if len(spaced_selected) < n_grids:
                                spaced_selected += selected_list[:n_grids-len(spaced_selected)]
                            selected = torch.tensor(spaced_selected, device=device)
                    # pick the selected ones
                    est_rots = torch.index_select(est_rots, 0, selected).clone() # [n_g,3,3]
                    est_trsl = torch.index_select(est_trsl, 0, selected).clone() # [n_g,3,1]
                    est_scls = torch.index_select(est_scls, 0, selected).clone() # [n_g]
                    samples_trg = samples_trg[:n_grids]
#                     trg_points_ctr_orig = trg_points_ctr_orig[:n_grids]
#                     trg_points_bboxdiag_orig = trg_points_bboxdiag_orig[:n_grids]
#                     trg_points_normal_ctr_orig = trg_points_normal_ctr_orig[:n_grids]
                    if not given_shape:
                        batch_faces = uni_faces.expand(n_grids, -1, -1)
                    elif use_gt_shape:
                        batch_faces = gt_shape_faces[gt_mesh_all[syn_cls][i+b]].clone().expand(n_grids, -1, -1)
                        batch_given_verts = gt_shape_verts[gt_mesh_all[syn_cls][i+b]].clone().expand(n_grids, -1, -1)
                    elif use_closest_shape:
                        batch_faces = closest_shape_faces[shape_cls_results[repackedd2clsimg[b]]].clone().expand(n_grids, -1, -1)
                        batch_given_verts = closest_shape_verts[shape_cls_results[repackedd2clsimg[b]]].clone().expand(n_grids, -1, -1)
                    pca_est_params = torch.index_select(pca_est_params, 0, selected).detach().clone().requires_grad_() # [n_g,K]
                    if em:
                        sigma_sq_inv = torch.index_select(sigma_sq_inv, 0, selected).clone()
#                         sigma_sq_inv_shape = sigma_sq_inv * torch.square(trg_points_bboxdiag_orig.squeeze(-1)) # [n_g,n]
                    # shape optimization set-up
                    if shape_mode == 0:
                        optimizer_shape = torch.optim.SGD([pca_est_params], lr=lr_shape_m0, momentum=mm_shape_m0) # SGD object not subscritable cannot pick
                    if shape_mode == 1:
                        optim_scls = torch.ones_like(est_scls, dtype=torch.float32, device=device, requires_grad=True) # [bs]
                        optim_trsl = torch.zeros_like(est_trsl, dtype=torch.float32, device=device, requires_grad=True) # [bs,3,1]
                        optim_rots_so3 = torch.zeros_like(est_trsl.squeeze(-1), dtype=torch.float32, device=device, requires_grad=True) # [bs,3]
                        optimizer_shape = torch.optim.SGD([pca_est_params, optim_scls, optim_trsl, optim_rots_so3], lr=lr_shape_m1, momentum=mm_shape_m1)
                    # rendering camera set-up
                    if wt_render > 0 or render_criterion:
                        cameras_setup = setCameras(image_shape, n_grids, device, True, intrinsics)
                        cameras_setup.zero_camera_pose() # cam is world pose_c_w = identity
                        rasterizer.rasterizer.cameras = cameras_setup.cameras
                    if render_criterion and j+1 >= render_criterion_start_iter:
                        del rendered_depth_resample
                # create meshes from bases and transform to estimated pose and scale
                if linear_shape and not given_shape:
                    new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # [n_g,n_verts*3] dim broadcastable
                elif not linear_shape and not given_shape:
                    new_verts_0 = vae_model.decoder(pca_est_params.detach())
                else:
                    new_verts_0 = batch_given_verts
                new_verts_0 = torch.reshape(new_verts_0, (n_grids, -1, 3)) # [n_g,n_verts,3]
                if normalize_for_pose or center_for_pose:
                    new_verts_0, _, _ = normalize_center_vertices(new_verts_0, normalize_for_pose, center_for_pose, bbox_center=True) # [n_g,n,3], [n_g,1,1], [n_g,1,3]
                if mode1 == 0 or mode1 == 1:
                    new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
                    if verts_as_sppts:
                        samples_src = new_verts
                    else:
                        recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
                        # sample on the source mesh - inefficient
                        samples_src = pytorch3d.ops.sample_points_from_meshes(recon_meshes, n_sample_pts) # [n_g,n_samples,3]
                # pose estimation
                sim_trsfm = None
                
#                 print('trg', samples_trg.points_padded().shape) # TESTING
#                 print('src', samples_src)
#                 recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
#                 for errb in range(samples_trg.points_padded().shape[0]):
#                     pytorch3d.io.IO().save_pointcloud(samples_trg[errb], './nocs_umeyama_debug/error_trg0_{:03d}{:03d}.ply'.format(j,errb))
#                     pytorch3d.io.IO().save_mesh(recon_meshes[errb], './nocs_umeyama_debug/error_src0_{:03d}{:03d}.ply'.format(j,errb))
                
                # symmetry completion
                if symm_cplt and (j+1) >= symm_cplt_iter_range[0] and (j+1) <= symm_cplt_iter_range[1]:
                    inv_est_rots = est_rots.transpose(-1,-2)
                    inv_est_trsl = - torch.matmul(inv_est_rots, est_trsl)
                    inv_est_scls = 1.0 / est_scls
                    cplt_samples_trg = samples_trg.points_padded() # [n_g,n_p,3]
                    if rot_symm_on:
                        for symm_cplt_rot in symm_cplt_rots:
                            symm_cplt_rot_batch = symm_cplt_rot.unsqueeze(0).expand(n_grids, -1, -1) # [n_g,3,3] rotations around axis of symm
                            symm_rots, symm_trsl, symm_scls = compose_poses(symm_cplt_rot_batch, inv_est_rots, torch.zeros_like(inv_est_trsl), inv_est_trsl,
                                                                            torch.ones_like(inv_est_scls), inv_est_scls)
                            symm_rots, symm_trsl, symm_scls = compose_poses(est_rots, symm_rots, est_trsl, symm_trsl, est_scls, symm_scls) # symm rotation in world fr
                            symmrotcp_samples_trg_i = apply_trsfms_to_pts(
                                             samples_trg.points_padded(), # [WARN] if samples_trg not same len, padded zeros will be translated, here same len ok
                                             symm_rots, symm_trsl, symm_scls) # [n_g,n_p,3]
                            cplt_samples_trg = torch.cat([cplt_samples_trg, symmrotcp_samples_trg_i], dim=1) # [n_g,n_p*x,3]
                    if mirror_symm_on:
                        mirr_cplt_rot_batch = mirr_test_rot.unsqueeze(0).expand(n_grids, -1, -1) # [n_g,3,3] rotations around axis of symm
                        symm_rots, symm_trsl, symm_scls = compose_poses(mirr_cplt_rot_batch, inv_est_rots, torch.zeros_like(inv_est_trsl), inv_est_trsl,
                                                                        torch.ones_like(inv_est_scls), inv_est_scls)
                        symm_rots, symm_trsl, symm_scls = compose_poses(est_rots, symm_rots, est_trsl, symm_trsl, est_scls, symm_scls) # symm rotation in world fr
                        mirrcp_samples_trg = apply_trsfms_to_pts(
                                            samples_trg.points_padded(), # [WARN] if samples_trg not same len, padded zeros will be translated, here same len ok
                                            symm_rots, symm_trsl, symm_scls) # [n_g,n_p,3]
                        cplt_samples_trg = torch.cat([cplt_samples_trg, mirrcp_samples_trg], dim=1) # [n_g,n_p*x+1,3]
                    data_samples_trg = pytorch3d.structures.Pointclouds(cplt_samples_trg)
                    # reset em inv sigma to right size
                    if em:
                        n_points_multiplier = 1
                        if rot_symm_on:
                            n_points_multiplier = n_points_multiplier + len(symm_cplt_rot_angles)
                        if mirror_symm_on:
                            n_points_multiplier += 1
                        sigma_shape = list(samples_trg.points_padded()[:,:,0].shape)
                        sigma_shape[1] = sigma_shape[1] * n_points_multiplier
                        sigma_sq_inv = sigma_sq_inv_init * torch.ones(sigma_shape, device=device, dtype=torch.float32) # [n_g,n] sigma_sq_inv_shape updated later
                else:
                    data_samples_trg = samples_trg
                    # reset em inv sigma to right size
                    if em and symm_cplt: # symm_cplt == true but out of the iteration range for completion
                        sigma_sq_inv = sigma_sq_inv_init * torch.ones(samples_trg.points_padded()[:,:,0].shape, device=device, dtype=torch.float32) # [n_g,n]
#                         sigma_sq_inv_shape = sigma_sq_inv * torch.square(trg_points_bboxdiag_orig.squeeze(-1)) # [n_g,n]
                # knn_umeyama or icp
                if mode1 == 0:
                    try:
                        for aln_i in range(mode1_iters): # [DONE] Possibly wrong. Causing cuda runtime error
                            if not em:
                                sim_trsfm = knn_umeyama_hetero(data_samples_trg, samples_src, est_scale=True) # [DONE] update to EM
                            else:
                                sim_trsfm, sigma_sq_inv = knn_umeyama_hetero_em(data_samples_trg, samples_src, est_scale=True,
                                                                             update_sigma=(j > 0 or aln_i > 0) and not fix_sigma, knn_k=knn_k,
                                                                             sigma_sq_inv=sigma_sq_inv, bs=n_grids, deg=3)
                            if mode1_iters > 1:
                                est_scls = est_scls * sim_trsfm[2].detach() # [bs] accumuate increments
                                est_rots = sim_trsfm[0].detach().transpose(-1,-2).matmul(est_rots) # [bs,3,3]
                                est_trsl = sim_trsfm[0].detach().transpose(-1,-2).matmul( # [bs,3,3]
                                   (sim_trsfm[2].detach().unsqueeze(-1).unsqueeze(-1) * est_trsl)) + sim_trsfm[1].detach().unsqueeze(-1) # [bs,3,1]
                                samples_src = apply_trsfms_to_pts(samples_src, sim_trsfm[0].detach().transpose(-1,-2),
                                                                  sim_trsfm[1].detach().unsqueeze(-1), sim_trsfm[2].detach()) # transform with the increments
                                sim_trsfm = None
                    except:
                        pose_err_msg = syn_cls + ' batch {:07d}'.format(i) + ' batch instance {:03d}'.format(b) + ' iteration {:03d}'.format(j) +\
                                      'imid {:010d}'.format(image_id_all[syn_cls][i+b]) + 'instid {:03d}\n'.format(inst_id_all[syn_cls][i+b])
                        pose_err_msg_impath = dataset.image_info[image_id_all[syn_cls][i+b]]["path"] + '\n'
                        with open(os.path.join(save_dir, 'pose_errors.txt'), 'a') as f:
                            f.write(pose_err_msg)
                            f.write(pose_err_msg_impath)
                        sim_trsfm = None
                if mode1 == 1:
                    sim_trsfm = icp(data_samples_trg, samples_src, est_scale=True)
                # accumulate increments on estimates if mode1 is run
                if sim_trsfm:
                    est_scls = est_scls * sim_trsfm[2].detach() # [bs]
                    est_rots = sim_trsfm[0].detach().transpose(-1,-2).matmul(est_rots) # [bs,3,3]
                    est_trsl = sim_trsfm[0].detach().transpose(-1,-2).matmul( # [bs,3,3]
                       (sim_trsfm[2].detach().unsqueeze(-1).unsqueeze(-1) * est_trsl)) + sim_trsfm[1].detach().unsqueeze(-1) # [bs,3,1]
                    samples_src = apply_trsfms_to_pts(samples_src, sim_trsfm[0].detach().transpose(-1,-2),
                                                      sim_trsfm[1].detach().unsqueeze(-1), sim_trsfm[2].detach()) # transform with the increments (no use copied from above)
                    sim_trsfm = None
                if mode1 == 0 or mode1 == 1 or resample:
                    del samples_src
                # shape estimation
                if j+1 > shape_early_stop_iter or given_shape or skip_shape_optim:
                    continue
                if shape_mode == 1: # zero out the optimization variables for incremental update and accumulate updates on the estimates
                    optim_scls.data = torch.ones_like(est_scls, dtype=torch.float32, device=device)
                    optim_trsl.data = torch.zeros_like(est_trsl, dtype=torch.float32, device=device)
                    optim_rots_so3.data = torch.zeros_like(est_trsl.squeeze(-1), dtype=torch.float32, device=device) # requires_grad here should not matter
#                 try:
                # center and normalize the trg and apply the same to src for shape cplt need to perform this every est. iteration
#                 if symm_cplt and (j+1) >= symm_cplt_iter_range[0] and (j+1) <= symm_cplt_iter_range[1]:
#                     trg_points_ctr = torch.mean(data_samples_trg.points_padded(), dim=-2, keepdim=True) # [n_g,1,3] [DONE] use bbox center instead of centroid. may not need since here is just rough normalization not to gaurantee source model bbox center is at origin
#                     trg_points_maxcorner = torch.max(data_samples_trg.points_padded(), dim=-2, keepdim=True).values # [n_g,1,3]
#                     trg_points_mincorner = torch.min(data_samples_trg.points_padded(), dim=-2, keepdim=True).values # [n_g,1,3]
#                     trg_points_bboxdiag = trg_points_maxcorner - trg_points_mincorner
#                     trg_points_bboxdiag = torch.norm(trg_points_bboxdiag, dim=-1, keepdim=True) # [n_g,1,1]
#                     # trg_points_ctr = (trg_points_maxcorner + trg_points_mincorner) / 2.0 # [n_g,1,3] bbx ctr
#                     trg_points_normal_ctr = (data_samples_trg.points_padded() - trg_points_ctr) / trg_points_bboxdiag
#                     trg_points_normal_ctr = pytorch3d.structures.Pointclouds(trg_points_normal_ctr) # ptc same len but for mesh_deform_loss compatibility use Pointclouds
#                     if em:
#                         sigma_sq_inv_shape = sigma_sq_inv * torch.square(trg_points_bboxdiag.squeeze(-1)) # [n_g,n]
#                 else:
#                     trg_points_ctr = trg_points_ctr_orig
#                     trg_points_bboxdiag = trg_points_bboxdiag_orig
#                     trg_points_normal_ctr = trg_points_normal_ctr_orig.to(device)
                for k_sh in range(n_shape_iters):
                    optimizer_shape.zero_grad() # should zero out all the stored gradients
                    shape_start_time = timeit.default_timer()
                    if linear_shape:
                        new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # [n_g,n_verts*3] dim broadcastable
                    else:
                        new_verts_0 = vae_model.decoder(pca_est_params)
                    new_verts_0 = torch.reshape(new_verts_0, (n_grids, -1, 3)) # [n_g,n_verts,3]
                    shape_time[syn_cls][0] += (timeit.default_timer() - shape_start_time)
                    shape_start_time = timeit.default_timer()
                    # original # TESTING
#                         if shape_mode == 0:
#                             new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
                    # method 1 - use the estimated scale and translation to normalize - made it not optimize shape anymore
#                         if shape_mode == 0:
#                             new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, torch.zeros_like(est_trsl), torch.ones_like(est_scls))
#                         trg_points_normal_ctr = (data_samples_trg.points_padded() - torch.transpose(est_trsl,-1,-2)) / est_scls.unsqueeze(-1).unsqueeze(-1)
#                         trg_points_normal_ctr = pytorch3d.structures.Pointclouds(trg_points_normal_ctr) # same len
                    # method 2 - just center and normalize the trg and apply the same to src [DONE] optimize speed by computing these only once -> moved out of shape loop for shape cplt and out of est. loop for no cplt
#                     trg_points_ctr = torch.mean(data_samples_trg.points_padded(), dim=-2, keepdim=True) # [n_g,1,3] [DONE] use bbox center instead of centroid. may not need since here is just rough normalization not to gaurantee source model bbox center is at origin
                    trg_points_maxcorner = torch.max(data_samples_trg.points_padded(), dim=-2, keepdim=True).values # [n_g,1,3]
                    trg_points_mincorner = torch.min(data_samples_trg.points_padded(), dim=-2, keepdim=True).values # [n_g,1,3]
                    trg_points_bboxdiag = trg_points_maxcorner - trg_points_mincorner
                    trg_points_bboxdiag = torch.norm(trg_points_bboxdiag, dim=-1, keepdim=True) # [n_g,1,1]
                    trg_points_ctr = (trg_points_maxcorner + trg_points_mincorner) / 2.0 # [n_g,1,3] bbx ctr
                    trg_points_normal_ctr = (data_samples_trg.points_padded() - trg_points_ctr) / trg_points_bboxdiag
                    trg_points_normal_ctr = pytorch3d.structures.Pointclouds(trg_points_normal_ctr) # ptc same len but for mesh_deform_loss compatibility use Pointclouds
                    if shape_mode == 0:
                        new_verts_src = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls)
                        new_verts = (new_verts_src - trg_points_ctr) / trg_points_bboxdiag
                    if shape_mode == 1:
                        optim_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3)
                        # method 2 - corresponding to above
                        new_verts_src = apply_trsfms_to_pts(new_verts_0, optim_rots.matmul(est_rots), optim_trsl+est_trsl, optim_scls*est_scls) # incremental pose from current estimate
                        new_verts = (new_verts_src - trg_points_ctr) / trg_points_bboxdiag
                    recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
                    shape_time[syn_cls][1] += (timeit.default_timer() - shape_start_time)
                    shape_start_time = timeit.default_timer()
                    if em: # moved out of shape loop for efficiency just like method 2 above
                        sigma_sq_inv_shape = sigma_sq_inv * torch.square(trg_points_bboxdiag.squeeze(-1)) # [n_g,n]
                    else:
                        sigma_sq_inv_shape = None
                    shape_time[syn_cls][2] += (timeit.default_timer() - shape_start_time)
                    shape_start_time = timeit.default_timer()
#                         for errb in range(data_samples_trg.points_padded().shape[0]): # TESTING
#                             pytorch3d.io.IO().save_pointcloud(data_samples_trg[errb], './nocs_umeyama_debug/error_trg1_{:03d}{:03d}.ply'.format(j,errb))
#                             pytorch3d.io.IO().save_pointcloud(trg_points_normal_ctr[errb], './nocs_umeyama_debug/error_trg1_{:03d}{:03d}.ply'.format(j,errb))
#                             pytorch3d.io.IO().save_mesh(recon_meshes[errb], './nocs_umeyama_debug/error_src1_{:03d}{:03d}.ply'.format(j,errb))
                    [loss_chamfer, loss_edge, loss_normal, loss_laplacian, loss_ptedge, loss_ptface, shape_loss_times_names] \
                        = mesh_deform_loss(recon_meshes, trg_points_normal_ctr, n_samples_trg=n_sample_pts, ############### TESTING
                                           use_temp_verts=verts_as_sppts, n_samples_temp=n_sample_pts,
                                           chamfer_on=wt_chamfer, edge_on=wt_edge, normal_on=wt_normal,
                                           laplacian_on=wt_laplacian, ptedge_on=wt_ptedge, ptface_on=wt_ptface,
                                           given_trg_samples=True, double_chamfer=False, em_weights=em, knn_k=knn_k,
                                           sigma_sq_inv=sigma_sq_inv_shape, rec_runtime=True) # [DONE] update to EM
                    shape_time[syn_cls][11] += (timeit.default_timer() - shape_start_time)
                    shape_time[syn_cls][3:11] += shape_loss_times_names[0]
                    shape_start_time = timeit.default_timer()
                    # all losses avg. across batch. multiplied by n_grids 20240709 to avoid downscaling with smaller n_grids. removed 20240716
                    total_loss = (wt_chamfer * loss_chamfer + wt_edge * loss_edge + \
                             wt_normal * loss_normal + wt_laplacian * loss_laplacian + \
                             wt_ptedge * loss_ptedge + wt_ptface * loss_ptface) # removed multiplying by bs_actual 20231228
                    shape_time[syn_cls][12] += (timeit.default_timer() - shape_start_time)
                    shape_start_time = timeit.default_timer()
                    if wt_modsink > 0:
                        loss_modsink = torch.sum(torch.min(torch.mean(torch.abs( # changed 20240709, torch.sum->torch.mean for inner -1 dim o.w. loss scales with K
                            pca_est_params.unsqueeze(-2) - pca_basis_params_torch.unsqueeze(0)), dim=-1), dim=-1)[0]) # [n_g,K]->[n_g,1,K] [n_obj,K]->[1,n_obj,K]
                        if n_iters >= shape_early_stop_iter:
                            n_iters_for_shape = shape_early_stop_iter
                        else:
                            n_iters_for_shape = n_iters
                        if wt_modsink_schd == 0:
                            wt_modsink_ac = wt_modsink
                        if wt_modsink_schd == 1:
                            wt_modsink_ac = wt_modsink / (n_iters_for_shape * n_shape_iters) * (n_shape_iters * j + k_sh + 1)
                        if wt_modsink_schd == 2:
                            if n_shape_iters * j + k_sh + 1 <= n_iters_for_shape * n_shape_iters * wt_modsink_thpct:
                                wt_modsink_ac = 0
                            else:
                                wt_modsink_th = np.floor(n_iters_for_shape * n_shape_iters * wt_modsink_thpct)
                                wt_modsink_ac = wt_modsink / (n_iters_for_shape * n_shape_iters - wt_modsink_th) *\
                                                             (n_shape_iters * j + k_sh + 1 - wt_modsink_th)
                        if wt_modsink_schd == 3:
                            wt_modsink_ac = wt_modsink / (n_iters_for_shape * n_shape_iters)**2 * (n_shape_iters * j + k_sh + 1)**2
                        total_loss += wt_modsink_ac * loss_modsink
                    shape_time[syn_cls][13] += (timeit.default_timer() - shape_start_time)
                    shape_start_time = timeit.default_timer()
                    if wt_render > 0 and j+1 >= wt_render_on_iter:
                        recon_meshes_src = pytorch3d.structures.Meshes(verts=new_verts_src, faces=batch_faces)
                        rendered_depth = rasterizer.rasterizer(recon_meshes_src).zbuf.squeeze(-1) # [n_g,h,w]
                        rendered_depth = torch.flip(rendered_depth, (-1,-2)) # to be consistent with normal image coords
                        rendered_mask = torch.ge(rendered_depth, 0)
                        rendered_depth = rendered_depth * rendered_mask + (1.0 - rendered_mask.to(torch.float32)) * render_bgval
                        # debugging ##########################
                        # plt.figure(figsize=(7,7))
                        # depth_img_rendered = rendered_depth.detach()
                        # plt.imshow(depth_img_rendered.cpu().numpy()[0])
                        # plt.savefig(os.path.join(viz_dir, 'iter_{:d}_rend.png'.format(j + 1)))
                        # plt.figure(figsize=(7,7))
                        # plt.imshow(masked_depth_batch_torch[b].cpu().numpy())
                        # plt.savefig(os.path.join(viz_dir, 'iter_{:d}_data.png'.format(j + 1)))
                        # debugging ##########################
                        loss_render = torch.abs(rendered_depth - masked_depth_batch_torch[b]).mean((-1,-2)).mean(0) # mean across pixel and sum over batch (n_grids) (mean across batch before 20240709 and after 20240716)
                        total_loss += wt_render * loss_render
                    shape_time[syn_cls][14] += (timeit.default_timer() - shape_start_time)
                    shape_start_time = timeit.default_timer()
                    total_loss.backward()
                    shape_time[syn_cls][15] += (timeit.default_timer() - shape_start_time)
                    shape_start_time = timeit.default_timer()
                    optimizer_shape.step()
                    shape_time[syn_cls][16] += (timeit.default_timer() - shape_start_time)
                    shape_start_time = timeit.default_timer()
#                 except:
#                     shape_err_msg = syn_cls + ' batch {:07d}'.format(i) + ' batch instance {:03d}'.format(b) + ' iteration {:03d}'.format(j) +\
#                                   ' imid {:010d}'.format(image_id_all[syn_cls][i+b]) + ' instid {:03d}\n'.format(inst_id_all[syn_cls][i+b])
#                     shape_err_msg_impath = dataset.image_info[image_id_all[syn_cls][i+b]]["path"] + '\n'
#                     with open(os.path.join(save_dir, 'shape_errors.txt'), 'a') as f:
#                         f.write(shape_err_msg)
#                         f.write(shape_err_msg_impath)
                if shape_mode == 1:
                    est_scls *= optim_scls.detach()
                    est_trsl += optim_trsl.detach()
                    est_rots = pytorch3d.transforms.so3_exp_map(optim_rots_so3.detach()).matmul(est_rots)
                if wt_render > 0 and j+1 >= wt_render_on_iter and n_shape_iters >= 1:
                    del rendered_depth
#                 del trg_points_normal_ctr
            # end of optimization loop -- n_grids = bs = 1 should reduce to one
            assert(n_grids == 1)
            total_time[syn_cls] += (timeit.default_timer() - start)
            # record shape (no pose transformation) results
            if linear_shape and not given_shape:
                new_verts_0 = torch.matmul(pca_est_params, pca_bases) + pca_mean # dim broadcastable
            elif not linear_shape and not given_shape:
                new_verts_0 = vae_model.decoder(pca_est_params.detach())
            else:
                new_verts_0 = batch_given_verts
            new_verts_0 = torch.reshape(new_verts_0, (n_grids, -1, 3)) # n_grids = 1
            est_rots_v2 = est_rots.clone() # for viz results save original rots
            if normalize_for_eval or center_for_eval: # need normalize first to make sure it rotates around bbox center. I think does not matter.
                new_verts_0, norm_factor, centroid = normalize_center_vertices(new_verts_0, normalize_for_eval, center_for_eval, bbox_center=True) # [bs,n,3], [bs,1,1], [bs,1,3]
            if center_for_eval:
                est_trsl += est_rots.matmul((est_scls.unsqueeze(-1) * centroid.squeeze(-2)).unsqueeze(-1)) # [bs,3,1] t' = t + Rsc
            if normalize_for_eval:
                est_scls *= norm_factor.squeeze() # [bs] s' = s*norm
            if convert_to_snv1_rot:
                new_verts_0 = torch.matmul(new_verts_0, rot_v2_to_v1.transpose(-1,-2)) # [n_g,-1,3] should be pre-multiply but doing post-multiply so transpose
                est_rots = torch.matmul(est_rots, rot_v1_to_v2) # [1,3,3]
            recon_meshes_0 = pytorch3d.structures.Meshes(verts=new_verts_0, faces=batch_faces)
            shape_pred[syn_cls].append(recon_meshes_0.detach().cpu())
            # transformed shape results for viz (put here so the trsl and scls convention changes do not affect)
            new_verts = apply_trsfms_to_pts(new_verts_0, est_rots, est_trsl, est_scls) # estimated pose already transformed to current convention
            recon_meshes = pytorch3d.structures.Meshes(verts=new_verts, faces=batch_faces)
            # record pose scale results
            if nocs_1000factor:
                est_scls /= 1000.0
                est_trsl /= 1000.0
            xyz_maxmin_diff = torch.max(new_verts_0, dim=-2).values - torch.min(new_verts_0, dim=-2).values # [n_g,3] diagonal line
            xyz_scls = xyz_maxmin_diff * est_scls.unsqueeze(-1) # [n_g,3] 3d bbox x, y, z scales (diagonally unnormalized bbox edge lengths)
            pose_scl_pred[syn_cls].append((est_rots.detach().cpu().numpy(), # [1,3,3]
                                           est_trsl.detach().cpu().numpy(), # [1,3,1]
                                           est_scls.detach().cpu().numpy(), # [1] 
                                           xyz_scls.detach().cpu().numpy())) # [1,3] for bbox_xyz_scales. the batch dim 1 is good for concat later
            # save some results for viz (inside batch loop)
            if (i_for_save % save_freq == 0 and b == 0) or fr_results_save[syn_cls][i+b]:
                if fr_results_save[syn_cls][i+b]:
                    fr_path = dataset.image_info[image_id_all[syn_cls][i+b]]["path"]
                    fr_info = fr_path.split('/')[-2] + '_' + fr_path.split('/')[-1]
                    fr_save_dir = os.path.join(viz_fr_dir, fr_info)
                    if not os.path.exists(fr_save_dir):
                        os.mkdir(fr_save_dir)
                    viz_save_dir = fr_save_dir
                else:
                    viz_save_dir = viz_dir
                # these should not be indexed by b because they were constructed inside batch loop. batch size is n_grids and they are the same copies across n_grids.
                pytorch3d.io.IO().save_mesh(recon_meshes[0], os.path.join(viz_save_dir, syn_cls + '_iter{:d}_b{:d}_src.ply'.format(i + 1, b)))
                pytorch3d.io.IO().save_mesh(recon_meshes_0[0], os.path.join(viz_save_dir, syn_cls + '_iter{:d}_b{:d}_src_shape.ply'.format(i + 1, b)))
                posed_verts_init = apply_trsfms_to_pts(verts_init[0:1], est_rots_v2, init_trsl[0:1], init_scls[0:1])
                init_posed_meshes = pytorch3d.structures.Meshes(verts=posed_verts_init, faces=batch_faces)
                pytorch3d.io.IO().save_mesh(init_posed_meshes[0], os.path.join(viz_save_dir, syn_cls + '_iter{:d}_b{:d}_initscltrslshape_estrot.ply'.format(i + 1, b)))
                if normalize_for_eval or center_for_eval:
                    converted_verts_init, _, _ = normalize_center_vertices(verts_init[0:1], normalize_for_eval, center_for_eval, bbox_center=True)
                else:
                    converted_verts_init = verts_init[0:1]
                if convert_to_snv1_rot:
                    converted_verts_init = torch.matmul(converted_verts_init, rot_v2_to_v1.transpose(-1,-2)) # [DONE] better to also normalize
                init_meshes = pytorch3d.structures.Meshes(verts=converted_verts_init, faces=batch_faces)
                pytorch3d.io.IO().save_mesh(init_meshes[0], os.path.join(viz_save_dir, syn_cls + '_iter{:d}_b{:d}_src_initshape.ply'.format(i + 1, b)))
                pytorch3d.io.IO().save_pointcloud(samples_trg[0], os.path.join(viz_save_dir, syn_cls + '_iter{:d}_b{:d}_trgdepth.ply'.format(i + 1, b)))
                if symm_cplt:
                    cplt_samples_trg_ptc = pytorch3d.structures.Pointclouds(cplt_samples_trg[0:1])
                    pytorch3d.io.IO().save_pointcloud(cplt_samples_trg_ptc[0], os.path.join(viz_save_dir,
                                        syn_cls + '_iter{:d}_b{:d}_cplttrg_cpltlastiter{:d}_totiter{:d}_1sthypo.ply'.format(i + 1, b, symm_cplt_iter_range[1], n_iters)))
                # save image path for reference
                if data != 'demo':
                    with open(os.path.join(viz_save_dir,  syn_cls + '_iter{:d}_b{:d}_iminfo.txt'.format(i + 1, b)), 'w') as f:
                        f.write('im id: {:010d}, inst id: {:03d}\n'.format(image_id_all[syn_cls][i+b], inst_id_all[syn_cls][i+b]))
                        f.write(dataset.image_info[image_id_all[syn_cls][i+b]]["path"] + '\n')
            # release memory
            del samples_trg#, trg_points_normal_ctr_orig
            if symm_cplt:
                del data_samples_trg
                del cplt_samples_trg
        # end of batch loop
        torch.cuda.empty_cache()
        gc.collect() # does not help with memory exhaustion
    # end of data loop
# end of class loop
try: 
    shape_time_names = shape_time_names[:3] + shape_loss_times_names[1] + shape_time_names[3:]
except NameError: # if shape optim is not run, no shape_loss_times_names
    print('shape_loss_times_names not exist')
    shape_time_names = shape_time_names[:3] + ['n/a'] * (shape_time[syn_cls].shape[0] - len(shape_time_names)) + shape_time_names[3:]
```

```python
# evaluate shape
# user input
n_chamfer_pts = 5000
eval_sprior_matched_gt_argmax = True # false: matching by iou greedily over a threshold (sprior exact). true: pick the max iou greater than a threhold
shape_save_freq = save_freq * bs
if not eval_sprior_matched_gt_argmax:
    eval_shape_dir = os.path.join(save_dir, 'shape_eval')
else:
    eval_shape_dir = os.path.join(save_dir, 'shape_eval_argmax')
if not os.path.exists(eval_shape_dir):
    os.mkdir(eval_shape_dir)

# build gt shape meshes and points from vertices and faces
gt_shape_points = {}
gt_shape_meshes = {}
for obj in gt_shapes.keys():
    verts_torch = torch.tensor(gt_shapes[obj][0], dtype=torch.float32, device='cpu').unsqueeze(0) # [1,n_v,3]
    faces_torch = torch.tensor(gt_shapes[obj][1], dtype=torch.long, device='cpu').unsqueeze(0) # [1,n_f,3]
    gt_shape_mesh = pytorch3d.structures.Meshes(verts=verts_torch, faces=faces_torch)
    gt_shape_meshes[obj] = gt_shape_mesh
    gt_shape_points[obj] = pytorch3d.ops.sample_points_from_meshes(gt_shape_mesh, n_chamfer_pts)
# if use shape prior matched gt shape
if eval_sprior_matched_gt_argmax:
    assert(sprior_gtshape_match and use_mrcnn_mask)
    gt_mesh_all_eval = gt_mesh_all_alt
else:
    gt_mesh_all_eval = gt_mesh_all
# evaluate
chamfer_all = []
for syn_cls in synset_names[1:]:
    print(syn_cls)
    assert(len(shape_pred[syn_cls]) == len(gt_mesh_all_eval[syn_cls]))
    if len(shape_pred[syn_cls]) == 0:
        print(syn_cls + ' has no predicted instance!')
        continue
    shape_viz_dir = os.path.join(eval_shape_dir, 'viz_'+syn_cls)
    if not os.path.exists(shape_viz_dir):
        os.mkdir(shape_viz_dir)
    chamfer_per_cls = []
    # iterate instances in a class
    cls_loop = tqdm(range(len(shape_pred[syn_cls])))
    for s in cls_loop:
        if gt_mesh_all_eval[syn_cls][s] == -1:
            continue # no gt shape matched
        pts_gt = gt_shape_points[gt_mesh_all_eval[syn_cls][s]]
        pts_pred = pytorch3d.ops.sample_points_from_meshes(shape_pred[syn_cls][s], n_chamfer_pts)
        chamfer_metric, _ = pytorch3d.loss.chamfer_distance(pts_gt, pts_pred) # see src code, squared distance due to use of knn_points (also squared in the shape prior paper)
        chamfer_metric_cpu = chamfer_metric.cpu().numpy()
        chamfer_all.append(chamfer_metric_cpu)
        chamfer_per_cls.append(chamfer_metric_cpu)
        if (s + 1) % shape_save_freq == 0:
            pytorch3d.io.IO().save_mesh(gt_shape_meshes[gt_mesh_all_eval[syn_cls][s]], os.path.join(shape_viz_dir, 'iter_{:d}_gt.ply').format(s + 1))
            pytorch3d.io.IO().save_mesh(shape_pred[syn_cls][s], os.path.join(shape_viz_dir, 'iter_{:d}_pred.ply').format(s + 1))
            # save the reference image path
            with open(os.path.join(shape_viz_dir,  'iter_{:d}_iminfo.txt'.format(s + 1)), 'w') as f:
                f.write('im id: {:010d}, inst id: {:03d}\n'.format(image_id_all[syn_cls][s], inst_id_all[syn_cls][s]))
                f.write(dataset.image_info[image_id_all[syn_cls][s]]["path"] + '\n')
    # show evaluation results per class
    avgstd_err_msg = 'mean error: {:.10f} +- {:.10f}'.format(np.mean(chamfer_per_cls), np.std(chamfer_per_cls))
    minmax_err_msg = 'max min error: {:.10f}, {:.10f}'.format(np.max(chamfer_per_cls), np.min(chamfer_per_cls))
    print(avgstd_err_msg)
    print(minmax_err_msg)
    # save evaluation results
    with open(os.path.join(eval_shape_dir, 'chamfer_'+syn_cls+'.pickle'), 'wb') as f:
        pickle.dump(chamfer_per_cls, f)
    with open(os.path.join(eval_shape_dir, 'eval_chamfer.txt'), 'a') as f:
        f.write(syn_cls + '\n')
        f.write(avgstd_err_msg + '\n')
        f.write(minmax_err_msg + '\n')
# all classes all instances
print('all classes all instances')
avgstd_err_msg = 'mean error: {:.10f} +- {:.10f}'.format(np.mean(chamfer_all), np.std(chamfer_all))
minmax_err_msg = 'max min error: {:.10f}, {:.10f}'.format(np.max(chamfer_all), np.min(chamfer_all))
print(avgstd_err_msg)
print(minmax_err_msg)
# save evaluation results
with open(os.path.join(eval_shape_dir, 'chamfer_all.pickle'), 'wb') as f:
    pickle.dump(chamfer_all, f)
with open(os.path.join(eval_shape_dir, 'eval_chamfer.txt'), 'a') as f:
    f.write('all classes all instances\n')
    f.write(avgstd_err_msg + '\n')
    f.write(minmax_err_msg + '\n')
```

```python
import os
import glob
import torch
import pytorch3d.structures
import pytorch3d.io
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from tqdm.notebook import tqdm

# user input
data = 'real_test'
fr_results_dir = '/home/yihao/Research/ShapeBasis/Records/20240816_060105_NOCSrealShapePoseEstMultiTrackEMShapeClassInit_20231026_re20240724_013920_butNoSinkLoss/viz_fr/'
device = 'cuda'
depth_bgval = 1000 / depth_factor # set to a max depth in the original nocs data (my saved mesh fits the depth backprojected. depending on the division depth_factor on the depth set in nocs data loading)
cm = plt.get_cmap('viridis') # color map

# set up a dir to save all the rendered frames
fr_render_save_dir = os.path.join(fr_results_dir, 'rendered_frames')
if not os.path.exists(fr_render_save_dir):
    os.mkdir(fr_render_save_dir)
color_fr_save_dir = os.path.join(fr_results_dir, 'color_frames')
if not os.path.exists(color_fr_save_dir):
    os.mkdir(color_fr_save_dir)
    
# set up rendering
rasterizer_viz = rasterization(image_shape) # assume image_shape and intrinsics is set in the nocs loading part
cameras_setup = setCameras(image_shape, bs=1, dev=device, input_camK=True, cam_K=intrinsics)
cameras_setup.zero_camera_pose() # cam is world pose_c_w = identity
rasterizer_viz.rasterizer.cameras = cameras_setup.cameras

# go through all frames
frame_dirs = glob.glob(os.path.join(fr_results_dir, 'scene*'))
fr_loop = tqdm(range(len(frame_dirs)))
for fr_idx in fr_loop:
    f_dir = frame_dirs[fr_idx]
    # load estimated meshes and render
    lumped_rendered_depth = depth_bgval * torch.ones(image_shape, dtype=torch.float32, device=device)
    lumped_mask = torch.zeros(image_shape, dtype=torch.float32, device=device)
    src_meshes_paths = glob.glob(os.path.join(f_dir, '*_src.ply'))
    for src_mesh_path in src_meshes_paths:
        src_mesh = pytorch3d.io.IO().load_mesh(src_mesh_path, include_textures=False, device=device)
        rendered_src_depth = rasterizer_viz.rasterizer(src_mesh).zbuf.squeeze(0).squeeze(-1) # [h,w]
        rendered_src_depth = torch.flip(rendered_src_depth, (-1,-2))
        rendered_mask = torch.ge(rendered_src_depth, 0)
        rendered_src_depth_bg = rendered_src_depth * rendered_mask + (1.0 - rendered_mask.to(torch.float32)) * depth_bgval
        closer_depth_mask = rendered_src_depth_bg < lumped_rendered_depth
        lumped_rendered_depth = closer_depth_mask * rendered_src_depth_bg + torch.logical_not(closer_depth_mask) * lumped_rendered_depth
        lumped_mask = torch.logical_or(lumped_mask, rendered_mask)
    lumped_rendered_depth = lumped_rendered_depth.detach().cpu().numpy()
    lumped_mask = np.expand_dims(lumped_mask.detach().cpu().numpy(), -1)
    # convert the depth image to rgb
    rendered_depth_color = cm(lumped_rendered_depth) # to [0,1] (probably plt.Normalize) to color 4-channels
    rendered_depth_color = (rendered_depth_color[:, :, :3] * 255).astype(np.uint8) # [h,w,3]
    # load the rgb image
    image_path_file = glob.glob(os.path.join(f_dir, '*iminfo.txt'))[0]
    with open(image_path_file, 'r') as iminfof:
        iminfo = iminfof.readlines()
    impath = iminfo[-1].strip('\n') + '_color.png'
    frrgbimg = np.array(Image.open(impath)) # [h,w,3]
    # sythesize the rgb image with the depth rendering
    synth_img = lumped_mask * rendered_depth_color + np.logical_not(lumped_mask) * frrgbimg
    # save to both the frame subdir and the fr_render_save_dir
    synth_img = Image.fromarray(synth_img)
    synth_img_fname = f_dir.split('/')[-1] + '.png'
    synth_img.save(os.path.join(f_dir, synth_img_fname))
    synth_img.save(os.path.join(fr_render_save_dir, synth_img_fname))
    # save the color frame for convenience
    color_img = Image.fromarray(frrgbimg)
    color_img_fname = f_dir.split('/')[-1] + '_color.png'
    color_img.save(os.path.join(color_fr_save_dir, color_img_fname))
# note to get the mesh picture: render mode -> backface culling + flat, lighting on, view -> track ball off
```

```python
# test depth image to normal image code
# code from chatGPT and https://github.com/cobanov/depth2normal/blob/main/depth_to_normal_map.py
import cv2

def depth_to_normals(depth_img, toColor=False):
    gradient_x = cv2.Sobel(depth_img, cv2.CV_64F, 1, 0, ksize=5)
    gradient_y = cv2.Sobel(depth_img, cv2.CV_64F, 0, 1, ksize=5)
    normal_img = np.dstack((-gradient_x, -gradient_y, np.ones_like(depth_img)))
    # normal_img = normal_img / np.linalg.norm(normal_img, axis=-1)[:, :, np.newaxis]
    norm = np.linalg.norm(normal_img, axis=-1)[:, :, np.newaxis]
    normal_img = np.divide(normal_img, norm, out=np.zeros_like(normal_img), where=norm!=0)
    normal_img = (normal_img + 1) * 127.5
    if toColor:
        normal_img = normal_img.clip(0, 255).astype(np.uint8)
        normal_img = cv2.cvtColor(normal_img, cv2.COLOR_RGB2BGR)
    return normal_img

normal_image = depth_to_normals(depth, toColor=True)
cv2.imshow('normal map', normal_image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
# code from Ziqi
def camerasFromOpenCV(pose, K, img_size):
        """
        Build cameras from OpenCV convention camera matrices
        OpenCV camera convention: x-right-y-down-z-forward
        @param pose: [Bx4x4 tensor] object-to-camera poses
        @param K: [Bx3x3 tensor] camera intrinsics
        @param img_size: [Bx2 tensor] image sizes [[height, width]]
        @return cameras: Pytorch cameras
        """
        # OpenCV pose to Pytorch3D pose
        cv2torch3d = torch.diag(pose.new_tensor([-1, -1, 1, 1]))
        RT = torch.einsum("ik,...kl->...il", cv2torch3d, pose)
        R_torch3d = RT[..., :3, :3].transpose(-2, -1)
        t_torch3d = RT[..., :3, 3]
        # Focal length
        focal_len = K[..., [0, 1], [0, 1]]
        # Principle points
        principle_point = K[..., :2, 2]
        # Build Pytorch3D cameras
        cameras = pytorch3d.renderer.cameras.PerspectiveCameras(
            R=R_torch3d, T=t_torch3d, focal_length=focal_len,
            principal_point=principle_point, image_size=img_size,
            device=pose.device, in_ndc=False
        )
        return cameras 
```

## speed comparison of transforming ptc vs. transforming mesh vertices + sampling

```python
# CONCLUSION: transforming points directly is faster. Sampling is too slow (one order of magnitude)
import torch
import numpy as np
import timeit
import pytorch3d.transforms
import pytorch3d.structures
import pytorch3d.ops
import pytorch3d.utils

# copied function
def apply_trsfms_to_pts(pts, rot, t, scale=1.0):
    # pts: [bs,n,3] batched padded point clouds, rot: [bs,3,3], t: [bs,3,1], scale: [bs]
    # output: [bs,n,3]
    pts_new = scale.unsqueeze(-1).unsqueeze(-1) * pts # [bs,n,3] scale around center (assumed origin before translating)
    pts_new = torch.matmul(rot.unsqueeze(1), pts_new.unsqueeze(-1)) # [bs,n,3,1]
    pts_new += t.unsqueeze(1)
    return pts_new.squeeze(-1)

# user input
bs_test = 20
sphere_lv = 4
n_test_samples = 1000
dev = "cuda"

# obtain random transformation and test mesh
random_rots = pytorch3d.transforms.random_rotations(bs_test, torch.float32, dev) # [bs,3,3]
random_trsl = torch.rand((bs_test, 3, 1), dtype=torch.float32, device=dev) * 2.0 - 1 # (-1,1) [bs,3,1]
random_scls = torch.rand((bs_test), dtype=torch.float32, device=dev) * 2.0 # (0,2) [bs]
test_mesh = pytorch3d.utils.ico_sphere(sphere_lv, dev)
test_verts = test_mesh.verts_padded().expand(bs_test, -1, -1) # [bs,max(N),3]
test_faces = test_mesh.faces_padded().expand(bs_test, -1, -1)
test_mesh_samples = pytorch3d.ops.sample_points_from_meshes(test_mesh, n_test_samples).to(dev).expand(bs_test, -1, -1)
# apply the transformation to vertices and sample
tic = timeit.default_timer()
new_verts = apply_trsfms_to_pts(test_verts, random_rots, random_trsl, random_scls)
toc = timeit.default_timer()
print(toc - tic)
new_mesh = pytorch3d.structures.Meshes(verts=new_verts, faces=test_faces)
new_mesh_samples = pytorch3d.ops.sample_points_from_meshes(new_mesh, n_test_samples)
toc = timeit.default_timer()
print(toc - tic)
# apply the transformation to the sampled points directly
tic = timeit.default_timer()
new_points = apply_trsfms_to_pts(test_mesh_samples, random_rots, random_trsl, random_scls)
toc = timeit.default_timer()
print(toc - tic)
```

## Find minimum number of vertices for the object category

```python
min_n_verts = 1e5
for m in range(len(shapenet_dataset)):
    try:
        one_model = shapenet_dataset[m]
        one_mesh = pytorch3d.structures.Meshes(verts=[one_model["verts"]], faces=[one_model["faces"]])
        n_verts = one_mesh.num_verts_per_mesh()
        if n_verts < min_n_verts:
            min_n_verts = n_verts
    except:
        continue
print(min_n_verts)
```
